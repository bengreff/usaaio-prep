# k-nearest neighbors classifier in plain Python

A **k-nearest neighbors** (kNN) classifier predicts the label of a new point by looking at the $k$ training points closest to it and taking the most common label among them. Training is only storing the data. (The 2026 Round 1 paper had an open-ended problem that allowed kNN only.)

Implement the class

```python
class KNNClassifier:
    def __init__(self, k):
    def fit(self, X, y):
    def predict(self, X):
```

- `__init__(k)`: store `k` in the attribute `k`, and set the attributes `X` and `y` to `None`.
- `fit(X, y)`: `X` is a list of $n$ points, each a list of $d$ floats; `y` is a list of $n$ int labels. Store the training data in `self.X` and `self.y` as **copies**, so that later changes to the caller's lists (including the inner point lists) do not change the model. Return `self`. Each call to `fit` replaces the stored data.
- `predict(X)`: `X` is a list of $m$ points with $d$ floats each. Return a list of $m$ int labels. For each point $p$:
  1. Compute the squared Euclidean distance $\sum_{j} (p_j - q_j)^2$ to every training point $q$.
  2. Take the $k$ closest training points. If two training points are at the same distance, the one with the **smaller index** in the training data counts as closer. If $k > n$, use all $n$ points.
  3. Predict the label that appears most often among them. If several labels tie for most often, predict the **smallest** of them.

No method may modify its arguments.

**Example.**

```python
X = [[0, 0], [1, 0], [0, 1], [5, 5], [6, 5]]
y = [0, 0, 0, 1, 1]
model = KNNClassifier(k=3).fit(X, y)
model.predict([[0.2, 0.1], [5.5, 5], [3, 3]])     # [0, 1, 0]
```

For the point $(3, 3)$ the squared distances are 18, 13, 13, 8, 13. The 3 closest are index 3 (distance 8), then indexes 1 and 2 (distance 13, smaller indexes than 4), with labels 1, 0, 0, so the prediction is 0.
