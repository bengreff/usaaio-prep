- `logits`: float tensor `(B, K, H, W)`, raw scores for $K \ge 2$ classes.
- `target`: long tensor `(B, H, W)` with values in $\{0, \dots, K-1\}$.
- Let $p = \operatorname{softmax}(\text{logits})$ over the class axis and $g$ the one-hot encoding of `target`, both `(B, K, H, W)`. For each class $k$, sum over the batch and all pixels:

$$D_k = \frac{2\sum_{b, h, w} p_{bkhw}\, g_{bkhw} + \epsilon}{\sum_{b, h, w} p_{bkhw} + \sum_{b, h, w} g_{bkhw} + \epsilon}.$$

- Return the 0-dimensional tensor $1 - \frac{1}{K}\sum_k D_k$. Gradients must flow to `logits`. Do not modify the inputs.

**Example.** For `logits` of shape `(1, 2, 1, 2)` equal to zero everywhere (so $p = 0.5$ for both classes at both pixels) and `target = [[[0, 1]]]` with $\epsilon = 1$: for each class, $\sum p g = 0.5$, $\sum p = 1$, $\sum g = 1$, so $D_k = 2/3$ and the loss is $1/3$.

**Restrictions.** Only `import torch` and `import math`. Library Dice losses are not allowed.

**Constraints.** $B \le 8$, $K \le 10$, $H, W \le 256$. Time limit 1 second per call on CPU.
