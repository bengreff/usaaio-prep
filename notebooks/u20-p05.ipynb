
- `q`: a float array of shape `(A,)`, the action values $Q(s, \cdot)$ of the current state. Several entries may be equal.
- `eps`: the exploration probability, $0 \le \varepsilon \le 1$.
- `rng`: a NumPy `Generator` (from `np.random.default_rng(seed)`).

The policy: with probability $\varepsilon$, pick one of **all** $A$ actions uniformly at random; otherwise pick one of the $k$ maximal actions (those with `q[a] == q.max()`) uniformly at random. So a maximal action has probability $\frac{1 - \varepsilon}{k} + \frac{\varepsilon}{A}$ and every other action $\frac{\varepsilon}{A}$.

- **`epsilon_greedy_probs`** returns these probabilities as a float array of shape `(A,)`.
- **`epsilon_greedy`** draws one action and returns it as an `int`. Draw all randomness from `rng` (for example `rng.random()`, `rng.integers(...)`, `rng.choice(...)`), never from `np.random.*` directly, so that two generators with the same seed give the same actions.

Do not modify `q`. Note that `np.argmax` always returns the **first** maximal index: with a fresh Q-table of zeros it would always pick action 0.

**Example.** For `q = np.array([1.0, 3.0, 3.0, 0.5])` and `eps = 0.2`, `epsilon_greedy_probs(q, 0.2)` returns `[0.05, 0.45, 0.45, 0.05]`. Over 20000 calls of `epsilon_greedy(q, 0.2, rng)` with one generator, the share of each action must be within $5\sqrt{p(1 - p)/20000}$ of its probability $p$ (five standard deviations); one correct solution with `np.random.default_rng(0)` gets the shares `[0.04815, 0.4502, 0.4483, 0.05335]`.

**Restrictions.** Only `numpy`.

**Constraints.** $A \le 20$. The grader calls `epsilon_greedy` 20000 times per test.
