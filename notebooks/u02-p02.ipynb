# Unit rows and a cosine similarity matrix

Word embeddings represent each token as a vector. Two tokens are compared by the **cosine similarity** of their vectors: the dot product after both are scaled to length 1, which is the cosine of the angle between them. (This is Parts 5.4 and 5.7 of the 2026 Round 1, with the same restrictions.)

Implement

```python
def similarity_matrix(V):
```

`V` is a 2D float array of shape `(N, d)`; row $n$ is the vector $v^{(n)}$, and no row is all zeros.

1. Compute `W` of shape `(N, d)` whose row $n$ is $\hat v^{(n)} = v^{(n)} / \|v^{(n)}\|_2$, where $\|v\|_2 = \sqrt{\sum_j v_j^2}$.
2. Compute `S` of shape `(N, N)` with $S_{ij} = \hat v^{(i)\top} \hat v^{(j)}$, the dot product of rows $i$ and $j$ of `W`.

Return exactly `return W, S`, both NumPy float arrays. Do not modify `V`.

**Example.** `V = np.array([[3., 4.], [1., 0.], [0., -2.]])` gives

```
W = [[0.6,  0.8],
     [1.0,  0.0],
     [0.0, -1.0]]
S = [[ 1.0, 0.6, -0.8],
     [ 0.6, 1.0,  0.0],
     [-0.8, 0.0,  1.0]]
```

**Restrictions.** Only `import numpy as np`. You may not use anything in `np.linalg` (so no `np.linalg.norm`), the `@` operator, or any loop (no `for`, `while` or comprehensions). Everything else in NumPy is allowed.

**Constraints.** $1 \le N \le 1000$, $1 \le d \le 50$, entries of absolute value at most 100. Time limit 1 second on CPU.
