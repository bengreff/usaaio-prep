Write the two losses of one GAN training step.

```python

    return loss_d, loss_g
```

- `G` is an `nn.Module` mapping noise `z` of shape `(B, z_dim)` to fakes of the same shape as `real`.
- `D` is an `nn.Module` mapping a batch of samples to **logits** of shape `(B, 1)`: $D$'s probability that a sample is real is $\sigma(\text{logit})$.
- `real` is a batch of real samples, `z` a batch of noise (both with `B` rows; the batch sizes are equal).

Compute the fakes **once**, `fake = G(z)`, then return two 0-dimensional tensors:

$$\mathcal{L}_D = -\frac1B\sum_i \log\sigma(a^\text{real}_i) - \frac1B\sum_i \log\big(1 - \sigma(a^\text{fake}_i)\big), \qquad \mathcal{L}_G = -\frac1B\sum_i \log\sigma(a^\text{fake}_i),$$

where $a^\text{real} = D(\text{real})$ and $a^\text{fake} = D(\text{fake})$ (the non-saturating generator loss).

Gradient rules, checked by the grader:
1. `loss_d.backward()` must give gradients to D's parameters and **none** to G's parameters.
2. `loss_g.backward()` must give G's parameters the gradient of $\mathcal{L}_G$.
3. Both losses must be finite even when D's logits are as large as $\pm 300$.

Do not call `backward()` or any optimizer yourself.

**Example.** If D returns logits `[[2.0], [-1.0]]` on the real batch and `[[-3.0], [0.5]]` on the fakes, then `loss_d` is $0.7201 + 0.5113 = 1.2314$ (real term plus fake term, rounded) and `loss_g` is $1.7613$.

**Restrictions.** Only `torch` (with `torch.nn` and `torch.nn.functional`). You may not use `BCELoss` or `binary_cross_entropy` (the version that takes probabilities). `binary_cross_entropy_with_logits`, `BCEWithLogitsLoss`, `logsigmoid` and `softplus` are allowed.

**Constraints.** $1 \le B \le 1024$. Time limit 1 second for $B = 1024$ with small networks.
