
- `probs`: float tensor `(M, K)`, softmax outputs of the model on $M$ unlabeled examples (rows sum to 1; the top probabilities in a row are distinct).
- An example is a **candidate** when its largest probability is **at least** `threshold`; its pseudo-label is the class of that largest probability.
- For each class, keep at most `max_per_class` candidates with that pseudo-label: the ones with the highest confidence.
- Return two `torch.long` tensors of the same length: `idx`, the row indices of the kept examples in increasing order, and `labels`, their pseudo-labels in the same order. If nothing is kept, return two empty long tensors. Do not modify `probs`.

**Example.**

```python
probs = torch.tensor([[0.96, 0.02, 0.02],    # class 0, confidence 0.96
                      [0.10, 0.90, 0.00],    # class 1, 0.90
                      [0.98, 0.01, 0.01],    # class 0, 0.98
                      [0.50, 0.25, 0.25],    # below the threshold
                      [0.97, 0.02, 0.01]])   # class 0, 0.97
select_pseudo_labels(probs, 0.9, 2)
# idx = tensor([1, 2, 4]), labels = tensor([1, 0, 0]): row 0 is the least confident of three class-0 candidates
```

**Restrictions.** Only `import torch` and `import math`. Loops over classes are fine.

**Constraints.** $M \le 100{,}000$, $K \le 100$. Time limit 1 second per call on CPU.
