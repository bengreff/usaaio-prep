# Category share report

A sales team wants each region's category mix as a table of percentages, plus one row for the company as a whole.

Implement

```python
def share_report(sales):
```

`sales` has one row per sale and the columns `region` (string, possibly with spaces at either end), `category` (string, already clean) and `amount` (float64, positive, at most two decimals).

1. Strip whitespace from `region`.
2. For each region and each category that occurs anywhere in `sales`, the **category amount** is the sum of `amount` over that region's sales in that category (0 if there are none). The region's `total` is the sum of all its sales.
3. Add one more row whose `region` is `"ALL"`: its category amounts and its `total` are over all sales.
4. In every row, replace each category amount by its share of that row's `total` in percent, `100 * amount / total`, **rounded to 1 decimal place**. Round `total` to 2 decimal places. Compute every share from the unrounded amounts.

Return a DataFrame with the columns `region`, then one column per category, named exactly as the category and in alphabetical order, then `total`. The rows are the regions sorted by `total` (after rounding) from high to low, ties by `region` A to Z, and then the `ALL` row last. The index is 0, 1, 2, .... The report has no missing values. Do not modify the input.

**Example.**

```python
sales = pd.DataFrame({"region": ["north", " south", "north", "east", "south ", "north", "east", "west"],
                      "category": ["Toys", "Home", "Home", "Toys", "Home", "Toys", "Garden", "Home"],
                      "amount": [30.0, 20.0, 10.0, 25.0, 40.0, 60.0, 15.0, 60.0]})
share_report(sales)
#   region  Garden   Home  Toys  total
# 0  north     0.0   10.0  90.0  100.0
# 1  south     0.0  100.0   0.0   60.0
# 2   west     0.0  100.0   0.0   60.0
# 3   east    37.5    0.0  62.5   40.0
# 4    ALL     5.8   50.0  44.2  260.0
```

`" south"` and `"south "` are the same region once stripped. South and west tie at 60, so south comes first. North has no Garden sales, so its Garden share is 0.0. In the `ALL` row, Garden is $100 \cdot 15 / 260 = 5.769\ldots$, written 5.8.

**Rounding.** The tests contain no share that lies exactly halfway between two one-decimal values, so `Series.round(1)`, `np.round(x, 1)` and Python's `round(x, 1)` all give the same result. Values are compared with a tolerance of $10^{-9}$, so an unrounded share fails.

**Restrictions.** Only `import numpy as np` and `import pandas as pd`.

**Constraints.** Up to 50,000 sales rows, 30 regions and 12 categories; no region is named `"ALL"`, and no category is named `"region"` or `"total"`. Time limit 2 seconds on CPU.

````python
#@title Grader: run this cell to grade your solution { display-mode: "form" }
# The grading code is hidden on purpose (double-click to see it). It writes the site's test harness and this
# problem's grader to a temporary folder, imports your file the way the contest grader does, checks the import
# and ban rules, runs every test (including hidden ones) and prints a table.
import ast, base64, contextlib, importlib.util, io, json, os, signal, sys, time, traceback, zlib
_P = json.loads(zlib.decompress(base64.b64decode("eNq9XAt32zaW/isY95wxmSqM7DTtjFN3103dNrN59CSZM52RdVRKhGSOKVJDULFdj//7ft8FQIKS7Cbp7LqntoTHxb0X9w0gN3t5tnek9tbDPz9cDb/aG6i9pW5SNN1s9zR5U2i2PUsbvajqa2XO01qrWq+quuGIQhtTlX5acXAoAPNy3WiD1sMhvmb5fJ7P1kVzjZYv0KDLpubnPYE26aAtq2xd6EmZLmXVVV1NC72cmPV0mRuTYyEMSouiutTZJF9yGlcZ7ZXr5eqanau0zFKzN8bHaVqWmhSN3LdJUVUrjp+nhdEkT5tm0uRLXa2bCTseE90mXVigFlZSa6C50h30ZFFX69U0WC9BQ5nl5UIWNtW6ngkBVZ0v8jIt7Mi6OURjU6/1Lb7P80JYdLMHJpRgY7ISnpzJfz/qYqVro5rztFGOD2pRpxkbLeUq+mVeV0vlpvvWB7/EyVl5Vp644So3alaVDZj+aF3mjXn0Nf9888hBxfdVnn3zyI4GEol63qhMz3NAVWSRmq/LWQPuG8WdyaRx8sDi1qQXGp/0WdntkrL7qH5B0y+JOv355OVPL07fqiI3DUnSFiyBGUhUdVkqTHqzLlX0Pjc5sIqfqrcE16h6jWX1ew3Z4ySQ9n2aF2tsiqrT3GjFr+oyb85VqgDPpAutLuu8AcFqXtXqW10eKZNeq0uie5ka9FblYqBKQlUZJEed51mG4Xm5WjeG3LO7UDqWzqrVdfuF8kL+uq8ieQpQyxVbP1PPqiKdqmc//dUoqoqhsNbkVUlU9o16eah+qquBMpUqqllaECJEx6jlGkyZajIVLKqoNDm6i5x8mIL+c2zlPJ01VZ1wpcNkyM1N1WLN/Y/sygawCv2wOa91mimz0tgwjIKELs6La0hLMVfABVjMqpqMrjWZ3qh8TgSdbhtHh9MOMuXtT6en302+P3n27vUbdczVSfBZOStSrM5tiE6vZnpFSYmPzkqFnxW67CgIFDiuswiymA3U0iz8GKxbVmRymbkW/nSbG3FsB2RSrqIrPxd2JJjjd6iqZ+dB6xz056Vp0nKmo6uB7U/e6dJUdRxMl2V1s65LdZVkMIqz8yhOZqs1fss2R7EdrIVM9VyWO63rqg6gWJIDWOUqSU1a1+k18O7oEKMSLapmAKEs+RvyeXy2B8mGqTzb85gZ8LpZrwodkXKMjxOZ6nARnmKMH0RY8UDNz/ZuCPAWG27sWurGwO7oq5WeNZCJm2D8LZbrMJsVlbkPs4Gqm6o4PtAPvxyo1H38k8cXmnUJlFtkU9Ncg9B5UaVEjO2yZr8jIGZh6SNJl/bjHeRYAtzweBdtvkvIaxfAhuQG5i1vMBt4FEUUK1gKSmHYd+n6wuVpSdNcLGHJOXk5b4E7QeZ+F4VjIpjh2MVfjl38FQqermtwTBhB5JbpFf+kUxMt1EN1GccxYYPQ/FetNHyXGlL5dqhKn09i6NT7tIB9UFGR1guaXU2BVTf4c5Q8XtzGvb3X/1qnxQdIZV+p7FAiDW9ISR/EwlEgkTZN7frP9mQ7OiA7xCUQkG7M/7lYhKIh+E8cH4DcljKFXG0hcFeONjGGMBBbIb6DAWN8g64/1D3UOApNvc3IzURURLaDn8yd+yELBjvSTdnCn61ghQMcJxOJtyaTHjoLmD/unFt13w/a91D7mE5Mma7gyZv/X9NcVKX+SKOMJb1UXlEkxbfvxXCJmWhvH58oy2dgJAOXOMSnT1wPOwKMMq1Xx+8Q6QUC5hB8B/ZtorcNIO65EDYlhCl9Vz3GY1uidKCmn8739IP4bodYvZD1PpXtKdkucEzfFLiFUruIiabxLluD6Z2d2Tm9tRNT91H2tq/ZpGBgrR0kuwz3ysMRCEHwUk2WaygFoptoXg7UA5hTaMaDBxeX/OQxsaHjM3gANS9l5TkDVNKAGAdR4ALqlZYIGevFeomgHC5ErYp0phMfdYrXpz7BLo46zUpjiWdTTuCKYzsS0RmdRxltImS7OSWHIINgE3OqRiija+Rz0a/5KrJzZI2egIs9aYXLGZFrJDaQxUwzwM/nOSNLhswdLZ/d5OpzdXD7VF1W9QWj+lSEV3H7EI+29tIxGch3HGYQn5eR0QwFgdWHcvlpyGJY1zDk3hFJS/4If2gQe4MCMkhi3f12JoJ3m0NxqdcAUTMFQ4K0NCRJ4vCX6UzNsACngEgbdTNbkcCewQGyi4fTdQZLymaJwRGW50yg1iuok2ou85kmQHAoh4eu5k6DgbPnqBERYssc8RLdNyFgxhp9iXrHjrxGM4gp9rkE/YJdjBzWWU+smiGjSUBIkF3OJwJG197a/IYoZc3uyYhRGheOOHqPldtE9UiFaUMnkhOKYk1tiA5DucMmYpmvjx2kDVM0RU5z0TWlhwN1cYjV+vZRUB5sNPYo+W1WCJ5kxKGw4TBoFjYgY4syutg72BHGmyFBokdNVYncIPnPmqPkcH6LKB9Spp+6rKsVYhEmqJ0AUuuS+fyNY42bJzkkEju/xTt+ZElklkiD5xCtBQNYm2TeuH0SQNSOexW0LhfQTp35DesSHOxkVi3B7XkK8Zu0A8+kYrM2aZpXE6kw9KocP7AFGgDMuuJBLkbjPC8yat2MlREkzDl1gSP/8vb1K+BaiuaYJgOGoqur6+YcLQ+fK7+O+toXkLK8/kZ93a2BTnyHVv5b2ppvtlJ+KFuQ/mNjr5oin7ZN9g9aknWTF11z1X78p6m6IsEqbc7D6SZflGk3zVybjfKC/1LDL0zT2UXbAikuWSywtQYzy0FmqVtzh4aLvHlYaAxTB8mXyYGL6cwR60RLlusGYlKQXTjn443fKs3B5cVZ6ddI5nkBifZfo7M9oF3Vmhmgq7QgEk0e/LW8KFnDMVXBikolFQAsmK+4a8kDFyn+ePrmFJrjeJH8hL/RZMIy2GQSs8rG6YjqVmkNzM5KcCXh4ATSqusmGkIZED4QTAyAu2pfUgtSn8FX/ys9UqdfDA+lEvbibyd/fzv59uTVq9PvgMHN2d7EVQ8nExIDwvx+2q/6Ss/cJ8T59tOsWq6A69ne7Vn54vXrnyavXn93+hbgIshK8n0FXeWHv0Fwtf14Yq7LWdvxAjHkM8Cw397q4Mt3CDK7bz/okv65qk+vVnFYYnlnazG/VWWZpMj0lhFlbL2EvYHb0q3GSqroAYXVCEY2s8nsXM8uTGTqGbe4Sfv+9o3Vd0bDcFbkY3VhmbM0CzDmKcwiI0Bb4xO1Mc1+uzsSDLlSIh1VU+e2rNhzUdYJjca7otmm1gjshEuQEuTXQLQfh769Rn5+JXEoi3J6Oz4cObSPIC1SBnbYH9FMGpnuUuSqtHbmRif8W1a3R/yM0Uh+bh2CrhpNlwB+JTDM0JONEjV5NBo7RKdp2y5+svEuR8qhk7URYEH7lDVyWyt17RRxhIzQ3TyDqjASq22gYAEDgh0fMaNh0J1PWZAHORcSu+kioWC7BaQsHvB8mTbLdRG2zKU6AlNN2wzmX6bFRcTN2HDdQZjO4VaebWqwmVW0oSwHJoLu0bYDcyUVxMtRKoMYlC9ZWrBfE6wGRjJyjJaIPGGOmMwR+FKQtRsRH+32jcFeJGmWuTXi7cHd3gTjErMqcm431xw9PBjvmNhtXrJeZYy6d8wNk8TifjZ+D6u3SQ02FHslfHSFd5DfKdQOXsoMx0h8/h1c3OQgoG1wIWBdjwO7JGBj6l3M+6SpQufHMv3E684m7RsCIXhQ0T4I6qvAIt8HMM/upmr3oF0rds5qc01R/CRdrXSZRb7iYyed7VmTZ837f/X5tWsV8YLf5uVr58VO1osTQx8U29rKxoSkcuNeps1LhIxbMm1t0EciB8QCgQxAMkh0oO40/z4KsFpiRZ6tohAcl/yzYnoqBjYK1onj3dblc3oUMMaDutkfqH0LxGsUFXAf6yHvXeyzFHu7u564C3+WLHagbzzuHbDP2hPEaVoa8nZ27jyEhIKdg5Dys06RxP4sf+OnNhstLtNr89CeqCKTQVyYt4AolA5aVRbXA78mMpfO1URWCmwRyYVSDDquVaEXSM0vGTYpGZkkSRw4MwRlKqu4KRbweS75pd2FKHCafxTnGHhhi65zvhDDf6so8KV/VL24MBAhrPDRogMMjHL8IfzIxDtkB5Dj37PDZeXXaI9ku10G5n3iJ1aJJtXKxpP9ahuHWyXbUL0PIZgZJeTiv5WHzZQ10BsV2cjJrTAajkPh3kH+x7CgW/U+4sW4bZEsrZ9GsA2TRNh7hKKZakyhrvW5LiWJbWNHmfWfpd9RdntPpj6vc4ApricSyUYAgUR3wmyqH9G/LvVDwTPTZlbnklWoFUQV+r1Q7iRaBuT++JoJG9RVc5yRw/p+hWnKsorPWxPkzfjcTJpppJPJpO1AyueCTQJHpDkXpz7nOgCBnZoz/dQ+5mPm15LgotJLlkxYsKKVldqkEzr8ZjgWRO29Jm6FCAxHy3Fah75jJk9M5IREB+cjjP5vb2TVW473zF4iEo/aXAyWdjPLZSab1ov3ozZE9KTcOfJwHKTEjlMMwI9VO+SxD9ch92hnwSEpqjQzUSRIPKLYUCekFLFHePBZrGRE3tghfSJEh0w4wHNDqonHalsvbMJ45q7XnMldGtN9tYUZK7Iuk6YksuFVhS0I4Y86aGOic0cq2io6I1iqoBlZrMbiLfhNtGQchrM+pfPxrJG649ayoZWQqkUk7OTlEBPZ4aGXt1LiS55zIJ1XyduG5ZPnrz3zbJEnsX8i9+3t8x9OXrx5OXBZuhsKczaR4q4/BQ5sWf+SEnl5OPS49LNjuQTTFauwnVleayif3YwImG6GWR5H3eRcovZoPn/3/OXpm8mb05MXgxa7jTDHrDSlp18KS9g6YW3GVnZYnZQjk4Ci4JaXFY2uMIdtCkzVxnrr6fZyDpYsyKUj/tqcCI2x48yI3ZIwjK3gb5MkOqTrhDWgiZ1GyxN/AuOG/dqEK7v0ShJOEJ1yjK01szTSAotVE4vbVNWFnCmoG78fR8mQFWBvutwq36ZGtxUiWwPpSmLfvjgdDg86BD6KjLswdgdDgiZPYnhDrHIb5YKge/xRq9ZboHcxytsVrgt5pkTJSXyEBPyL4XB4NP5kRZZbif06jjSF9zvsGH9W7E80QcvAdsW/vbkdr7JK26DJXraDdyGI2zBv/0gSFh+nk2d79tIftbB1Ge09wJZud4/wt3VvESjf4i5dstDcKHfdzxW2fJZnh9Ay+KuDPoh3Zo++hkFDhERlbg07P9G2v09r49dI8kYvTSS3Z8q+VxCr6s/9eYCSAo1oHocFL1gJHkJKSMIVN4JncceU/noNu2WFgtGK3CIoPWkbBpfmOS/XOty/mXWw1h4eCZDRk6Mx7LecCwNbV6VWYh7P9hxoPzhYjb2sBjPt2ZWmOHtrD33YNEyGtx9zJrZ98eETvM7v9TzucG7bKn+SSXObMOo4R01l4L19d2PLgneTbexhdVyMdVGVC38us2mzVZSX9paZRPMDJhDztFb+ZDAOjYBbXA43Nmrav5PgFmeG1zvuqnyUL/lP4RLeM5DziuwDfUgHqxVwwpO72dFdB7UD9XiXe3Ph7NgnZHVr2D7aFbUTbKx6zNQx6mLX+6JSMSrTqiqiLcRsH0E1PckluKYHrp3hE5Q7XYpca54rn/BY6zaZML/hQdmRPx5gumNPdltnwTPdXTei2wNQXpNn4ypzR3dz5oVyqdwf3rkG179keuS6EBTDg9555DeQ4/FBeEVn4G6TDHhoTbraS+hwG87866t0uaIZHfMs9Q0vhTC7B0MWTtyZAgLuzL1+GKh0SeEhe2e8nBHbGxi1niMnLGdaTTWvg/Dui2wpVphxNKbDEsiUhKiI72IuXzfn1iS/q64ltn8MwYGNR6cykC7X+2NlQ+XDtjeY6jsP2k7N0+se2MMn7PNyzqyIwFUfwBe7oHsIX+6A/kNaZ7qUtZ/43kvtez1YmQkWv3tuuc/6jB3xM/88aQGntvXvHcJB68/91lkwtoOgsmCw5Uen2+jPHMX/6LNaB61fPbHYvjn94fnrV05cAn4Eu9IxoiOae1x3R8btiPO8KIz9WKQX2n1EqFm7Krac5xmRxWcn79y638KPmE1ed5w92/ufvEGe7JrftoeNftvGXZHCSi60HWLu/XE6k9Djtot5FhD2gXov9gMDA0/DIyaMp2l3tzx4xRYp74oXpG/C0lY2mtHMZRJEz2jvYb3fuyBbXnL0F83kjAmw7Z3aKAwX2O8xynw8t+EABaZdUz7uWtdf6puhXQYFt1r0vMcVd7esKtbL0gxYASvl4omzCwPrUOCV5PUSRqzLzZamatIitjfyGcSwOMd7clhmQEqqmlYzLJt1iJHP4V7Zfpgg05W9ZWDsb6gJEdVlZAPW3jGdlMrXyyhreRs6O4c1dTKiaVEPNjatidWjRyo6RAccBNKDg6HdEbGAQGkslf5Sk+wGweQBj5wRQWoxf/6lT3g4V6e87j062FrtEZDtw946urcEOrQHBAVWC1ZDzwvhbMcoYeuFvj4u0uU0S9XiSEUPyRB0jBbjjisDtYj79dQRL1kvpCYxBitkAz4X38HNPeNLrxG5DpWxwGwesmj3l7NkxNneyYsXVEu7bYFS8j7WDtl7VTnPAUbCpTSXWt7liFeS1Y1LQd1zPOvqFB9k8F5GqvxzNAL4KDFz0LFBG1KzW1HHbWJcaF7cbCILAOO/lib3dfsWhqQmGyaA90qiEDxFcCTYjj9UpgnMGYsd1oQ/V5juQ43owEr9e8hRvHVazgcYVwgSJfqIrhgvdhPhjITKsAXa8eCB+nJXurNNtmuxeYYXiIW9+6Pl9hwCmmM4Vg9ODr0bWMBV012eBAE97jLeLReeLkIBD92sgMIltIIJPSLBOnHxiYkeD9SfqArbzV/2Kw4LidsgJc5LjnIbeOaCWDI7r3LkrRQAN4BVvfxXfYx1XVZ7LKyIAyV3Jm5E93c3QPZ6aBf3QDO6EEG+ue0Lx8LGxkLBZiLFKVBkIjHr2yKpT7lbjhG3fZh8NWbSNmLvaDgOjRUDye7uTHsJu9u0clNCFtw2ixMp5/k61ljaw4m2o78ryzgedxA+c1rtY1dkLWltNmoPUmaxNG5Cs/pr++I4ZKSvzvWpP3iyQ8gXNmppixXyB/9vrfaYuH+O8Z/3xm+NO9zChMz1OVlko5WNORB9GygPQUf/0sKWud3QTnaF9+TewDwibZTbZBIV2wu0WdogL256D2Dkll0PtoO5ypLvMP576b/pfMqRGi18yDUZuPvQmC6nJz7hkGFWFCdCazDsztu+5Lg17TL9vXWQHsgkCO94KNm9+uN5iy1lsm+gihTZzIChyzGPbIKXX7yilxs5yOmfYMs1pO6ZBIAl4RPvQcimePfDJUAZ9Jjm3i8JNrdH3euMdkBbp5xVhbHP1kSd5qFjk2XEwfCUdWaDu5gZrswKmBku5sZJWnizNfv2afdgat/u6/7AXn20+8cw6IS18H+4MHJf/OF+/zliCzfHPlwJTtJkbQUVU57DxfFdWHJFmcu31Osi401v6ABd1EAlSWJLdAxhwRAJR52VWBVrIPjiBUNbCDSfT6TzBv3GHj/EG+8mq0bOqHNTplGcNNXEvYpNeCJ3L3p2/7vXk3LcE7xNTD1KgqooWmFjyLTNvuEcYYiHHVLyvg6RSiPOmfJ9JcHXugljt3H4TtZGhXz2JofK9usl7LdMv+R0+4D2XloEUysFnuXnFX7pdHauJCdaQSTcuApSPbCs3gevNzbfvREcbeMwvguH2hcqXIRt3yPawNcGXOo8X5xT7grmGxKNT1vX0BNIbn6RssC610WAYKrloY90u63OWBE53nqyGz55bfjyEPQ83qKnfTZ8cNi9G/7zhn47Au56Z+mCBCrt6ODo4cG4/8ppV4Yw2Xj2JBscvkC19CZFNRvlA5t63Eexv0vZf+bbrUnChiGBv3kE0GrXJsz0che8PsNsztneS/FJHTb5AJnhLF+CnRIj7XjtysODP7DI5xKUTVGneNBmdNvhVovWRtubYTaHYEDOh01yQMlXWaK/g/sepXi5tndSVYqZ/KcbkL8ESmYteIj4/W+l/Su/yLGhHwJsPJTucdCbIsmnhFTrJNVNfvuBb6fDoqIcTriNDzzr6c/2eYErPG5OpqreMZPVM7lnkducenuy/UcswulSWpZ//6FUkXUHj0NpDMD3Eg8rYe4fxZB7CTdsvt1eU/gSLumC3xbcF0xinkg0Fl7vd0/9DpPhRtX2Q0KGrUiFNTXg4fhiQxb8zxr17e3/AuDjkMY=")).decode())
_dir = os.path.join("/tmp", "usaaio_grader_" + _P["id"])
os.makedirs(_dir, exist_ok=True)
for _n, _s in _P["files"].items():
    if _n.endswith("::b64"):
        with open(os.path.join(_dir, _n[:-5]), "wb") as _f:
            _f.write(base64.b64decode(_s))
        continue
    with open(os.path.join(_dir, _n), "w", encoding="utf-8") as _f:
        _f.write(_s)
if _dir not in sys.path:
    sys.path.insert(0, _dir)
for _m in ("harness", "usaaio_grade", "grader", _P["meta"].get("module_name", "submission")):
    sys.modules.pop(_m, None)
import harness as _harness
_harness.SPEED_FACTOR = 1.0   # Colab is the reference machine: time limits apply as stated
import usaaio_grade as _G

def _usaaio_grade():
    meta = _P["meta"]
    mod = meta.get("module_name", "submission")
    path = os.path.abspath(mod + ".py")
    rows, static = [], []
    if not os.path.exists(path):
        print(f"{mod}.py not found: run the %%writefile cell above first.")
        return
    src = open(path, encoding="utf-8").read()
    static = _G.static_checks(src, meta)
    per = float(meta.get("test_timeout_s", 20))
    def alarm(*_):
        raise TimeoutError()
    use_alarm = hasattr(signal, "SIGALRM")
    if use_alarm:
        signal.signal(signal.SIGALRM, alarm)
    def limit(s):
        if use_alarm:
            signal.setitimer(signal.ITIMER_REAL, s)
    buf = io.StringIO()
    error = None
    if any(not s["ok"] and s["msg"].startswith("syntax") for s in static):
        error = "syntax error"
    else:
        try:
            with contextlib.redirect_stdout(buf):
                limit(per)
                spec = importlib.util.spec_from_file_location(mod, path)
                sub = importlib.util.module_from_spec(spec)
                sys.modules[spec.name] = sub
                spec.loader.exec_module(sub)
                limit(0)
        except TimeoutError:
            error = f"importing your file took over {per:.0f} s"
        except BaseException as e:
            limit(0)
            error = "your file failed to import: " + _G.friendly_error(e, path)
        entry = meta.get("entry")
        if not error and entry and not hasattr(sub, entry):
            error = f"your file does not define {entry}"
    if not error:
        gspec = importlib.util.spec_from_file_location("grader", os.path.join(_dir, "grader.py"))
        grader = importlib.util.module_from_spec(gspec)
        sys.modules["grader"] = grader
        gspec.loader.exec_module(grader)
        visible = set(getattr(grader, "EXAMPLES", []))
        for name, fn in [(n, f) for n, f in vars(grader).items() if n.startswith("test_") and callable(f)]:
            rec = {"name": name[5:].replace("_", " ") + (" (example)" if name in visible else ""), "passed": False, "msg": ""}
            t0 = time.perf_counter()
            try:
                with contextlib.redirect_stdout(buf):
                    limit(per)
                    fn(sub)
                    limit(0)
                rec["passed"] = True
            except TimeoutError:
                rec["msg"] = f"took longer than {per:.0f} s (infinite loop, or far too slow)"
            except _harness.Fail as e:
                limit(0)
                rec["msg"] = str(e)
            except BaseException as e:
                limit(0)
                rec["msg"] = "your code raised " + _G.friendly_error(e, path)
            rec["seconds"] = time.perf_counter() - t0
            rows.append(rec)
    print("Rule checks")
    for s in static:
        print(f"  {'PASS' if s['ok'] else 'FAIL'}  {s['msg']}")
    if error:
        print("\n" + error)
    if rows:
        w = max(len(r["name"]) for r in rows)
        print("\nTests")
        for r in rows:
            print(f"  {'PASS' if r['passed'] else 'FAIL'}  {r['name']:<{w}}  {r['seconds']:6.2f} s  {r['msg']}")
    out = buf.getvalue()
    if out.strip():
        print("\nPrinted output (last 2000 characters)\n" + out[-2000:])
    n_ok = sum(r["passed"] for r in rows)
    ok = not error and rows and n_ok == len(rows) and all(s["ok"] for s in static)
    print("\n" + (f"Accepted: all {len(rows)} tests passed and every rule is met." if ok else
                  f"Not accepted: {n_ok} of {len(rows)} tests passed" + ("" if all(s["ok"] for s in static) else ", and a rule check failed") + "."))

_usaaio_grade()
````
