````python
#@title Grader: run this cell to grade your solution { display-mode: "form" }
# The grading code is hidden on purpose (double-click to see it). It writes the site's test harness and this
# problem's grader to a temporary folder, imports your file the way the contest grader does, checks the import
# and ban rules, runs every test (including hidden ones) and prints a table.
import ast, base64, contextlib, importlib.util, io, json, os, signal, sys, time, traceback, zlib
_P = json.loads(zlib.decompress(base64.b64decode("eNq9XAtz20aS/iuzStUa0FIwSTu5LB3lTraVRFt+qCxt7d6KKgQkhhRCEEAwoB7W8n779dczAwxIULYc33k3IjGP7p6efs+A93tJvDcSe6vBs4OiP9zrib2lrCJqut/uqZIqlWh7Ke/yLBbVlRRVGSVZks1FGWVzORKRKGSZ5HEyFSqZZ1GKialUKs8stNTgSbJVJRW1Dr+lxziZzZLpKq3uqOU5NcisKvF9ryglgat4Th6vUhlm0ZLpUKvJMlEqIdjUGaVpfiPjMFkWeVkB8MXeb0r3LaPqCp+5wt+CntJkwl+T6YIWxatb8uc0L+7wma2W+ouiAVHJYH7LJ5h3SV8nUZZJcOjCPIVpnhfAOotSJQFQqioE1HxVhegYDPtojuaaNlqMTIl1QV7ILJRZTOD0GnWzvCXmFnkaVWaBdc8sX5WJLMOZjKpVKZXZtjKZqmCpJNOnaMyUuZSXyTwxO1FEZTWkxqpcyTU9z5KUt+B+74qWSNsUFMzzMf/vF5nSbira6KgSRUmLl0sxL6MYjZrNwvt1VuZLYabb1v1f/WCcjbMjM1wkSkzzrKJNfbrKkko9/QEfPz41UOm5SOIfn+rRREQgTioRy1lCUAU4KWarbApOKIHtj7kx3Ne0VdFCQhzHWSMSQguL+JWafg3E8T+P3p6+OT4TaaIqpWUXYAFMCXWV32SCJn1YZcK7TlRCVPkvxBnAVaJcEVp5Lcs7nkRL+ylKUrCe5D5RUuBR3CTVFSkAwVPRXIqbMqlowWKWl+KlzEZCRXfiBuTeRIp682zeExmgiphkTVwlcUzDk6xYVQrc07uQGZZCLusHiBX4ax5ZVgVBzQq0fiNekdhMxKvTvysRlVIoaEYJXmUg5YkSb4fitMx7QuUizadRCogkWkosV8SUiQRTiUU5lDKh7jQBHya0/ivaylk0rfIyAKZh0MfmRmK+wv57GrMiWKk8qK5KGcVCFZI2jEaV+Wp+ld6RtKQzQbQQFdO8BKNLCaZXIpmBQHpkHTbrMEoEppydHh+/Dn86enX+/oM4BHYseJxN04iwYxu849upLCAp/micCfpXUJceRQJFHJexR7IY98RSze0YwpvlYHIWmxb8azbXw9gGSJgV3q2dS3bKmWN3KC+nV07rjNafZKqKsqn0bnu6PziXmcpL35nOaCWpdSZug5hs8fTK84NpsaK/vM2erwdLXqY4YXTHZZmXDhS9ZAdWVgSRisoyuiO6m3Woq6iQ3jyveiSUGf6SfB6O90iyyRSP9yxlinhdrYpUelg5jfcDnmpoYZ7SGDsIsPyemI337gFwTRuuNC5xr8juyNtCTiuSiXtn/JrQNZRN01w9RFlPlFWeHg7kwXc9EZmv31t6SbNuiOSa2EhVd7TQWZpHIAztjLPd4SxmrteHJd3orzuWoxdghvtda7NdvLwaAW1Iosi8JRXNJjrS1PMFWQpIodt3Y/pc9LCk5HZhCTPMSbJZDdwIMvY7TQ0TiRmGXfhj2IU/ruDJsiSOMSNA3DK6xUc0Ud5cHIgb3/cBmxaafJRCkosTfShfh6q0+cSGTlxHKdkH4aVROYfZlRBYcU8fo+DZfO239l7+vorSz5DKtlLpoSA6i1nSez5zlIiIqqo0/eM93o4GSIe4OALSjPk/FwtXNJj+0PCBiNtSJperNQTsymiTYhIGUMuLb2CQMb6nrj+VLdIwippam5GokFWEtwPf1M79YITOjjRTtuhHK7HCAPaDkIO6MGyRMyfzh50zWJ/YQU8s1Dalocqigjx59f9rmtM8k480yoTSSuUtRJJ9+55PLjFm7W3T4yH87XHg4rv0tBfXog4AvVjK4vCcIj1HwAyB58S+TfK2AfgtF4KmADC577bFeNoWL+qJyZfzPfosvushWi8Y35eyPQLbGY5qmwKDKNJIlDfxu2wNTW/sTOf02k5MzFfe27ZmYwU9be1IsjN3rywchuAEL3m4XFWcEXizrCf2yZySZuzvL27wzVKiQ8dX5AHELGPMMwSoWAPFOFdI1WJqppCxnK+WFJSTCxFFGk1lYKNO9vrQJ7KLF41mRT7HsxEmAOOlHknRGZxH5m0SpLsxJSFBpgUrH1MlhTKyjMi/fUwKT89hHC0BZ3tSC5cxIneU2JAsxhIBfjJLEFkiZG7W8s19Iv4iBusX4iYvF4jqIxZege2jeLS2l4bJRHzDYQTxSeYpiVCQqPpcLr9wWUzW1Q25OyJpTlbJHyqKvWkFYBDHuk/qmRS86xwKqN4TiBIpGCVIS4UlcRz+NpqKKSHAFFqkjrqRrXBgj+CAsouDySomS4pmjsEpLE+QQK0KUidR3SRTCYDEoYQ8dD4zGkw0W44qFiG0zChegvsGBJqxor5AnKMjKamZFpM+AQr4BY0MHJZxS6yqPqJJghBQdjkLGYwsrbX5hCjFVfdkilEqE46Y9R4Ks4niqXDThkYkQ4giFy68oSt3tImE5odDA2nDFE0op1k0TdGwJxZDwta2j0xyb6OxtZJPs4LpBCOGzIah08xsoIzNi+Fid7DDjTfdBbEeVXnOckPJf1yNguFsTVE+SZl8YbKuWohZmEjtGJBYZcjn7w1rzDzOISmxs1vc8Y9RUmZJafCMRGuOAFYnmfdmnxgQtONBBS2zOWmnjO2GNQkO7WScL4nbs4jEL6wHjrl6slJRlOQhVxhaVY6f0UIaQJQ1xYOEjcZVksbQuikqI5QwJ9AFjPzb2ft3RGvGmqOqmChkXS3uqitqOTgRFo/4wZQ4wjgpfxQ/NDiok55JK//NbdWPWyk/KZuT/tPG3lZpMqmb9Ae1BKsqSZvmvP6K8lf9YKpe9bOuzzWPd2qjvGAfSvILk2i6qFtIilHzU7rWoKYJLTOTtbmjhkVSHXDVTAyC74KBienUCHWiJcqBPTYplF0Y52ONXxElxOX5OLM4glmSkkTbR2+8R2TnpUQGaCotFIkG+3/PFhlqOCpPUVHJuQJACJMCuxbsm0jxl+MPx6Q5hhfBKX16YYgyWBj6ARkuTKeorohKomycEVcCDA5IWmVZeX1SBgofAMYngF21L64FiW/IV/8ejcTx8/6QK2Fv/nH032fhy6N3745fEwX3473QlCrDEIuhhdn91I/yVk7NN4rz9bdpviyI1vHeepy9ef/+NHz3/vXxGYHzSFaCn3LSVXz5Bwmu1F+P1F02rTveUAz5imDopzPpPLymILN5+llm8M95eXxb+G6J5VzXYj5VZQkjyvSWHmRstSR7Q25L1hrLqaIF5FYjENlMw+mVnC6Up8optriK2v72g9Z3RMPkrMDHfKGZs1RzYswLMouIAHWNj9VGVU/q3eFgyJQS4ahQNWVRabko7YQuLrui2aqUFNgxl0hKKL8mQttx6Nkd5ee3HIeiKCe348MLQ/aIpIWrxYb6Ecyk4ukmRc4zbWfuZYDPLF+P8J1GU/KzNgSa0jdcAvErIMNMerJRDwePLi4NoZOobmc/WVmXw+XQcKUYmNM+IX0Ida3UtEPEKWQk3U1iUhVEYqUOFDRggqDHe8hoEHQnExT8aTkLjt1kGkCwDQKunjs8X0bVcpW6LTOujpCphm0m5t9E6cLDZmy4bidMx3Atzzo12Mwq6lAWAwMmd7TtwExJheJlL+JBCMqXKC3ox4CwESMROXpLijzJHCGZA/AlE6s3wh91+0ZnL4Iojg0Of3twszfOuEAVaYLtBs6Lg8Flx8Rm84JVESPq7pjrJonpw2z8iaze5mpoQ2mvmI+m8E7LbxSqg5c8wzCSvv8BLm5ykKBtcMFhXYsDXRKwMXUX875oKq/zsUw/srqzufYNgWA6oGifBfWdY5EfApjEu1fVPagLY+OsNnGy4gdRUcgs9mzFR08a72mTp837f7b51YWFveDLJHtvvNjRan6k4IN8XVvZmBDkZtzbqHpLIeOWTGsb9EjiiDBHIB2QCBINqJ3m30YBWku0yKOVFQLjgt9ypKdsYD0Hj+93W5e/wKMQYyyo+yc98UQDsRoFBXxC+CjvnT9BKXbdXU/soh8liw7ylaW9AfZNfYI4iTIF3k6vjIfgULBxEFx+lhElsf/kT/+FzkbTm+hOHeiDV8pkKC5MakAQSgMtz9K7nsVJmUvjajwtBbqIZEIpBB13IpVzSs1vEDYJHhkEge84MwrKRJxjUzTgq4TzS70LnuM0/8zO0fHCmlzjfEkM/y08x5f+WbTiQkeECMOjRYcoUMLwB/A95XfIDkH2/8gOZ7nFUR/JNrtMlLcXH2olCvNCx5PtahuGayXbUL3PWTAySpKL/xIWNlJWR2+EpyMng+Gif+kKd8fyH8OCButDi2fjtrVkbv2yBeswiYW9tVBqhhpDqEt5JTNOYuvYkWd93fWbla0fyNRnZUJg0ruQI1mPQFCiGyKbakf07zN5wHTGUk3LhLMKUZCoVrhYYk6ieUBij6+RsJG6SoxTfFjfrjBNUFaxeau+STGtwmriySAM6w5K+UywCeAUac7Yqc+Ah0DQTs2Qfkob8yHzq5dgotIblExQsIKV5dqkETr6i3DMidpbTdgKFhiM5uO0hnzDTJyY8AmJdM5HEP2v7xnrGuMts5cUiXt1LkaWdjPLRSYblfPrizpEtEvZOXJ46aTEhlMIwA9FPeSZDddJ7qkdBYcgzaNYeR4T8RRiA53gUsQe4JHPQiXDs8aO0idANMS4Ayw3uJp4KLb1QieMY3MLZ8xXblTzqAszWmRNJg1JRMO7nLbAhX/RQLsEOTtS0VrREcFCBdWFpuqSvQWeWEsu3XDWpnQ2nlVcd9xC61oJrlp4zE5cDlGeHu56eS0ltuQ5I6KTPDirUD45eW+Zp4s8gf7wzNPZyc9Hbz687Zks3QwlcxZycdeeAju2rH2XCbwc9i0t7eyYL8E0xaoAl7dKScqnN8MjSjfDLEujrBKgKC2ZJ+cnb48/hB+Oj970auo2whxVSEhPuxQWoDVEbUZXdlCd5CMTZ0XOVTItGk1hjrbJMVUb+FaTbXQGFiMEag9/NieSxuhx6gLdnDBcasHfXhLrkCwD1IBCPQ2Wx/8CxvXbtQlTdmmVJIwgGuW41NZMrxEWmK0aW9wqzxd8piDu7X6Mgj4qwNZ0GSwvIyXrCpGugTQlsZdvjvv9QUPAo5axi2JzMMRk4iQGN8Rys1EmCHrAH9VqvQW6i1HWrgAvyTMkik/iPUrAn/f7/dHlFysy33ps13G4yb3focfYs2J7oklr6eku/9Ob2/AqzqUOmvRlO/IuALF28/ZHLmH+OJ0c7+lLf9DC2mXU9wDrdZt7hJ/WvbmjfPNduqShmVHmup8pbNksTw+BZbBXB20Qb8wefA2CBo8SlZk27PgG234dlcriCJJKLpXHt2eytldgq2rP/XGAEhEZ3sx3C15kJXAIySEJMG4Ez+yOIf3liuyWFgpEK3yLILNL2zC4MM9JtpLu/k21g9X2cMRALr4dXZL95nNhotZUqQWbx/GeAW0HO9jQi2ow0p6uNMXYW33og6Z+0F8/5kxs++LDF3idP+p5zOHctlX+IpNmNuGi4Rw0FYH39t2NLQveTNaxh9ZxNtZpns3tucymzRZekulbZhzN95BAzKJS2JNB3zUCBjkfbmzUtP/ggmuaEV533FV5lC/5WrS49wz4vCL+TB/SwKoFHPBKkuLY23VQ2xPPutybCWcvbUJW1obt0a6onqBj1UOkjl4Tuz4UlbJRmeR56m0RpvsAqmpJLsBVLXD1DJug7HQpfK15JmzCo61bGCK/wUHZyB4PIN3RJ7u1s2jOdM1bC6PmbYQ4qqKevWPNYS4lGlOUvHByVdECyYJI3BbIZx1vNvDp7inljNEcbZ5+pUHo9xme98QpEsxkKv2Rnkgbi/vfKG5REIX70grxCMJnskw9xmBc2seEMvrWsbA2ZzRk2SO3nEoCM84wQy9ilfEU4mAuIpJIqa6I4yl7K1t+44PJinNT61yUMG9SEG3cHaHmRwToubTC1wSetEbXF+b6BJAEH1BGYubd+rjlQEqgVstwISL6T9FKh6JIxELckuM+pc7iKgkX5CEWNHgQBM96pKk471Mx2fnB9r32HafYraNrvL5RP+TNObW6as0hR1sguOg+xyam6U5evMl47bHgKWPovlC/65C3xxcieriR0Nu4YTLO6vcNKEIwnl7eRssCHpN08vzD0cm78M3J25NzGvEdX6U9/XD8+uTVed36jFvfHv0z/NfJafj2Je67U8P50Yefj9FP3cNx9u79ydkxPw1wGu9picfNEy51Vz0BrEm1iiGMtCMRbQnuRbCWFhQ+mgoshIs06vXR+dHZ8fkZxwOmSlNTPhLeIHhG9qof4GUWrx/8Fd91y6BP2z4Y+D07zWjbEc/rB98Ne+JAD6an/+CJ3/LEbzFxuDXxpUH4PY96jnmDoEaOv5j3bGveKzOvzzj6Gt/3Fga1ML7nmEdxR/jq6NUv4OD92jm/nuHm5SmxkfmneoZTG7dOjDZ4kdjHDRStDvyVdGKfdGDf0QsdIy70PTg8bl+EczHhpgEZj8OB7161XEYLyQVnS4lLI2/robBbeME5piEYTqi5EcNGWO89tQcrMrl5ucT1hpp+Apd8lIfWMwF7XEY3XprDjpLwuxHVbQuMHcIAsq08QRBrdzKYWFUGGUGJUpDD4r0ByUBhauA76f/k6ohhTst3TcPBoIc3yoYY43CS7XMIO+PBGFD4M/nNLokNMN7LMl3jvRtErTKjgIDs/yF5mGp28D3H7mRsHUbUPs0jcEgMHIwF+w+pUzUd2bOz6OEit/smiomntWRu3x7Q7e7u6pdabAZhUxjGQ8YHOEA+il4+B++78z/tvfhvqNdOtJlvxnv5dVLG9ysP2Xh61vgGy0WM7x55m1lye2jwE9dCylHC5pwM9oYmMwzkfHi2ESe+A1BSx/1dFNFsBvLULtJUGXtNMw01jRqMu+suSEp9bsd7xnlf9C+DKscNE09zq+kZND229O2CbHhlAVJLJzi0dwAzLHb5UgcQzuUUjGnz5ypHjZqcI0LA6U1cX6dUwfQK4wCvs2L36XyL5CjQcoEQvWGbry9D2UWbRyslm7cVH7606eQXmy9gtQZ8diaw8UIKlGVtdtFG9Duq67V4z/DaZOryquYm2O1vXDd1fXrXldPue5htyjhp4wuZA+dCZlLp9/KQrVGwR+Pq+5f9R9y/9O4dEu3U3Fy+a7T6o9GtT6j1qkBhhRWa5dRYt4AcmSWDzaiJu4J/JcVP9Ol9dEwqm8+PDoM/BqxNMF2BvZdHDqKcgkuHbpXWLQ7ZakkB1phTWasl5TzNJ954DzcAN2sAZGvxqhXj8boKBJacoqGhCEqJd3KvZVjltbAHkQqLXCW3zbEGeaxwiWoxLxiVH9Jz+gj5DaqnYiC/c9+eM8NJOJqIry0glAAYHvPrO2aGlpS3L1+IhZSFk26Yi7oNNIyqWaZD56Bc4gaVaxwm+e2n916nLo4x373THVtszuOQNwIbhGeVYWkybr3LZM/W7CD4DzTVxWY7pm0KH1t/bGQq5NSqkJBqikP4jlRTADOn6Fh1b6eaPfBve6382C3S+j7VIw4aatNUc6XT2u+qiDZ3pr6OoW2ODyC3lq8bea4u1I++kiF2wyJciV0iDSKm98QmSzrjqCZMM2lysaC87pYiU7KOso7POqDiAHUxb8rCNjjRGz6zeUl4jwCcg7CAknc5l6XyBv39/b/6691BioXXhBQX+njuWucT17B5t+py/bmBwMPy8eloQN+NxU02yyiOCiydX+j3SRDe5dVPKJN1Xpyd6UN4bZi4Ulgfzfs+H6a1CuMzG0N7m6GhGxNuvWXWFmXedxJFs07iVKKPlZAZtItEHHB694x27b8wLwHp4k0sJpJ2StZwHvDSbOTUSEBD+SaTPn1T0TXe2LK2Xd+e1xeSIMw6wPd36uTmQmgjUn16RJpl3uxBMqDp/1O5DsQZYeRVZCrB6xEmLgVeXj11md872LGaMf/Mhr6npetNn7psv/d1DNDmYr9qsNdSccp30yRbeJuBYLuQ84hQcEvgOqNBfR1mqUWsOxBsv2vOK6cd9DEB+UY3NvuylmEVJnS+lxuJU/2aC0C1MZFCakSHhwLfb+2bv7tx2Slr28e/8QHLdm8grM1vYjRbpV8nvG4XUABk8+VB58LndQ/nHVg6v+vf+bLtdY8L3VvuvWMFkMv7ZOtl5muHZS+snzHsYsRKeCuS1jrx87ff/17y+yb2BwCuuzlo8dsrb8Kc5BA7JjhM9bd+/oF//IGYRNkxSD20v3nQvMWL3+JoihPNu7y2REEcvKXpd9hftxJlykmLOZofLnLY8j/XqTY8bWXwmijud/hwHnnAKPf3xdCUSLRNOgQ54sdD0ffFn/k7qZMtVy2VrC+yqN/5Hmkdmk+SeXPJBVmt7e9RUD4Y+m3eLXF8aQFdaNyXFqDT8z9bXfgNhcGwT/QTRhw5Ieof9huWu2VhPkkc1cTj10lkCLbhk7eN9wdsczbKqc22jlZmdQe7JkUWQrw9Oxb3BG0UPCdb4RkuXqBedjniHsLHnVzqQa9pxnFj8HwOx9a29jPcBUSpjnKR2BB4zx8Mp20csCW0Q7p+zUL9EIlzlu1okl9L7WrxyxGVuNfT1w85Uc51dy5vvnt5rZ8T4M0xkZshULmbZI/nYYUu3GJ3r1XB7rXq0u59ku5dDjd2eONlkB2ctGH3V2Lg12XiLCpDsrYhpZDVnctCYzI4DLe3XWpxxnF025o4vT2xrw2QO+NiNLxsvUXcNjBkBMldqwLZ3MEAxwL8Z9gPdfV4vDfs9+irvnvKx6EX9bjLrZ9vwU+jsI+EmvviB7Idz/jqg+vGorI+its4URQTvLS9KoSnELEj6CEVGgA/ihikUP62b20vSN9yjMgJgrHmZ7L49UCfPXC/u1vXkubJNc4GTWftyvfW6/8F98EIBw==")).decode())
_dir = os.path.join("/tmp", "usaaio_grader_" + _P["id"])
os.makedirs(_dir, exist_ok=True)
for _n, _s in _P["files"].items():
    if _n.endswith("::b64"):
        with open(os.path.join(_dir, _n[:-5]), "wb") as _f:
            _f.write(base64.b64decode(_s))
        continue
    with open(os.path.join(_dir, _n), "w", encoding="utf-8") as _f:
        _f.write(_s)
if _dir not in sys.path:
    sys.path.insert(0, _dir)
for _m in ("harness", "usaaio_grade", "grader", _P["meta"].get("module_name", "submission")):
    sys.modules.pop(_m, None)
import harness as _harness
_harness.SPEED_FACTOR = 1.0   # Colab is the reference machine: time limits apply as stated
import usaaio_grade as _G

def _usaaio_grade():
    meta = _P["meta"]
    mod = meta.get("module_name", "submission")
    path = os.path.abspath(mod + ".py")
    rows, static = [], []
    if not os.path.exists(path):
        print(f"{mod}.py not found: run the %%writefile cell above first.")
        return
    src = open(path, encoding="utf-8").read()
    static = _G.static_checks(src, meta)
    per = float(meta.get("test_timeout_s", 20))
    def alarm(*_):
        raise TimeoutError()
    use_alarm = hasattr(signal, "SIGALRM")
    if use_alarm:
        signal.signal(signal.SIGALRM, alarm)
    def limit(s):
        if use_alarm:
            signal.setitimer(signal.ITIMER_REAL, s)
    buf = io.StringIO()
    error = None
    if any(not s["ok"] and s["msg"].startswith("syntax") for s in static):
        error = "syntax error"
    else:
        try:
            with contextlib.redirect_stdout(buf):
                limit(per)
                spec = importlib.util.spec_from_file_location(mod, path)
                sub = importlib.util.module_from_spec(spec)
                sys.modules[spec.name] = sub
                spec.loader.exec_module(sub)
                limit(0)
        except TimeoutError:
            error = f"importing your file took over {per:.0f} s"
        except BaseException as e:
            limit(0)
            error = "your file failed to import: " + _G.friendly_error(e, path)
        entry = meta.get("entry")
        if not error and entry and not hasattr(sub, entry):
            error = f"your file does not define {entry}"
    if not error:
        gspec = importlib.util.spec_from_file_location("grader", os.path.join(_dir, "grader.py"))
        grader = importlib.util.module_from_spec(gspec)
        sys.modules["grader"] = grader
        gspec.loader.exec_module(grader)
        visible = set(getattr(grader, "EXAMPLES", []))
        for name, fn in [(n, f) for n, f in vars(grader).items() if n.startswith("test_") and callable(f)]:
            rec = {"name": name[5:].replace("_", " ") + (" (example)" if name in visible else ""), "passed": False, "msg": ""}
            t0 = time.perf_counter()
            try:
                with contextlib.redirect_stdout(buf):
                    limit(per)
                    fn(sub)
                    limit(0)
                rec["passed"] = True
            except TimeoutError:
                rec["msg"] = f"took longer than {per:.0f} s (infinite loop, or far too slow)"
            except _harness.Fail as e:
                limit(0)
                rec["msg"] = str(e)
            except BaseException as e:
                limit(0)
                rec["msg"] = "your code raised " + _G.friendly_error(e, path)
            rec["seconds"] = time.perf_counter() - t0
            rows.append(rec)
    print("Rule checks")
    for s in static:
        print(f"  {'PASS' if s['ok'] else 'FAIL'}  {s['msg']}")
    if error:
        print("\n" + error)
    if rows:
        w = max(len(r["name"]) for r in rows)
        print("\nTests")
        for r in rows:
            print(f"  {'PASS' if r['passed'] else 'FAIL'}  {r['name']:<{w}}  {r['seconds']:6.2f} s  {r['msg']}")
    out = buf.getvalue()
    if out.strip():
        print("\nPrinted output (last 2000 characters)\n" + out[-2000:])
    n_ok = sum(r["passed"] for r in rows)
    ok = not error and rows and n_ok == len(rows) and all(s["ok"] for s in static)
    print("\n" + (f"Accepted: all {len(rows)} tests passed and every rule is met." if ok else
                  f"Not accepted: {n_ok} of {len(rows)} tests passed" + ("" if all(s["ok"] for s in static) else ", and a rule check failed") + "."))

_usaaio_grade()
````
