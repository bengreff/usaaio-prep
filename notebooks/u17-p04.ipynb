````python
#@title Grader: run this cell to grade your solution { display-mode: "form" }
# The grading code is hidden on purpose (double-click to see it). It writes the site's test harness and this
# problem's grader to a temporary folder, imports your file the way the contest grader does, checks the import
# and ban rules, runs every test (including hidden ones) and prints a table.
import ast, base64, contextlib, importlib.util, io, json, os, signal, sys, time, traceback, zlib
_P = json.loads(zlib.decompress(base64.b64decode("eNq9XI132zaS/1dw7rsNmcqs7KTdXafOnZO6ae7y9ZL0be9JfiwkQjRWFMglSNuy1//7/WYAfkmy89G9S99aEgjMF2YGM4Ph3uzpZO9I7NUHf94vxo/3RmJvpSqJoZvtJ5WuMkVjJ6JSV5WYZ9JavdCqFLkRUjx/96uY1UmqKpqdKWtz04DIxgcMXJu6Uhajh9/jZ6IXCz2vs2qNkccYUKYq6fveQldxUapEzxnYKk/qTMVGrpgAW89WGrgBHw9lluWXKon1qsjLioBP9ky9KtZMdF7OzxmErPhznmeZmldYa+lnqfivNEm+om+6UmWV55ndO8OvmTRGkRwm/lec5XlBKBYys4rgK1vFlV6pvK5ievDDmEZl6unIimgm0zhfxJd5mTBKGjNpKVfWs6YybdIoy1NtKz2PS5WWquWufU5IYi9fosbmdTlnceSlTrWRGU0vZFkdYrAqa3U7IkFmLO+bvXNZGoCNChbwlP/7RWWFKq2ozmUlijKfZWolQFpCg06eIvh9UeYr4Zc3ow9/D6OpmZoTP11oK+a5qbCD39VGV/a7H+nj6XceKn4XOnn6nZsNIiLxshKJWmhAFSRFsaiN2xdB+5zwYPzQ0VbJpcI3NTXd3gunFeJ3DP0eidPfTl6/e3X6QWQQI7GkHFgCZoU9zy8N6en72ojgQlsNqsIn4gOBq0RZA626UOWaF4G1n6XOauyDKKW2StBPcamrcyg64FmZKnFZ6goMi0VeimfKHAkr1+KSyL2UFk9zk46EIagigT6Kc50kmK5NUVeWpOd2wXiRzvNi3f6g3Sb5+p+szwJQTUGj34jneSZnZHJWyFIJSyZQkqwMkfLAiteH4l2Zj4TNRZbPZUYQoUZWrGoIZaZIqBBRThao8TjTJIcZ+D/HVi7kHJYTEabDaEybK0Va0/4HDrMFrEztV+elkomwhcKGYVaZ1+l5toa2ZAsBWkDFPC9J0KUioVdCL4hA/GRj9Xx4AyKhfHh3evpT/PPJ849v34tjwk4MTw17G96G4PRqrgrSlPBoagT+FXjkZkGhIHGVBNDFZCRWNm3mAK/JScgm8SP0r9vcgOZ2QGJTBFfNWjil3ppmh8iz9EYX4F8bW0kzV8HVyD2PPipj8zLsLWe0qqpLI66iBN52fh6E0byo8Ze3OQjdZMVsipeM7rQs87IHxbHcg2WKSFpZlnINujs+7LksVJDm1QhKaegv9PN4Cr9n4Xenew1lFrKu6iJTAXGO+WHESz0tLFPMaSYRrHAkFtO9GwJ4iw23Dpe4sfA76qqAk4VO3PTm3wJdR9k8y+19lI0EHHF2fKD2fxgJ6b/+paEXlnUJkltipa3WYHSR5ZIIo3HGOXzQYyZ1/BFLl+7rHew4Bvz0cBdvzSNmr0WADdEW7g0nSpCCjiwLQgFPQVrYf3bpn/XRkyeVmj2hoTXaLFrgXpFpv7PMCxHC8OKiP15c9KeveKosITEWBBG3klf0IWc2SMW+uAzDkGCDUX2thMLxJsZkfDtMZSgndnTiQmbwDyLIZJmS21WksOIGH0fRo/Q2HOy9+kcts8/QyqFRualEtElY00chSxREyKoq/fPpHm9HB2SHuvQUpJvzf64WfdVg+mMvBxC3ZUx9qbYQaFeONimGMhC1zHwHA874Bo/+rRyQRrMwNNgMbWM2Ed4O+mbv3A9G2NuRbskW/TQKUXjAYRRz9BbHA3IQy/DOeawPmkkPGqhDSmNrZIGTvPr/dc1ZbtQXOmWgbLTyilSSz/a9EEdiwtY7pCegOHfEgUvYp2fI3IA6AhgkShXHHxHp9RTME/gR4tskbxtAODhCaCgimPzsaiB4bEsgR2L29XKXnyV3N8XZBeP7WrFLEjvDsUNX4BFJh8QGs3CXr8Hyzs/sXN76iZn/yns7tGziYOS8HTTb9PeqgcMQesFLHq9qGAWim2BhRuIh3Cks4+HD5SV9ayhxoeNznABiYRjzggJU4gExDqLAFOYlDULGMq1XCMpxhIgik3MVNVEnn/pkT/CLk86yZMjxrKQFhPHMzUR0RoeHCTYJco9piYYig2Eb0lKFUEaVEufbtS4Ct4ZxDBSc/UmrXN6JrJHYQBcTRQE+5ZeJC5k7Xr650eJbcXD7RCCpWrrsk3RW0PYhHm39pRcyiO8kTEG8NoFVFAqCqs+V8pO+iOFd+yH3jkias1KchxaxNzggAXGs+6BdieDd5VCE6i1AlJSCcWJILHEc/lrOxRwIaAmYdFE3ZSsc2FNwgOxi32WFNMwxOMJyTQlUXcCcRHWp54oAQkIaJ3S+8BYMmhuJWlYhGlkgXqLjmyBgRY1nkfhID3SJYTCTPSAUdC44ZCRhlQzUqhpTNAkIEbLLRcxgVNl4m0+oUlLtXowYpfLhiOf3WPhNFN+JftrQqWRMqliSNQSHfb3DJgLNj8ce0oYrmiGnWXZD8nAklofANvSPTPJoY3DAyadFwXSSIA5ZDIe9YRYDMrYgoSP2DnH0480+Q2xHVZ6z3iD5T6qj6HBxiygfWqae+KyrVWJWJpgdAxK1oXz+xovGr+McEolds8U7/jFKZJZIgxdQrZQCWJdk3vh9YkBkHfcaaGlSWKdKmg3rEhxXpoG0FxLqF7cTp1wpqa2UOo+5wjCocrygEVgAKOuKB5qdxrnOErK6OVVGkDBrsgWa+V8f3r4BrYYtx1YJKGRbLdbVOUb2X4oGj/jRlzjiRJdPxY8dDjzEb1jlP3mserqV8sPYeum/obJapmftkPvASFRXOuuG8/br323eFQkKWZ33l1udGtkts2u7UV5ofpQ4F2ZyvmxHoMWGigWu1mDnGmwa1bo7DCx1tZ8pTBMH0Q/RgY/p7BHViVZU+xuxS0F24Q+fxvkVUkPK6dQ0OKKFzqDRzc9gugey81JRBugrLYhEo4e/mqWhGo7NM6qo5FwBAEJd0K5FD32k+Mvp+1NYjpdF9A6fQRxTGSyOwwiOi5YjqitkCcqmBlKJaHIEbVVlFYxhDAgfCEwIgLtqX1wLEt/grP6HPBKnj8eHXAl79beT//kQPzt58+b0J1BwM92LfU0yjokZMNbsp/uprtTcf0Oc777N81UBWqd7t1Pz6u3bd/Gbtz+dfgC4ALoS/ZzDVunL36C4yn09sWszbx+8Qgz5HDDcrw+q9+MnBJndrxfK0Pmcl6dXRdgvsXx0tZhPVVliiUxvFZCO1Sv4GxxbqrVYThUbQP1qBEU283h+ruZLG9hyTltcyeF5+97ZO0XDOKxIjvnSCWdlUwjmCdwiRYCuxsdmY6sH7e5wMORLiXRQVaV2ZcXBEeUOocnZrmi2KhUCO5YStAT5NQgdxqEf1sjPrzgOpaKc2o4PJ57sI2gLV4o99UfkJi0v9ylybpyfuVERfZr89oi+YzaSn1tPoK9x05EAeUVwzLCTjcI3yWhy5gmdyXacz8mqOXK4HBrXloH1xmewh9jVSv04qThCRtiuTmAqXOl3gYIDDAhufkAZDQXdekbVfbCz5NhNZREptkfAlfOezFeyWtVZf2TB1RG4avLNEP6lzJYBbcbG0d0L02m602eXGmxmFW0oSxMjJvdo+wDzJRXEy4HkSRSUr6i04H5GwAZBUuQYrBB5wh1RMkfAV0ys24jwaPfZ2NuLSCaJxxFuT+72pjcvskWmabsJ52T/4GzHwm7zorpIKOresbafJGb3i/FneL1NbrCh2CuWoy+8g/3OoHbIkld4QeL7H5DipgQBbUMKPdENJLBLAzaW3iW8r1rKfH6p0E8a29nkfUMhmA4ytM+C+qbnke8DqJO7udo9aRfG7rDaxMmGH8miUCYJmoqPWzTdcy7Puff/GMprFxY+BZ9p89afYid1emLpDApdbWVjQZT7ea9l9Roh45ZOOx/0hcSBsJ5C9kBSkOhB3en+myjAWYlTeRplg6B50d9zSk/ZwQY9PGG427t8SycKBNOAunkwEg8ckMaiyAAfAB/y3vQBlWJvd9cTd9FPJYsd5NuG9g7YN+0N4kwaS7Kdn/sTgkPB7oDg8rOSSGJ/48/wictGs0u5tvvu0hWZDOJC3QIipfTQcpOtRw1OZC7dURM4LXBFJB9KUdCxFplKkZpfUtgkeGYURWHvMENQJpKcNsUBPtecX7pdCHqH5p/4cOydwo5cf/hCDf8pgt5Z+icxiAt7KgQMX6w6oMAKLx+CH9hwh+4AcvhHdtjkDY72SrbbZVA+ZD52RhTnhYsnh9U2mu6MbMP0PodhyiihF/8pGtiUsvbsRgQucvIYJuOzvnLvYP9LRNBhvY95dm5bLPPo1zHswiRW9gGjGCYzJqUu1bkynMS2sSOv+tfy7zm7vSdTX5QaYLJ1zJFsABBIdGPKpoYR/Vuj9pnORNl5qTmrEAVUFfadCn8TzRN0c31NCRvMVdE8y5f1wwrTjMoqTd4aIW/G9yquZoGK4rh9gJTPB5sEHJHmgg/1BeEBCOzUgtJP1cR8lPm1LPio9JJKJlSwIi/LtUmvdPhL4Vgvah8M0VawwtBsvk7ryPfCpBsTviFRvfsRiv5vbxjrLc1vhL1CJB60uRg87WaWS5msLNOLSRsiNqzcOfPwrJcSe0lRAH4s2imPmnAdeo9xKjhEWS4TGwRMxHekNmQTXIrYI3g4s6iSETTODukTQfTE9Cc00uBq4rHYtguXME59B86U221s99MVZpzK+kyaNJEG3uTYgj78SQftjMi5IxVtDZ0iWDJBO3FUnfFpQb/YSs764WyT0jXxrOW64xbavpfgqkXA4qTmEBu46f1T3mlJU/JcgGidRx8qKp+8fNsIzxV5IvcR+F8fXr44efX+9chn6X4q3FnMxd3mFrjny4Z9TCTLw3FDyzA75iaYrlgVUZdWqWB8bjMCULoZZjU0qkoTirIh8+XHl69P38fvT09ejVrqNsIcWyjSnmEpLKLRmGozrrJD1Um+Mulx1OsZc6rRFeawTT1XtYGvnm2j87AYIaEO6M/mQliMm2cn9JgThjOn+NsssQ2pMqIaUOyWkecJv0Jw42FtwpddBiUJr4jeOM6cN3M8kgdmr8Yet8rzJd8piJtmP46iMVWAG9flsTyTVrUVIlcD6Upiz16djscHHQFfxMZdFPuLISaTbmKoQyz3G+WDoHvOo9ast0DvElTjVwgv9Jk0im/iAyTgj8fj8dHZVxsytzgO6zg81O/vcHOau+LmRhO8jNyj8NOb28kqyZULmlyzHU4XAnHbz9u/kIX0y2xyuuea/sgK2yOj7QNs+fZ9hJ+2vbRnfOldtuSg+Vm+3c8Xtposz00hz9C0DjZBvHd7dNZQ0BAgUVk4x07fyLdfyNI2OCJdqZUNuHvGDE8F9qrNvT9doEiQESzCfsELXoIuITkkIYwbwTMfx6T9ZQ2/5ZSCohXuIjANaxsOl9yzNrXq79/cHbDOHx4xkMn3R2fw33wvDGp9lVqwe5zuedDN5B42ekrVYEp7dqUp3t+6Sx8aGkfj2y+5E9tufPiKU+ePnjz+cm7bK3+VS/ObMOkkR5ZKgfd278aWB+8Wu9jD2Tg76yw3aXMvs+mzRaCN6zLjaH5ECcRClqK5GQz7TsAj58uNjZr2H2S4pZnC6x29Kl90lvyraOn3GfB9RfKZZ0gHq1VwgldCi5PgrovakXi063jz4exZk5CVrWP74qOoXeBi1WNKHYMudr0vKmWnMsvzLNgizD0jUNVAcwlcNQDXrmgSlDuPFG5rXogm4XHeLY4pv6GLsqPmeoDSHXez2x4W3Z2ufy/hSMy0kTgs6ZpRc3NI81qCO33oBolaj9Ulcy/Jz15gGrY7kyatZUo7TD3KCTU/y/m8RvK4jsQblToA3Lc/NYtMF5ynGnr3oUNHj/n+vDaaejd80wTlF1mGyTC3cfTDn1mONJcvQGGZM55tw6apvFRpjVhdXxNlmtZNDRe28ETOMWNGV1bj6K+Hkei9HcF998CEVbmZKxEU9SzTc0bn+8055G9kUOVQoHO+8j+vFwusg69nEt6tP1LZTNCFMt2Au9ZuV5nr+j3moMW3s8xyrOJtjzaute9rXh+2it11vTriVoTRRlvH1LRN/jiW/fHqGGa3B/WbmlcvX7/8iOePcOjs7FP4pm0bybtr6efvfn3iWm4suUvfveNgDbtLmlaIqfl48v7F6ceRcJ/xu1+fvXr5HJixSTD4cfSXvxLJL96+/Yn9TaFTNRPXeaHhhC8klW5mIGVV0+ZIMLzUIoFLSpUFGUt8m8HQRZpfiKVe0CwjrmUhLtRFbsVMFRLn8UpfC0gAK3WaT/f83QPs5tmJw3otEyhXDQTX9bW4Vheqwl+6NKDaFlyKVbChSiyJEKtSJVIJJa1NjhB1pUWRX+diAdW8wJ5lUMHrPMH3TK/wzYpUXwzQ/vzy1Su6rQa7CnQRuzW9IQFGSm2Jy0TDtgG9qJf055peH1CYgL8JdWhciETOlLjWMymWkm4E8QtsatBaIB6CqYFemYELC3Nb5lDpQVsK1+01dcRlOQRGlXZTp4iDLzRUkSxqhdUXtA8GJ8i1WKgMnBQwNgtOaWcu6gSYF3WpCM2C3/G4IBu1gJuRRC6ox+VCLTcwV2B6BVQrHKJWl9D5JZRsBhDXoiJS6hJrIW5bI0wF5us6we5e1yRMiKjGJlyoAgEXvXoC7wkxL0gLDBavYLtUYWmF/eb0xQk00luDqV3ctczZDpq6Ueou+hVFsZaNqoiNSo/H0SP6ilg9878PYHU5TkJ8HT8eVvBOyQU5H3IkDijxerThA+HS2E8Z8pvwEOxTKNyXMw1y1w4rdRZrzCJvCoOv3PnymAAejClfyZAnuLKbgykho0uYI3YdHm8OFXF+2uGhrgcHsKJSId0lcK5DFx3wa675C7uBwIcOEPEKmpWBfi4skmMrcpi7vvAxWHGOHaMLhrqCz5pxWKUcJsyhviALJ0FNdFSfhQMu5aV5Ih7/O50yGcF2T4i7ghowrRPooHSZ0i3qfb1MLve4quyogXnM5a5JW5CktC5TJuBZofhRmF6wtqQUCmdvGs3Pc43AfoJ9PUQAAhDF8WQcPSbf9Ij/nPVTvPl5bZaWunTJuhzOXv//Rj/dcjPcLnJadODeIXDcBUTbOPrelT/3N6I4asgn5zhJI9ALz4OUiriisTA8I0AE86kYu/XwaJtTMRSGZ1sX0AP8rHc7UgPHbhN5TRpb6qM4DCGCy10X7o7ZfXwMn+2ot+/AtQXSifzbYzEACE78XuBM+WSGt7E/TgVaVrDpBwdhuNUfAS7IZW8Klsa2JPu5Eluc7diExs+4vZws+iJg6wQlKYXObfsz0eFQ9rV07dXbSebpIMzfxMrGt8HxmpV0X6x72SaZUcMW57/uHq1ybc0kU0dhG/U6siZ6YD/OWBs46+G9ycCgnXOOn588/4V61W5ue21VSPM55ezegMFRWlLUTYjdmu12Izc+aaa2bU0xve1QxsR0ew78gKwB2jA+8BRKmiSHcw79nKYjdUZzZrvnNJnNb5R0xFfIoTDbd1n5jd3t7w5wymzt+G9t+YXjT34vja9zEbfG8MqxH2/KBe6h3Xh4sLN0/enCg2sLg/yjXowdUCtaQMIM3Usa9H2N75PfoAIbSnK22cN7fytzL+vefC1xMOG/1XrztYfeC1nT/gvTTTrbLsFpyeNZni8RQvE1NmIrtXbv1vo7Ve1ORJrZvW/kkH92cj58R2wHTTuvu1xZYHIEhTrr3pZa0CvN2WD7dm61/ww3WsN3he73dUv3ieXCCjdNH/SaphG18buzVFGx5ySvtkd6POiRDm4YeTPcZBnh8P1AlgT0LaQNIK3aIqPprfdSo7md3J50b5C8cz3JTcMk9TAPUZFdMSYcJM7G7sHVTL5tog/S7xu37HaQTA6RcJGA6LxgjkAFx3YXJCjqqj1wFVT+TfBd3yu9cc14mveSO340vdA2pprVQYuJyHX+RK2Kah003CSE+ZheqzTVD497kyfOKLGGb65peNNXUrrCU0kBx2ej9iWeQLKV87NDLrX0ns34WVIN3fy2H24c+2au2nfyhRyJNf5XzPA5Gzm/0Z4E3kvP5+29XVDwa0NrGXL7THvR6oou2FlfBOBSYFvTEDf4dhQ9XtyOxD3avrGrQPv0eJjisvJ8AgWpAHY1vxQ3g7XDF/pYKK5M8S8QyoyFMtsQyjYbTH+vOHI//fx/JEAvtVYNK46Hvdvb/wVsaqn5")).decode())
_dir = os.path.join("/tmp", "usaaio_grader_" + _P["id"])
os.makedirs(_dir, exist_ok=True)
for _n, _s in _P["files"].items():
    if _n.endswith("::b64"):
        with open(os.path.join(_dir, _n[:-5]), "wb") as _f:
            _f.write(base64.b64decode(_s))
        continue
    with open(os.path.join(_dir, _n), "w", encoding="utf-8") as _f:
        _f.write(_s)
if _dir not in sys.path:
    sys.path.insert(0, _dir)
for _m in ("harness", "usaaio_grade", "grader", _P["meta"].get("module_name", "submission")):
    sys.modules.pop(_m, None)
import harness as _harness
_harness.SPEED_FACTOR = 1.0   # Colab is the reference machine: time limits apply as stated
import usaaio_grade as _G

def _usaaio_grade():
    meta = _P["meta"]
    mod = meta.get("module_name", "submission")
    path = os.path.abspath(mod + ".py")
    rows, static = [], []
    if not os.path.exists(path):
        print(f"{mod}.py not found: run the %%writefile cell above first.")
        return
    src = open(path, encoding="utf-8").read()
    static = _G.static_checks(src, meta)
    per = float(meta.get("test_timeout_s", 20))
    def alarm(*_):
        raise TimeoutError()
    use_alarm = hasattr(signal, "SIGALRM")
    if use_alarm:
        signal.signal(signal.SIGALRM, alarm)
    def limit(s):
        if use_alarm:
            signal.setitimer(signal.ITIMER_REAL, s)
    buf = io.StringIO()
    error = None
    if any(not s["ok"] and s["msg"].startswith("syntax") for s in static):
        error = "syntax error"
    else:
        try:
            with contextlib.redirect_stdout(buf):
                limit(per)
                spec = importlib.util.spec_from_file_location(mod, path)
                sub = importlib.util.module_from_spec(spec)
                sys.modules[spec.name] = sub
                spec.loader.exec_module(sub)
                limit(0)
        except TimeoutError:
            error = f"importing your file took over {per:.0f} s"
        except BaseException as e:
            limit(0)
            error = "your file failed to import: " + _G.friendly_error(e, path)
        entry = meta.get("entry")
        if not error and entry and not hasattr(sub, entry):
            error = f"your file does not define {entry}"
    if not error:
        gspec = importlib.util.spec_from_file_location("grader", os.path.join(_dir, "grader.py"))
        grader = importlib.util.module_from_spec(gspec)
        sys.modules["grader"] = grader
        gspec.loader.exec_module(grader)
        visible = set(getattr(grader, "EXAMPLES", []))
        for name, fn in [(n, f) for n, f in vars(grader).items() if n.startswith("test_") and callable(f)]:
            rec = {"name": name[5:].replace("_", " ") + (" (example)" if name in visible else ""), "passed": False, "msg": ""}
            t0 = time.perf_counter()
            try:
                with contextlib.redirect_stdout(buf):
                    limit(per)
                    fn(sub)
                    limit(0)
                rec["passed"] = True
            except TimeoutError:
                rec["msg"] = f"took longer than {per:.0f} s (infinite loop, or far too slow)"
            except _harness.Fail as e:
                limit(0)
                rec["msg"] = str(e)
            except BaseException as e:
                limit(0)
                rec["msg"] = "your code raised " + _G.friendly_error(e, path)
            rec["seconds"] = time.perf_counter() - t0
            rows.append(rec)
    print("Rule checks")
    for s in static:
        print(f"  {'PASS' if s['ok'] else 'FAIL'}  {s['msg']}")
    if error:
        print("\n" + error)
    if rows:
        w = max(len(r["name"]) for r in rows)
        print("\nTests")
        for r in rows:
            print(f"  {'PASS' if r['passed'] else 'FAIL'}  {r['name']:<{w}}  {r['seconds']:6.2f} s  {r['msg']}")
    out = buf.getvalue()
    if out.strip():
        print("\nPrinted output (last 2000 characters)\n" + out[-2000:])
    n_ok = sum(r["passed"] for r in rows)
    ok = not error and rows and n_ok == len(rows) and all(s["ok"] for s in static)
    print("\n" + (f"Accepted: all {len(rows)} tests passed and every rule is met." if ok else
                  f"Not accepted: {n_ok} of {len(rows)} tests passed" + ("" if all(s["ok"] for s in static) else ", and a rule check failed") + "."))

_usaaio_grade()
````
