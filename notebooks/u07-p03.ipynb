# Logistic regression by gradient descent, with a stable loss

Logistic regression predicts $p_i = \sigma(z_i)$ with $z_i = \sum_j X_{i,j} w_j + b$ and $\sigma(z) = 1/(1 + e^{-z})$, for labels $y_i \in \{0, 1\}$. Train it by full-batch gradient descent on the mean log-loss

$$L(w, b) = -\frac{1}{n}\sum_{i=0}^{n-1}\Big[y_i \ln p_i + (1 - y_i)\ln(1 - p_i)\Big].$$

Your loss values must be correct and finite even when some $|z_i|$ are in the hundreds, where $p_i$ rounds to exactly 0 or 1.

**Initialization.** Store `learning_rate` and `steps` under those names. Set `self.w = None`, `self.b = 0.0`, `self.losses = []`.

**Training.** `fit` gets `X` of shape `(n, d)` and `y` of shape `(n,)` (0s and 1s, int or float). At the start of every call reset `self.w` to `np.zeros(d)`, `self.b` to `0.0` and `self.losses` to an empty list. Then repeat `steps` times: append the loss $L$ at the current `w`, `b` to `self.losses` (as a float), compute both gradients at the current `w`, `b`, and update both with `learning_rate`. So `self.losses[0]` is the loss at zero, $\ln 2$, and `len(self.losses) == steps`. Return `self`.

**Prediction.** `predict_proba(X)` returns the probabilities $p$ as a float array of shape `(m,)` with no `nan`. `predict(X)` returns an integer array of shape `(m,)` with 1 where $p \ge 0.5$ and 0 elsewhere. Neither method changes the parameters, and no method modifies its inputs.

**Example.**
```python
X = np.array([[1.0, 2.0], [2.0, -1.0], [-1.0, 0.0], [0.0, 1.0]]); y = np.array([1, 1, 0, 0])
model = LogisticRegressionGD(learning_rate=0.5, steps=2).fit(X, y)
# model.w = [0.45384104, -0.00023671], model.b = -0.01530742
# model.losses = [0.69314718, 0.57977575]
model.predict_proba(np.array([[1.0, 1.0], [-2.0, 0.0]]))   # [0.60785315, 0.28434915]
model.predict(np.array([[1.0, 1.0], [-2.0, 0.0]]))         # [1, 0]
```

**Restrictions.** Only `import numpy as np`. You may not use `np.linalg`, SciPy, or `expit`.

````python
#@title Grader: run this cell to grade your solution { display-mode: "form" }
# The grading code is hidden on purpose (double-click to see it). It writes the site's test harness and this
# problem's grader to a temporary folder, imports your file the way the contest grader does, checks the import
# and ban rules, runs every test (including hidden ones) and prints a table.
import ast, base64, contextlib, importlib.util, io, json, os, signal, sys, time, traceback, zlib
_P = json.loads(zlib.decompress(base64.b64decode("eNq9Wwtz2za2/iu47syGTGVGspN0x6lz67RuNzt5TZKd9R3bw1IiRGNNkVyCiix7vb/9fucAIEGJdh7de93GFkHg4LxfgG52VLpzIHaW4x92q/H+zkjsLGSTYOhm+02jmlzS2KsyU7pRM1HLrJZaq7IQ07XI6iRVsmhEKvUMf0dipZoLkQjdJNNcirzUmuDkWFIWDng+fsLbqmLZSI3RyZ/xmKr5XM2WebPGyD4GAK9e+5u/b/f+7RcGUKbLXMZFsmAcq7rEnov9WC+nC8XzaFaS5+VKprFaVGXd0HanO8VyUa13zvF2mhSFTHkwV0WSZ7REzxReEwpXlWrcvDgvy4rWz5NcS+KO1E3cqIUsl01ML/bHNJpkZpO8BLpZlKXMAfPgGAIEZK1mOiJGqVw1vF21bi7KIprlidZgDO2ry2U9Y/LKWmWEIU9M6mYPg029lLd4nqucGXmzc5HUBXgUVcy5M/7vLzKvZK1Fc5E0wnKJRUeDhi0i+H1elwthl7vRh7+H0VlxVhzZ6UJpMSuLBqJ5tCxUox/9SH+eP7JQ8Vyp9PkjMxtIROIlKcdcAaogfon5spg1EI0WJLeUB+OHBrcmuZT4JM+KToTCSFn8jqHfI3F8cvT63avjDyKHShBJ0oAlYFroi3JVCCx6vyxE8ElpBazCZ+IDgWtEvcS28pOs17wIpP2aqHwJpRJ1orQU9OhUGPB0kkmxqlUDgsW8rMULWRwInazFitBdJRpvyyIbiYKgihRqJS5UmmK6Kqplo4l7RgqFZemsrNbtAykP8dc+sloKQC0qGv1O/FzmyVT8/O5vWiS1FJo0uSZeFYTKAy1e74l3dTkSuoStzZKcIKoi02KxBFOmkpgKFpVkWgqvc0V8gOU2FxDlPJk1ZR3RTnvRmISbiGxJ8g/MzhqwcrnbXNQySYWuJASGWXW5zC7yNbQlnwvgAixmZU2MriUxvRFqTgjikW3O0mFNhZjy4d3x8S/xr0c/f3z7XhzS7kTwWcGqz2IIjq9msiJNCQ/OCoGfCq/MLCgUOC7TALqYjsRCZ24O9i1KYnKR2hH66YQb0NwOSFxUwZVbC2/jrXESKuvZhTc6B/2qgNUWMxlcjcz76KMsdFmH3nLeVjbLuhBXUQoHO7sIwmhWLfGbxRyEZrJkMsVL3u64rsvag2JI9mAVVZTopK6TNfDu6NAXSSWDrCQHnLAbhn4enu1As+FQz3YcZhq8bpZVLgOiHPPDiJdaXJinmOMmEaxwJOZnOzcE8BYC12YvcaPhd+Ae5ayBTtx482+xXYfZDB7vPsxGom7K/HAid5+ORGI//tnhC8taAeUW2UQ3axA6z8uEEKNx3rP/wiMmM/QRSSvz8Q5yDAF2ejhEm3vF5LUbQCBKw72pBquBR54HoYCnIC30363sO3978qSJYk9Y0BpVzFvgVpFJ3nlumQhmWHbRL8su+uUrnqxrcIwZQcgtkiv6k0x1kIldsQrDkGCDUHUthUQgE2MyvgFT6fOJHZ34lOTwDyLIkzojtytJYcUN/hxE+9lt2JO9/Ocyyb9AK/tGZaYS0kXKmj4KmaNAImma2r4/22FxdEAG1MVTkG7O/7la+KrB+MeWD0Buy5h8rrYQSCoHmxhDGQhbJr6DAWd8g1f/VfdQo1kY6glD6ZhNhMVBn/Sd8uANPYl0S7bwp1GwwgIOo5izsTjuoZPB/ZHk7K4P3KQHDmof01gXSYVI3vz/uua8LORXOmVs6bTyilSSY/tOiJCYsvX28QlSNQMjKXEJfXz6xPWwI4BBKmV1+BGZnqdgFsGPYN8metsAwl4IoaGIYPK7qx7jIZYgGYnpt/M9+SK+mynGLni/b2V7QmxnOLrvCuxGidlEB9NwyNdgeednBpe3fmJqP7Js+5ZNFIyMt4NmF76sHByG4CUvZbxYwiiQ3QTzYiQewp3CMh4+vFzRJ4eJSR1/RgQQ84J3nlOCSjQgx0EWmMG8kgIpY52hoEAdpgpR5clMRi7r5KhP9gS/eNpZVhJyPpvQAtrx3MxEdkbBowg2ETKvaYmCIoNgHdJSyXVMgvh2rarArOE9egrO/qRVLutE1ihsoIuppARfzRVllpQyd7R8d6PE92Jy+0ysyvqSsvqElVeQ+JCPtv7SMhnIdxymJF4VgZaUCgKrL+XyM5/F8K5+yj2QSXNxiXiokXuDAmIQ57oP2pVI3k0NRVu9BYiaSjAUSAtNJHEe/jqZiRk2oCVUPHPWTdUKJ/aUHKC62J0uU3hSGuYcHGm5ogJqWcGcRLNSM0kAwSGFCF3OrQUDZ8dRzSpEI3PkSxS+CQJWLPEuEh/phaoxDGLyB7QFxQWzGXFYpj21asaUTQJChOpyHjMYWTtv8xlVSpvhxchRGpuOWHoPhRWieCT8sqFTyZhUsSZrCPZ8vYMQsc2PhxbShiuaoqa57IaSvZG43MNuff/IKI82BnuUfJ4VjCcxYo/ZsOcNMxtQsQUphdg72OHnmz5BbEdNWbLeoPhPm4Nob36LLB9aJp/ZqqtVYlYmmB0DEsuC6vkbyxq7jmtIFHZOxAM/vCUqS5TBc6hWRgmsKTJvrJwYEFnHvQZaFxmsU6ZOYF2BA0mm5QLcnidQv7ideMbtnKVOElXG3GHodTl+oxFYADDrmgeKncaFylOyuhl1RlAwK7IFmvnXD2/fANeCLUc3KTBkWzWNGLH7Urh9xI+2xRGnqn4ufuz2wEs8wyr/xWPN862SH8bmlf8Q7FWTq2k7ZP5gJFo2Ku+Gy/bjP3TZNQmqpLnwl2uVFUm3TK/1RnvBPdSIC9NkdtmOQIsLahaYXgM3vaAbrbvDwKVqdnOJaWISPY0mNqfTB9QnWlBTb8QuBdWFDT7O+VWJApezs8LtEc1VDo12j8HZDtAua0kVoO20IBONHv6tuCyoh6PLnDoqJXcAsKGqSGrRQ5sp/uX4/TEsx/Iieoe/QRxTGyyOwwiOi5Yjq6uSGpidFeBKRJMjaKusm2AMY0D6QGBCABzqfXEvSHyHWP3P5EAcPx7vcSfs1d+P/udD/OLozZvjX4DBzdlObFuLcUzEgDAnT/Mor+TMfkKebz7NykUFXM92bs+KV2/fvovfvP3l+APABdCV6NcStkof/g7FlebjkV4Xs/bFK+SQPwOGefogvYdfkGR2T7/JguJzWR9fVaHfYvloejGf67LECSq9RUA6tlzA3yBsydZiuVR0gPxuBGU2s3h2IWeXOtD1jETcJP14+97YO2XDCFbEx/LSMGehMzDmGdwiZYCmx8dmo5sHrXQ4GbKtRApUTa1MW7EXokwQOj0fymabWiKxYy5BS1BfA9F+Hvphjfr8ivNQasrJ7fzw1KJ9AG3hnrDF/oDcpObltkQuC+NnbmREf4vy9oA+YzaKn1uLoG1VU0gAvyI4ZtjJRv+aeHR6bhGdJu04x8nGhRxuh8ZLzcC88SnsITa9UjtOKo6UEbarUpgKZWK1SRQMYEAw8wOqaCjpVlNq24OcS87dZB6RYtsNuEfu8XyRNItl7o/MuTsCV02+GcxfJfllQMLYCN1emk7TjT6b0mCzqmhTWZoYMboH2wHMtlSQLwcJT6KkfEGtBfNIvXgwkjLHYIHME+6IijkCvmBkjSDCg+HY6MkiStLU7hFuT+5k482LdJUrEjftebo7OR9Y2AkvWlYpZd0Da/0iMb+fjb/C621SA4FCVsxH23gH+Z1BDfCSV1hG4vMf4OImBwFtgwse63ocGNKAjaV3Me+bljKdX8v0I2c7m7RvKATjQYb2RVDfeB75PoAqvZuq4UlDO3bBanNPNvwoqSpZpIHr+JhFZzvG5Rn3/t99fg3twlHwhSre2ih2tMyONMWg0PRWNhZEpZ33OmleI2Xc0mnjg74SOSDmKaQHkpJEC+pO9++yAGMlRuVplA2C5kX/KKk8ZQcbePuE4bB3+Z4iChjjQN08GIkHBoizKDLAB9gPdW/2gFqxt8P9xCH8qWUxgL52uHfAvmtPEKdJoYm3swsbITgV7AIEt59lgiL2hP+Gz0w1mq+Std41B6+oZJAXqhYQKaWFVhb5euT2ROXShZrAaIFpItlUipKOtchlhtJ8RWmT4JlRFIVeMENSJtKShGIAXyiuL40UAi9o/omDoxeFDbo2+EIN/yUCL5b+SfTyQk+FsMNXqw4w0MLyh+AHOhzQHUAO/4iEi9Lt0R7JdlIG5n3iY2NEcVmZfLLfbaPpxsg2TO9LCKaKEnrxk3CwqWT17EYEJnOyO5yOz33lHiD/a1jQ7Xof8ezctkjm0W8j2KRJrOw9QjFMZkxKXcsLWXAR2+aOvOo/S7+l7PaeSn1eK4DJ1zFnsgFAoNCNqZrqZ/RvC7nLeNJlkFpxVSEqqCrsOxP2JJonKHd8TQUbzFXSPM2H9f0O05TaKq5ujVA343MTN9NARnHcvkDJZ5NNAo5Mc85BfU77AAQkNafyU7qcjyq/lgSbla6oZUINK/Ky3Ju0SofflI55WXtviETBCkOz+TitQ98yk05M+IREeucjlP3f3vCutzTfMXuBTDxoazF42s0qlyrZpM4+nbYpoiPlzpl7515JbDlFCfihaKfsu3Qdeo9xajhEeZmkOggYiUekNmQT3IrYIXiIWdTJCJyzQ/lEEC0y/gTHDe4mHoptuzAFIw2dntMzlXq6ezSNGaOytpImTaSBNyVE4MM/7aCdEzp3lKKtoVMGSyaoTw1W5xwt6Imt5NxPZ11J5/JZzX3HrW19L8Fdi4DZSZdDdGCm+1HeaIlrec6BtCqjDw21T16+dcwzTZ7I/Ans04eXvx29ev96ZKt0OxXuLObmrjsF9nxZ/8YS8XJv7HDpV8d8CaZrVkGcqaoljM8IIwCmm2mWw1E2iraoHZovP758ffw+fn989GrUYreR5uhKkvb0W2ERjcbUmzGdHepO8pGJR5F3B8yoRteYg5g8V7Wx33K6vZ2FxRvS1gH92lwIizHz9Cm95oLh3Cj+NklsQ7KOqAcUm2XkecJvYNy435uwbZdeS8IqojWOc+PNDI3kgdmrscdtyvKSzxTEjZPHQTSmDrBzXXaXF4mWbYfI9EC6ltiLV8fj8aRD4KvIuAtjezDEaNJJDN0QK62gbBJ0TzxqzXoL9BCjnF+hfaHPpFF8Eh+gAH88Ho8Pzr/ZkPnuYr+Pw0P+/Q4zx50VuxNN0DIyr8LPC7fjVVpKkzSZy3aILgTi1q/bv5KE7Ots8mzHXPojK2xDRnsPsKXb3iP8vO1lnvFld9mSgWZn2et+trHlqjwzhTyDuzroknjr9ijWUNIQoFCZG8dOn8i3f0pq7faIVCMXOuDbM0U/KrBXdef+dIBCd1+Deeg3vOAl6BCSUxLacSN55nBM2l8v4beMUlC2wrcICkfahsMl96yKpfTlNzMB1vjDAwZy+uTgHP6bz4WBre1SC3aPZzsWtJvs7UZvqRtMZc9QmWL9rTn0oaFxNL79mjOx7YsP3xB1/mjksYdz2175m1yaFcJpxzmyVEq8t+9ubHnwbrHJPYyNs7POyyJz5zKbPlsEqjC3zDibH1EBMU9q4U4GQ98J2M35cGOjp/0HCW5xpvR64K7KV8WS/xQu/j0DPq9IvzCGdLBaBSd4NbQ4De46qB2J/aHwZtPZc1eQ1a1j++pQ1C4wueohlY5Bl7vel5WyU5mWZR5sIWbeEaimp7kErumBa1e4AuXOkMLXmufCFTzGu8Ux1Td0UHbgjgeo3DEnu22woDPdrSPKoSvSdx3bjfiIe2Qun46EveBnb8CN/Ns3I3tRZETn0Qbp9oY5YoL17fIqWVTkI8+9Qy6VBdfOGZlzpqKK5KJq1jEdjQTXI5HSfof+pVSUuZh4LZ4finG79BSjJMsJYmYwgXYSoKsq2L3mNy5SyWu7B15dn/6bX3Uw/m2BYJaFIq/vO36Xc5Se1EE9GYn1SOR0PaCRVXsrBiyhk4ATc+3JFsl80Ym+w8DnR8DlWtalDlJEEjj/Ue+Mx7uV0QNMP0RJcCIeilUY6eUiSK6UPpyEQNpLos0+zmhsM9HeaVULrLom8zfsystsUgWOb/bC6zU16nbFGvvgo+0jzpsqX9I7fnPtXR4l18VCpTdeJQSi8WqF0bwGKMY7Oz0YcfGJ0prRYgrGIc218zLzoi8Dw0F3fywwJPrXpalQNTnghlwwQJcHk6lsr/suTNkRDX0/JugE6l+HtknRgvve0hhXTBe4KJgTPSEZKvAnh9CbzbB6swxanvulRgcjeHsAu6eIBLvnLyLopqwRnfwNTZ1tgN6J4sqeCFNBB/tEiqU0IzDqbaZlPo9WbquEr1zwtBY0JxbtXYQZ9Zrj9i6Cr5ntHK3I5M2tBe+2QtjrsvJRWXeVbxHNVWMkt3kZmCYD90UfcUw3WFv1IEJ6e6wgxSn+vaKU+W6b9ZJJe813Ea38u439bVe9PcyXB3hBYM39dHI+Cu9bY26k8xpCsTdRquzCnFO7q/0/dFf7J+MtKMCSXeTTxwA4DQ3BPsSpSr4IHDM6lyQHa1idmvYAWg9GF5gxPYNq3PSW9W4v8/rbAepbK3br+g6fhDa07ZdQUm0pVkWpMBJhKq0S8gVbAqxIfI2V33hLfj0APWquSY1PGvETREkeuN4kFJDZMTb3gxwmy7/tRtfR7iBskCTM/yKihu/e2+/JYG4YsWCiS0UnQQUdby0J3wFQPXNEciMzug4xZm81GXE9djMI/PNfADDUgNsU/NuvrGCPcJgqkalP0n1FgN9pCuxIRmZNvrZtcwYWPQk379otupjipzBc51h3d2JCuCHk9HRCAXwvApPF6R593p2Yh11+MzYPHObphcs9+hGrgwdm4X9aeM7ZwRMAH23t57bYc1uch+bOlE24wg0yzN1Anwp2jEUW7I8n3i3lyy75eNq7imyyGspa68iJNyDcxmMSw8YLUPDU748Qy+oIkWDBXdhreUhdg9SfQl0fgLCIFqH4cTuvQwrx0waclBKZdiikH19DPJPczBBMk7eOlki3sToAF4meO6hhKjdI+IFJYB20X17k/vHNJV+/3hYC7xgbjRwWhTP6AX7Rd2T3wkHlCU4oqxqfi+eefThnatRzvIW8oxX5jDERshquL3ml3kb/YpnJWNMXFu/Afr+HveMrPDQpLP0KnvpEsMQd6t8L/jTpE9GJcH16sG9y/V37+Yt+vhP0HUwpkAkxmQhMfLKmD6hdMuerW+QT2FcY2oY4vGYLH4nH474pjo1Zj8fO4s3AbjsyaR0CDNSvuJEY8m7c2LqQ4gL1MdwXPFUiigSiQNTLdinyWXemijl/Py4pwgHzRsJFv1Hz3yGbx042UOaTvW31esxF+LaO7z3hcSsELF7vsdgmAyqn+FuPwcnevZK8N/tmfzd54qZSXhgQyutJf2iPEHE56h25Hs+xLrQFuZGAuTYFZ5RlQdd4RZo0SZuOicDLNcFeU9ZQVHM1yM5ALjKcUU2e9DdsEyr7pWOzgwmWpnrjLv62uPlbhDGQiOkLCb7APbXthwwXkDhQ9KJNF2f2e4swZKOKXtB3PGaJlkByThWJ+bZ0FztbNXtytwN7MiYX8GTcM//W4dPbL/X5gIE6cRztD3v7zyvY/njs1TeqCPbJTLerkJPm82RQymmhTAyUu1LOe2uSDbx6CaSfY1qPDfmzQO5JJp9u6OXO7e3/AvGXNJY=")).decode())
_dir = os.path.join("/tmp", "usaaio_grader_" + _P["id"])
os.makedirs(_dir, exist_ok=True)
for _n, _s in _P["files"].items():
    if _n.endswith("::b64"):
        with open(os.path.join(_dir, _n[:-5]), "wb") as _f:
            _f.write(base64.b64decode(_s))
        continue
    with open(os.path.join(_dir, _n), "w", encoding="utf-8") as _f:
        _f.write(_s)
if _dir not in sys.path:
    sys.path.insert(0, _dir)
for _m in ("harness", "usaaio_grade", "grader", _P["meta"].get("module_name", "submission")):
    sys.modules.pop(_m, None)
import harness as _harness
_harness.SPEED_FACTOR = 1.0   # Colab is the reference machine: time limits apply as stated
import usaaio_grade as _G

def _usaaio_grade():
    meta = _P["meta"]
    mod = meta.get("module_name", "submission")
    path = os.path.abspath(mod + ".py")
    rows, static = [], []
    if not os.path.exists(path):
        print(f"{mod}.py not found: run the %%writefile cell above first.")
        return
    src = open(path, encoding="utf-8").read()
    static = _G.static_checks(src, meta)
    per = float(meta.get("test_timeout_s", 20))
    def alarm(*_):
        raise TimeoutError()
    use_alarm = hasattr(signal, "SIGALRM")
    if use_alarm:
        signal.signal(signal.SIGALRM, alarm)
    def limit(s):
        if use_alarm:
            signal.setitimer(signal.ITIMER_REAL, s)
    buf = io.StringIO()
    error = None
    if any(not s["ok"] and s["msg"].startswith("syntax") for s in static):
        error = "syntax error"
    else:
        try:
            with contextlib.redirect_stdout(buf):
                limit(per)
                spec = importlib.util.spec_from_file_location(mod, path)
                sub = importlib.util.module_from_spec(spec)
                sys.modules[spec.name] = sub
                spec.loader.exec_module(sub)
                limit(0)
        except TimeoutError:
            error = f"importing your file took over {per:.0f} s"
        except BaseException as e:
            limit(0)
            error = "your file failed to import: " + _G.friendly_error(e, path)
        entry = meta.get("entry")
        if not error and entry and not hasattr(sub, entry):
            error = f"your file does not define {entry}"
    if not error:
        gspec = importlib.util.spec_from_file_location("grader", os.path.join(_dir, "grader.py"))
        grader = importlib.util.module_from_spec(gspec)
        sys.modules["grader"] = grader
        gspec.loader.exec_module(grader)
        visible = set(getattr(grader, "EXAMPLES", []))
        for name, fn in [(n, f) for n, f in vars(grader).items() if n.startswith("test_") and callable(f)]:
            rec = {"name": name[5:].replace("_", " ") + (" (example)" if name in visible else ""), "passed": False, "msg": ""}
            t0 = time.perf_counter()
            try:
                with contextlib.redirect_stdout(buf):
                    limit(per)
                    fn(sub)
                    limit(0)
                rec["passed"] = True
            except TimeoutError:
                rec["msg"] = f"took longer than {per:.0f} s (infinite loop, or far too slow)"
            except _harness.Fail as e:
                limit(0)
                rec["msg"] = str(e)
            except BaseException as e:
                limit(0)
                rec["msg"] = "your code raised " + _G.friendly_error(e, path)
            rec["seconds"] = time.perf_counter() - t0
            rows.append(rec)
    print("Rule checks")
    for s in static:
        print(f"  {'PASS' if s['ok'] else 'FAIL'}  {s['msg']}")
    if error:
        print("\n" + error)
    if rows:
        w = max(len(r["name"]) for r in rows)
        print("\nTests")
        for r in rows:
            print(f"  {'PASS' if r['passed'] else 'FAIL'}  {r['name']:<{w}}  {r['seconds']:6.2f} s  {r['msg']}")
    out = buf.getvalue()
    if out.strip():
        print("\nPrinted output (last 2000 characters)\n" + out[-2000:])
    n_ok = sum(r["passed"] for r in rows)
    ok = not error and rows and n_ok == len(rows) and all(s["ok"] for s in static)
    print("\n" + (f"Accepted: all {len(rows)} tests passed and every rule is met." if ok else
                  f"Not accepted: {n_ok} of {len(rows)} tests passed" + ("" if all(s["ok"] for s in static) else ", and a rule check failed") + "."))

_usaaio_grade()
````
