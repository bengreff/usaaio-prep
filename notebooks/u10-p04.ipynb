**SmallNet.** Store one `nn.Sequential` named `self.net` holding, in this order: `nn.Linear(in_features, hidden)`, `nn.BatchNorm1d(hidden)`, `nn.ReLU()`, `nn.Dropout(dropout)`, `nn.Linear(hidden, num_classes)`. `forward(x)` maps `(N, in_features)` to logits `(N, num_classes)`.

**save_checkpoint(model, model_config, path).** `model` is a trained `SmallNet` and `model_config` is the dict of keyword arguments it was built with, for example `{"in_features": 4, "hidden": 16, "num_classes": 3, "dropout": 0.2}`. Save, with `torch.save`, a dict with (at least) the keys `"model_config"` and `"model_state_dict"` to the file `path`. It must load with a plain `torch.load(path)`.

**load_for_inference(path).** Return a new `SmallNet` built from the saved config, holding exactly the saved parameters **and buffers**, on the CPU, ready for inference (every module in eval mode).

**predict(model, X).** Return the predicted class of each row of the float32 tensor `X`, as a `torch.int64` tensor of shape `(N,)`, computed in eval mode without building a graph.

**Example.**

```python
config = {"in_features": 4, "hidden": 16, "num_classes": 3, "dropout": 0.2}
model = SmallNet(**config)
# ... training ...
save_checkpoint(model, config, "ckpt.pt")
loaded = load_for_inference("ckpt.pt")
loaded.training                    # False
predict(loaded, torch.randn(5, 4)) # e.g. tensor([2, 0, 0, 1, 2])
```
