`A` is a 2D NumPy float array of shape `(m, n)` that is not all zeros, and `k` is an int with $1 \le k \le \min(m, n)$. Return

- `A_k`: the best approximation of `A` of rank at most `k` in Frobenius norm, a NumPy array of shape `(m, n)`;
- `rel_err`: the relative squared Frobenius error $\dfrac{\lVert A - A_k \rVert_F^2}{\lVert A \rVert_F^2}$, a single float.

**Restrictions.** Only `import numpy as np`. Everything in NumPy is allowed, including `np.linalg.svd`.

**Constraints.** $1 \le m, n \le 300$. Time limit 1 second on CPU.
