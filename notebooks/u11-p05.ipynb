# Training on badly scaled data

A tabular dataset has features measured in very different units: one is around 1000 with spread 200, another around 0
with spread 0.01, and so on. A plain training loop does badly on it. Find out why, fix it, and return a model that works
on **raw** features.

```python
def train_model(X, y):
    # Your code
    return model
```

- `X`: a `torch.float32` tensor of shape `(n, d)`, raw features, one example per row. `y`: a `torch.int64` tensor of shape
  `(n,)` with labels in `0, ..., num_classes - 1`, where `num_classes = int(y.max()) + 1`.
- Return a trained `nn.Module`. The grader calls `model.eval()` and then `model(X_test)` under `torch.no_grad()`, where
  `X_test` holds **raw** features of new examples from the same distribution, in batches of any size **including 1**.
  The output must be logits of shape `(m, num_classes)`; the predicted class is the index of the largest logit.
- The model must reach **at least 90% accuracy** on the held-out examples of every test. Predictions must not depend on
  which other examples share the batch.
- Do not modify `X` or `y`. Set a seed inside `train_model` so the result is repeatable.

**Example.**

```python
g = torch.Generator().manual_seed(0)
U = torch.randn(1000, 4, generator=g)
r = U[:, :2].norm(dim=1)
y = (r > 1.0).long() + (r > 1.6).long()                        # three rings in the first two coordinates
X = U * torch.tensor([200.0, 0.01, 50.0, 0.001]) + torch.tensor([1000.0, 0.0, -300.0, 5.0])
model = train_model(X, y)
model.eval()
with torch.no_grad():
    model(X[:5]).shape                                         # torch.Size([5, 3])
```

The same network that reaches about 95% here when the features are rescaled gets about 47% when trained on `X` as given.

**Constraints.** $500 \le n \le 3000$, $d \le 8$, at most 5 classes. Time limit 10 s per call to `train_model`, CPU; a
small MLP trained full-batch for a few hundred steps needs well under a second.
