# Tokenize and evaluate arithmetic (a small calculator class)

Implement a calculator for arithmetic expressions such as `"3.5 * (x - 12) / -rate_2"`, in two steps. **Tokenizing** splits the text into its pieces (tokens). **Evaluating** computes the value, respecting operator precedence and parentheses.

```python
class Calculator:
    def __init__(self, variables=None):
    def tokenize(self, text):
    def evaluate(self, text):
```

**`__init__(variables=None)`.** `variables` is a dict from variable names to numbers, or `None` (no variables). Keep your own copy: later changes to the caller's dict must not affect the calculator, and you must not modify it.

**`tokenize(text)`** returns a list of tokens, scanning `text` from left to right:

- A space is skipped. It only separates tokens.
- `+ - * / ( )` are tokens on their own, returned as one-character strings.
- A **number** is one or more digits `0-9`, optionally followed by `.` and one or more digits. It is returned as a `float`: `"12"` gives `12.0`, `"3.25"` gives `3.25`. Take the longest number possible at each point. A `.` that is not between digits (as in `"3."` or `".5"`) is an error.
- A **name** is a letter or `_`, followed by any number of letters, digits and `_`. It is returned as a string: `"rate_2"`.
- Any other character raises `ValueError`.

**`evaluate(text)`** tokenizes `text` and returns the value as a `float`, by this grammar (written from the loosest operator to the tightest):

```
expression := term   (("+" or "-") term)*       evaluated left to right
term       := factor (("*" or "/") factor)*     evaluated left to right
factor     := "-" factor  or  number  or  name  or  "(" expression ")"
```

So `*` and `/` bind tighter than `+` and `-`; `8 - 3 - 2` is `(8 - 3) - 2 = 3`; `-` in front of a factor is negation (`2 * -3` is `-6.0`, `- -3` is `3.0`). There is no unary `+`.

- A name evaluates to its value in the variables; an unknown name raises `KeyError`.
- Division by zero raises `ZeroDivisionError`.
- Anything that does not fit the grammar raises `ValueError`: empty text, a missing or extra parenthesis, two operators in a row such as `2 * * 3`, a leading `+`, or tokens left over after a complete expression such as `2 3` or `(1))`.

**Example.**

```python
calc = Calculator({"x": 2, "rate_2": 0.5})
calc.tokenize("3.5*(x - 12)")       # [3.5, "*", "(", "x", "-", 12.0, ")"]
calc.evaluate("2 + 3 * 4")          # 14.0
calc.evaluate("8 - 3 - 2")          # 3.0
calc.evaluate("2 * -(x + 1)")       # -6.0
calc.evaluate("3.5 * (x - 12) / -rate_2")   # 70.0
calc.evaluate("y + 1")              # raises KeyError
calc.evaluate("(1 + 2")             # raises ValueError
```

````python
#@title Grader: run this cell to grade your solution { display-mode: "form" }
# The grading code is hidden on purpose (double-click to see it). It writes the site's test harness and this
# problem's grader to a temporary folder, imports your file the way the contest grader does, checks the import
# and ban rules, runs every test (including hidden ones) and prints a table.
import ast, base64, contextlib, importlib.util, io, json, os, signal, sys, time, traceback, zlib
_P = json.loads(zlib.decompress(base64.b64decode("eNq9XAl320aS/iu9zL4RIJMwSR9JmDg7TqLMeNeO/WzPsSvpISDRpDACAQwatMTR6L9vfVXduEhd9uxm3ohko7u6uu6qLvhqkMSDmRpsxpNRMX4+GKrBWlcRDV3tPqmSKtUY+5if6yz5h1ZRFiv9KUo3UUU/yqQ6o9XJQnmRMusoTdUiShebNKryUi3SyBgfcFJtTJ4RoGyTptgxyTaVNjTw5Bn9jJPlMqFV1ZZGMKCzqsT3wU81NEY0jzepDrNozUgVZT5P9fp5aDbzdWJMQlvQLMIiv9BxmKyLvKywyfEpDc+jLNNx61eY5nmBx8soNRqn1aYKq2St800V4sF0jNFoxTAGxbY6y7OAT0W40yM7YqoyyVbtkVIvNqXDx47pssxLntSgG5hCLwbAx+SbcsGnystklWRRykujsprSYFVu9DX9XiYpU+1qcBaVGRE1KJhMJ/y/P+q00KVR1VlUKUsctSqjGINCDeX9tizztbLL3ejhb35wkp1kL+10lRi1yLOK+PB4kyWVefw9Pn54bKHS7yKJf3gsswmJQL2qVKyXCUFVoKNabrJFRUc0CuyKeTA8FNyq6FzTN32SNaRQwlz1Gw39Fqijv7588+710QeVJqbCkbSABTCjzFl+kSla9H6TKe9TYhLCyv9OfQC4SpUb2lZ/0uWWF9HRfomSdFPSyjJKjFb4qS5IeFWkCJ6JVlpdkDDTgdWSJPdHnc2UibbqAuheRIae5tlqqDJAVfFmXaizJI5pepIVm8qAesKFzJJ0kRfb+geECvS1PzNav1UENSsw+pX6KU+jufrp3Z8M6ZRWBgJcglYZUDkw6s1UvSvzoTK5SnNSMUCEzKn1hogy1yAqkSiHHiX0OE1Ahzmd/4xYuYwWpEEBdpoGYzA3UqsN+O/JzoZgpXpUnZU6ihUJJTGMZpX5ZnWWbkla0iVpPrBY5CUIXWoQvVLJEgjST1Y1ew6rQiDKh3dHRz+Hv7z86ePb9+oFdseBTzJWImaDd3S50AUkxZ+dZIr+K+iRzCKBIorr2CNZjIdqbVZuDu2b5SByFtsR/Ncw18PcBkiYFd6lW0umpbXGcSgvF2et0SWdP8lMFWUL7V0O5XnwUWcmL/3Wct5WV5syU5dBTIZ0ceb5waLY0F9ms+fLZM3HVK94uyPYghYUOXILVlYEkYnKMtoS3s05zFlUaG+VV0MSygx/ST5fnAxIssl6ngwcZoZoXW2KVHs4Oc33A15qcWGa0hw3CbD8oVqeDK4A8JoYbmQvdWXI7uhLMlMVycRVa/41bddgtkhzcxtmQ1VWefpiokfPhyqyX79x+JJmXRDKNbKRqbZ00GWaR0AM47xn90HrMCs5H450IV9vOI4cwE73953NPeLj1RsQQxJD5i2paDXhkaaer8hSQArbzy7ss/b2sKRRwpYww5okW9bArSCD32lqiUjEsOTCH0su/GkLHrkTohgTAsito0t8RHPjrdRIXfi+D9h0UHhtTQ5OjaF8e1SlSyc2dAoenqyll0blCmaXnZe6oo9Z8GR17Xd4r/++idJ7SGVXqWQqkM5ilvShzxQlJKKqKu3zkwGzowGyR1xaAtLM+T8Xi7ZoMP6hpQMht6NMbarWEMCVWR9jEgZgy4dvYJAxvqJH/1Z2UMMsGuowIzEhqwizA9/MjfzgDVscaZbs4I9RIoUF7AchB2Fh2EFnReYPnLO7HrhJBw5qF9PQZFFBnrz6/zXNaZ7pBxpl2tJJ5SVEkn37wOdgGNrbxceLkwUREoGL38ane7gOdgDoxVoXLz5SpNcSMIvgRyJfH71dAH7HhWAoAEx+dtkhPLHFi4Zq/vl0j+5Fd5kiesH7fS7ZI5Cd4ZiuKbAbRbKJ8eb+PltDyxs7s3d5bSfm9ivztqvZOMFQrB1JdtbmlYPDEFrBSx6uN6QUFN14y2yoDsmckmYcHp5f4JvDRELHn5A+LTPeeYkAFWegGIeiwBWpV5RRyFiuNmsKysmFqCKNFjpwUSd7fegT2cXjRrMin+PZCAuw46nMpOgMziPz+gjJYyxJSJDpwMbHUk2hjC4p5fP+kRSerOE9OgLO9qQWLmtEtpTYkCzGGgF+skwQWSJkbs7y1VWiHqnJ9XfqIi/PEdVHLLwK7KN4tLaXlsiEfENhBPFJ5hmNUJCwui+Vv2uTmKxrO+TeE0lzTkn+0FDsTScAgTjWPahXUvAuORS2eksgSqRglCCtDY7EcfibaEH5ccZL6JASdSNb4cAewQFlF6P5JiZLimGOwSksT5BAbQpSJ1VdJAsNgEShhDx0vrQafKZrihoWIYwsKV6C+wYEWrGhZ4H6iAdJScN0mPQAW8AvyGagsI47YlWNEU0ShICyy2XIYHTprM0dohRX+xdTjFLZcMSe94WyTFSPVTttaEQyhCiW0AZv2pY7YiJt8/0LC6lniuaU05w3Q9F0qM6ntFvXPjLKw95g5yR3k4LxBCGmTIZpa5jJQBmbF8PF3kCOdrzZPhDrUZXnLDeU/MfVLJgurynKJynT39msqxZiFiZSOwakNhny+StLGruOc0hK7ByL9/zHW1JmSWnwkkRrhQBWkswryycGBO24VUHLbEXaqWPHsCbBIU7G+ZqovYxI/MJ64glXTTYmipI85ApDp8rxB4yQBhBmTfEgYaNxlqQxtG6BygglzAl0ATP/88PbXwnXjDXHVDFhyLoqpRk1eqXcPup7W+II46T8QX3fqtUUW/pNWvlPHqt+2En5Sdla6T8x9rJKk3k9JB80EmyqJG2G8/rr30zeFAmKqDprLzfJKouaZWZreuUF96MkvzCPFuf1CElxhmKB1BrMIqFjZro2dzRwnlSjVNM0NQmeBxMb05kZ6kRrFO+GbFIou7DOxxm/IkpQ+zrJ3B7BMklJot1P72RAaOelRgZoKy0UiQaHf8rOM9RwTJ6iopJzBYA2TApwLTi0keIfj94fkeZYWgTv6NMLQ5TBwtAPyHBhOUV1RVQSZicZUSXA5ICkVZeVNyZloPABYHwCuK/2xbUg9RX56r9HM3X0dDzlStjrv7z87w/hjy9//fXoZ8Lg6mQQ2opiGOIwdDDHT/mpL/XCfqM4X74t8nVBuJ4Mrk+y12/fvgt/ffvz0QcC55GsBL/kpKv48hcSXC1fX5pttqgfvKYY8ieCIb8+6NaPnynIbH79QWfwz3l5dFn47RLLR6nF3FVlCSPK9NYeZGyzJntDbkvXGsupogPUrkYgslmEizO9ODeeKRdgcRV1/e170XdEw+SsQMf8XIizNisizHdkFrkyyraF1cZUBzV3OBiypUQ4qqpMpKzYcVHihI5P90WzVakpsGMqkZRQfk2IduPQD1vKzy85DkVRTu/Gh8cW7RlJC9eKLfYzmEnDy22KnGdiZ650gM8sv57hO82m5OfaImgr1HAJRK+ADDPpSa9sDRodn1pE51E9zn6yci6Hy6HhxjCw1vic9CGUWqkdh4hTyEi6m8SkKojESgkUBDBBkPkeMhoE3ckcNXo6zjnHbjoNINh2A66dt2i+jqr1Jm2PLLk6QqYatpmIfxGl5x6Y0XPdrTAd00WeJTXoZxV1KIuJAaM723VgtqRC8bIX8SQE5WuUFuRnQLsRIRE5emuKPMkcIZkD8DUjK4zwZ/t9Y4sXQRTHdg9/d3LDm9a8wBRpAnZjz+PR5HTPwoZ5waaIEXXvWdtOEtPbyfgLWb3+aYihxCumoy280/EbhdpDS15hCUnfv4CKfQoStB4VWqTrUGCfBPSW3kS8z1rK53wo0V863emfvScQjAcU7V5Qf21Z5NsAJvHNp9o/ad+OjbPq78mKH0RFobPYcxUfWXQyEJMn5v0/uvTatwt7wR+T7K31Yi83q5cGPsiX2kpvQZDbeW+i6g2FjDsyLTbogcgRYi2BbIFEkGhB3Wj+XRQgWiIij1FWCMwL/pYjPWUD67X28f391uURPAoRxoG6OhiqAwHiNAoKeED7Ud67OkAp9np/PXEf/ihZ7EHfONwbYF/VN4jzKDOg7eLMeggOBRsHweVnHVES+1f+9L+TbDS9iLZmJNeulMlQXJjUgCCUFlqepduh25Myl8bVeCIFUkSyoRSCjq1K9YpS8wuETXwb7QVB4LecGQVlKs7BFAF8lnB+KVzwWk7zd+wcW15Y0LXOl8Twn8pr+dLfqU5c2BIh2uHBokMYGGXpA/ie8ffIDkH2v4TDWe72qK9kGy4T5t3Dh6JEYV5IPNmttmG6KFlP9e5zYGSUJBe/Vw42UtaW3ihPIie7w/H4tC3ce47/EBI0u952eDZuO0fm0c87sIRJLOydg9Iw1BhCXeoznXESW8eOvOpfe357sutbMvVlmRCYdBtyJOsRCEp0Q2RT3Yj+baZHjGeszaJMOKtQBYkq6fdK2ZtonpC462skbKSuGvMMX9Z3K0xzlFVc3hpQ3kzfq7CaezoIw/oBpXw22ARwijSX7NSX2IdAEKeWSD+1i/mQ+dVHsFHpBUomKFjBynJt0god/UU41oraO0NgBQsMZvN1WoO+JSZuTPiGRLfuRxD9X1/xrteY74i9pkjcq3MxsrT9LBeZbFSuPh3XIaI7yo0zp6etlNhSCgH4C1VPeeLCdZJ7GkfBIUjzKDaex0g8hthAJ7gUMQA88lmoZHjO2FH6BIgWmfYERw2uJr5Qu3ohCeOJ7cE54YYb0/yUwoyIrM2kIYkY+DUnFrThHzfQToHODaloreiIYKGC5liwOmVvgV+sJaftcNaldC6eNVx33Nm2bSW4auExOdEcYjyZ3vbyIiWu5LkkpJM8+MCtQ6/eOuJJkSeQD8/++vDqDy9fv38ztFm6nUrmLOTirrsFbtmybicTaDkdO1y62TE3wTTFKmJnnJSalE+Y4RGm/TDL4airBFuUDs1XH1+9OXofvj96+XpYY9cLc9DvhJN3SmHcBRWiNiOVHVQn+cqkdaJW65eIRlOYIza1TFVvv818dzsLizfE1h7+9BeSxsg8c4zHnDCciuDvHol1SJcBakChLIPl8T+DcONubcKWXTolCSuIVjlOxZrJGWGB2aqxxa3y/JzvFNSV48csGKMC7EyX3eXHyOi6QiQ1kKYk9uPro/F40iDwoGPchLG9GGI0cRODDrHcMsoGQbf4o1qtd0DvI5SzK9iX5BkSxTfxHiXgT8fj8ez0sxWZGxW7dRweavd3yBx3V+xuNOksQ3nk383chlZxriVokmY78i4Acd3O2x94hNXDdPJkIE1/0MLaZdR9gPW5bR/h3bq3ainf6iZdEmh2lm33s4Utl+XJFFgG1zrognhr9uBrEDR4lKgsxbDjG2z7p6g0bo8gqfTaeNw9k3W9AltVd++PC5SI0PCWfrvgRVYCl5AckmDHXvDM7hjSX27IbolQIFrhLoLMHa1ncGGek2yj2/xbiIMVezhjIMfPZqdkv/lemLC1VWrF5vFkYEG7ya3d8BTVYKQ9+9IUa2/l0gdD42B8/ZA7sd3Gh8/wOl/qeezl3K5V/iyTZplw3FAOmorAe7d3Y8eCN4sl9hAdZ2Od5tnK3cv0bbbykky6zDiaHyKBWEalcjeDftsI2M35cqNX0/7CA9c4I7ze06vyIF/yr8Kl3WfA9xXxPX1IA6sWcMArSYpj76aL2qF6ss+92XD21CVkZW3YHuyK6gUSq75A6ug1settUSkblXmep94OYvIMoKqO5AJc1QFXr3AJyo0uhdual8olPGLdwhD5DS7KZu56AOmO3OzWzgJ3ujfdyA359nqI6+OhbfLATnVbOBlya5D1ZbQuYNiIdH9++f6DGMZLGKopDBfaVsKp2K1n162LaYiJ4bYcklxpktvfD7XMev1i9DG7IR6QCfeW/31tmDMnwbvZZLvbjrZqHjRX8HsBZrm9nhLIt4NxBKrycxPm5/3Gzlu7B7nprdU8OAMUeXPE9Zbc1UYIEN3eW8roeTpkC9/77cp7N3GLruWhaZ+5BtLQzbU8cffmbstTTYPdhic5jY9iFn+9EC1bcRvnLpbqKrHtnP1mzn4np/7kkVjCRd4mnK7bL3Bv5XhY4f8rxbF+3+dGueQqRwd5tmoPlR7X771373uKD8Poyg83RVOANRIOorFmokffqkOFrulJQPYcU4S9d+1+vy7ctmXimMMe25ZOgua9Ju4W9WC46kjV6t0icGJN0dyT4Nmhd0lHmEx9juiOaQTm7VDCM08+LuVjhI/JFCcDWoNTaVCwhtLF0Y5FLdk5GUzJaz4h0jzlXSZPBYZdqw7qxwd3wPmGcH1C/58ynCc9MPXTgzvROVQjOvgjNZFzj57vYNTMuAsa0YxmOzpS6jJy3oEgfz3uQb5p9kHX2hovjdbzmNSps9cWGDHg/9Jbbi7oAOfH94PkTWiu4PhnHGsXmMw42BXBWoJul8HbJE/R4cePngbTZ5JKHNNPIdQjSBkesHitKSBIRnGySvgFp7lrL4j1IllL3+4tu0STcH4YLmQH+SmibOUbj3gbuYjhBIL3kk24y83g5aQ79pleyhZWNy4t1MiirCQPSbfkEGyxfk65u5I86zbAnjf62vcd/l5HK1kdv661sa2U0rh0po29V8ObkXccQfYQlV4X1ZZdxM2sl9DX3E8COBIkcPCAx9AXexP1TD6n6t/VE/kafWUZNAmeBdNe5aUvzDX27Ju6YkzmNkK+TPKexOh2Ru0fLWG4rp9HsToIDuQ1N/FOzdrG73tXAExG2N+lQ0EM1bGGk3kQCcRrMSG6ZlGsIp2ChidjMg7PyTRMLYvtMNSRZsPEPaWn3+Dp8+CZPJUnj+jJoWLCTsmm+W0C1tarHQPw/YIck0WgfkfkkCXnsZonuC1NVmeVSyEf8ZMRuedmYY88qV5CVsIS676EQi2Dz/Zezvocx3/qCPRNQyCaCJLy8evRKU0ElevJn08UbtMfNcyv7/+MWlF2VyjOPPiqiihwG4UyvZLS1+3EcUnHk+tbyTRiFz2qCcTui8kAzyZjIztST3pK5BpNRYjcHLH5vqOWrL6xnResgAe7rOc/vSd5I7FJwJ9IlnEXX2RfKVVMGzJdSZeCLT58zvK+Bjc28p48eHo7DzwWPd+pM/yZENXzLn1fvFqLPR7E2gMPLrHma7/DCG/Cvyf3JGfrMDeR7KYp/fDyPnbdUWRyvd/CO7v+yDqsWilhVqbu6SEslvx4ZD89/w6b381Hdmw+48Bvg9cl7WUid8d4aWIdlTcZ/Q5F9hxp6lAlzvj1+VCykRBZXbov3vQLDwHTwfcrkmPeG+Pboj1EstuduJGM+ca2THdikdsgTUjhJXqVuOR/dJn/nKDqm2c14NgOIM75B03YlbNPUZmg0m1CNMLweydxW+rifsVli2+jCVdb+LZTE4dQr+dUx7We7UuFOm9A8LtUvHSo9r5HBFmpkWPgSmpauEoKQ5RKUYJyb6IcS7hH207G49uyBNhI4cDI5i3cmtz8wxbM5nOtC4X4Ezzh95XyZQsdz65JSTrcK1yYDTT9O9IUiVGnn7t3N/tlOtZmYDzucuiLSetEZaUzSniL0iOpinVRnQ2lv6vXDc7vfCB5pkidxY4Vt0g30v+Nl8faT0mj38k7GrbL2mNlP0NGvtQlO3amod9pKCkpeEcV175j4jWXcYwaKgBjBJg873s1Dp5072mahTtPma5r9M7idMeoi5YB/dUrSkDwvgGKS3Jy3z/d+1ZktibarFuuAtdkO3CIVeNeA2UPrac9tD7JRcLVp+vgai8835+Np3HnstCihFr+J6l9SDPBlYxISN1ctzp6tbcmp1XQzjsCAMPjhGBnQ06KyNJVw/prsbvNsy/fxpNtuKnOr7MybOeG7CsEBV8jPBodPj4Z7GHr05qbESES3Q+TOU2d32+qkf2VtBn1OP3shpajiM5gcJC8cN/mjJ0dKeY9Wy5QQ/s6VyjvP7WNOa5s8SbWpDFQpm3fpejU1NRHtkAUcqfP1/XN4B7rDlDOFufcwDV2PVlys01j36sn43Gb6Rw2FdseDXvsIVI+R+UOpWPZ5vbbR4SCrmuG20eL7RBHtLXDmTwi28gOSwDyCxRydTEySdw2QOhg3bkG23G2d17qMgEekXdq/UMH+NcAbqjt3qeKav+NGm4TugL8L6+ofkFV9X7YRO1/LGFPvkrPpN9sj9S6Vy7x7ybxZT+sGITlky83/p/4lqsjOewS8a9kvJjC4ro3pOXtlo5BOPflquy8BwTWwYKYfPvttw4Ee7YXgsvx+JTNjWvqzdkAAVgOs0bwcMtAmw7t/Mms7lyo70CY+UxnQpS1DiiUWZTOZDPcBIM+NbN2JZ0tjOBw7DADasdwHPm1uirIPdyFGDqjbccGC6h9CZt53i8Z1LIrOZB/Z/nfvWn4EOFCoy2JDDaY7YgP6xYCphfW9h+qZ2Pmxqj9Y2I7r3072AvQ7nlGbORzNgisCHRGQktS3anz8e0y9pXUGPfdErQNrq//F/Nzh5I=")).decode())
_dir = os.path.join("/tmp", "usaaio_grader_" + _P["id"])
os.makedirs(_dir, exist_ok=True)
for _n, _s in _P["files"].items():
    if _n.endswith("::b64"):
        with open(os.path.join(_dir, _n[:-5]), "wb") as _f:
            _f.write(base64.b64decode(_s))
        continue
    with open(os.path.join(_dir, _n), "w", encoding="utf-8") as _f:
        _f.write(_s)
if _dir not in sys.path:
    sys.path.insert(0, _dir)
for _m in ("harness", "usaaio_grade", "grader", _P["meta"].get("module_name", "submission")):
    sys.modules.pop(_m, None)
import harness as _harness
_harness.SPEED_FACTOR = 1.0   # Colab is the reference machine: time limits apply as stated
import usaaio_grade as _G

def _usaaio_grade():
    meta = _P["meta"]
    mod = meta.get("module_name", "submission")
    path = os.path.abspath(mod + ".py")
    rows, static = [], []
    if not os.path.exists(path):
        print(f"{mod}.py not found: run the %%writefile cell above first.")
        return
    src = open(path, encoding="utf-8").read()
    static = _G.static_checks(src, meta)
    per = float(meta.get("test_timeout_s", 20))
    def alarm(*_):
        raise TimeoutError()
    use_alarm = hasattr(signal, "SIGALRM")
    if use_alarm:
        signal.signal(signal.SIGALRM, alarm)
    def limit(s):
        if use_alarm:
            signal.setitimer(signal.ITIMER_REAL, s)
    buf = io.StringIO()
    error = None
    if any(not s["ok"] and s["msg"].startswith("syntax") for s in static):
        error = "syntax error"
    else:
        try:
            with contextlib.redirect_stdout(buf):
                limit(per)
                spec = importlib.util.spec_from_file_location(mod, path)
                sub = importlib.util.module_from_spec(spec)
                sys.modules[spec.name] = sub
                spec.loader.exec_module(sub)
                limit(0)
        except TimeoutError:
            error = f"importing your file took over {per:.0f} s"
        except BaseException as e:
            limit(0)
            error = "your file failed to import: " + _G.friendly_error(e, path)
        entry = meta.get("entry")
        if not error and entry and not hasattr(sub, entry):
            error = f"your file does not define {entry}"
    if not error:
        gspec = importlib.util.spec_from_file_location("grader", os.path.join(_dir, "grader.py"))
        grader = importlib.util.module_from_spec(gspec)
        sys.modules["grader"] = grader
        gspec.loader.exec_module(grader)
        visible = set(getattr(grader, "EXAMPLES", []))
        for name, fn in [(n, f) for n, f in vars(grader).items() if n.startswith("test_") and callable(f)]:
            rec = {"name": name[5:].replace("_", " ") + (" (example)" if name in visible else ""), "passed": False, "msg": ""}
            t0 = time.perf_counter()
            try:
                with contextlib.redirect_stdout(buf):
                    limit(per)
                    fn(sub)
                    limit(0)
                rec["passed"] = True
            except TimeoutError:
                rec["msg"] = f"took longer than {per:.0f} s (infinite loop, or far too slow)"
            except _harness.Fail as e:
                limit(0)
                rec["msg"] = str(e)
            except BaseException as e:
                limit(0)
                rec["msg"] = "your code raised " + _G.friendly_error(e, path)
            rec["seconds"] = time.perf_counter() - t0
            rows.append(rec)
    print("Rule checks")
    for s in static:
        print(f"  {'PASS' if s['ok'] else 'FAIL'}  {s['msg']}")
    if error:
        print("\n" + error)
    if rows:
        w = max(len(r["name"]) for r in rows)
        print("\nTests")
        for r in rows:
            print(f"  {'PASS' if r['passed'] else 'FAIL'}  {r['name']:<{w}}  {r['seconds']:6.2f} s  {r['msg']}")
    out = buf.getvalue()
    if out.strip():
        print("\nPrinted output (last 2000 characters)\n" + out[-2000:])
    n_ok = sum(r["passed"] for r in rows)
    ok = not error and rows and n_ok == len(rows) and all(s["ok"] for s in static)
    print("\n" + (f"Accepted: all {len(rows)} tests passed and every rule is met." if ok else
                  f"Not accepted: {n_ok} of {len(rows)} tests passed" + ("" if all(s["ok"] for s in static) else ", and a rule check failed") + "."))

_usaaio_grade()
````
