# Logistic regression by Newton's method

This is the core of 2025 Round 1 Problem 3: the logistic regression loss, its gradient and Hessian, and training by Newton's method.

`X` is a float array of shape `(n, d)` whose rows are $x_i$ (any intercept column is already included), `y` is an array of shape `(n,)` with entries 0 or 1 (possibly an integer dtype), and `beta` has shape `(d,)`. With $p_i = \sigma(x_i^\top\beta)$ and $\sigma(z) = 1/(1 + e^{-z})$, the loss is the **sum** (not the mean)

$$L(\beta) = -\sum_{i=0}^{n-1}\Big[y_i \ln p_i + (1 - y_i)\ln(1 - p_i)\Big].$$


Start from `beta = zeros(d)` and apply exactly `num_iter` undamped Newton steps $\beta \leftarrow \beta - H^{-1}\nabla L$, where $H$ and $\nabla L$ are evaluated at the current $\beta$. Return the final `beta`, shape `(d,)`. With `num_iter = 0`, return zeros.

Neither function may modify its inputs.

**Example.** `X = np.array([[1., 1.], [1., 2.], [1., -1.], [1., 0.]])`, `y = np.array([1., 0., 1., 0.])`.
`logistic_loss_grad_hess(X, y, np.zeros(2))` gives loss $4\ln 2 \approx 2.772589$, gradient `[0, 1]` and Hessian `[[1, 0.5], [0.5, 1.5]]`. `newton_logistic(X, y, 1)` gives `[0.4, -0.8]`.

Values are compared with a relative tolerance of $10^{-6}$.
