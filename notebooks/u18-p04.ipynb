**`make_schedule`** returns three float64 tensors of shape `(T,)`: `betas` evenly spaced from `beta_start` to `beta_end` (both included), `alphas = 1 - betas`, and `alpha_bars`, where `alpha_bars[i]` is the product of `alphas[0]` through `alphas[i]` (inclusive).

**`q_sample`** noises a batch. `x0` and `eps` have the same shape `(B, ...)` with any number of axes after the batch axis. `t` is an integer tensor of shape `(B,)` of **zero-based** step indices, so `t[b] = i` means the math step $i + 1$ and uses `alpha_bars[i]`. `betas` is a 1-D tensor of shape `(T,)`. Return

$$x_t[b] = \sqrt{\bar\alpha[t_b]}\;x_0[b] + \sqrt{1 - \bar\alpha[t_b]}\;\varepsilon[b]$$

with the shape and dtype of `x0`. Compute $\bar\alpha$ from `betas` inside `q_sample`. Do not modify the inputs.

**Example.** `make_schedule(3, 0.1, 0.3)` gives `betas = [0.1, 0.2, 0.3]` and `alpha_bars = [0.9, 0.72, 0.504]`. With those betas, `x0 = [[1.0, 2.0], [0.5, -1.0]]`, `eps = [[0.1, -0.2], [1.0, 0.0]]` and `t = [0, 2]`, `q_sample` returns `[[0.98031, 1.83412], [1.05924, -0.70993]]`.

**Restrictions.** Only `torch`. No loops or comprehensions anywhere in your file (`for`, `while`): use broadcasting.

**Constraints.** $1 \le T \le 4000$, $0 < \beta < 1$, up to 4 axes in `x0`, batch up to 256 images of shape `(3, 64, 64)`. Time limit 0.5 seconds for `q_sample`.
