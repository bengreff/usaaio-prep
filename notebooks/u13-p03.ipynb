# Loan outcomes: tabular classification by macro F1

A lender wants to predict how each loan will end: `"repaid"`, `"late"` or `"default"`. Each row of the data is one applicant:

| column | meaning |
|---|---|
| `applicant_id` | a unique ID number |
| `age` | years, an integer |
| `income` | yearly income in dollars |
| `debt_ratio` | debts divided by income, between 0 and 1; **missing (empty) in some rows** |
| `accounts` | number of open credit accounts |
| `late_payments` | late payments in the past |
| `region` | `"north"`, `"south"`, `"east"` or `"west"` |
| `employment` | `"salaried"`, `"self-employed"`, `"student"` or `"retired"` |
| `label` | the outcome (training and validation files only) |

````python
#@title Grader: run this cell to grade your solution { display-mode: "form" }
# The grading code is hidden on purpose (double-click to see it). It writes the site's test harness and this
# problem's grader to a temporary folder, imports your file the way the contest grader does, checks the import
# and ban rules, runs every test (including hidden ones) and prints a table.
import ast, base64, contextlib, importlib.util, io, json, os, signal, sys, time, traceback, zlib
_P = json.loads(zlib.decompress(base64.b64decode("eNq9XAtz2ziS/itYT9WGzMqMZCeZOWU8d07GmfVWXhV7H7e2iwOJkMQxRXIJKrbj1X+/rxsACephx8nsJVUWBQKN7ka/Aeh2J012hmJnMdjfLfv7Oz2xM1e1RNPt+ps6rTNFbW8KmYtiUY+LudJDUcvRIpOVGGdS63SSjmWdFrkY3Yi5HFeFeD2g0ZnSusgdyKz/lCdL80WtNFr3nuFrkk4wfJHVN2jZR4PK64qed8pKJem45jFFsshUnMs5I6MXo3mKeQEbL2WWFVcqidN5WVQ1AT7b+U2bd3NZz+iz0PS3xLcsHfFjOr4EZUzinD/HRXlDn/libh5KmSeSx2l0lRUD/K0YEYQLPI5kniti2Jn9FmdFUdL8E5lpRaCVrmOCD8bF9GKw16dmOTVYgiyVpfk0KkqVxypPAM5Qa5otl+0SVelYR5OBh1BUpiX1ZPxbpkSyqtOJHIMXhJkuFtWY+VZU6TTNZWaoq+o9NNbVQi3xfZJmvCi3OzNAxsJFJa/COf//s8pKVWlRz2QtygpMUHMxrWRCjYbxIvh1UhVzYYe71se/htF5fp4f2u4i1WJc5DWW+ckiT2v95Ef6+OmJhYrvZZr89MT0BhKROK5FoiYgE/ODo2KyyMckblqQQCTcGD82uNXyUuFJnectP4QRH/Ermn6NxNE/Dt9+eHN0IrJU10SSMmAJmBZ6VlxB0nPxcZGL4FOqU2AVvhAnBK4W1QLTqk+quuFBIO21TLNFhZGVTLUS9FVcpfVMSAF4Wk6VuKrSGgSLSVGJlyofCi1vxBWheyU13hb5tCdygioSSJ+YpUmC7mleLmpN3DOrkFuWkqQ2X0i8iL/2K0uvANS8pNbvxKsikyPx6sNftZCVEpp0pSJe5YTKIy3e7okPVdETuhBZMZYZQYTwaTFfgCkjRUwFiwpS0xSvs5T4AE2vZ1hKErOiimimvahPiyvFdEHrH5iZNWBlareeVUomQpcKC4ZeVbGYzrIbSEs2EcAFWIyLihhdKWJ6LdIJIYivrNWWDqtMxJSTD0dHP8evD1+dvv8oDmh2Ivg8Z5PEyxAcXY9VSZISDs9zgX8lXpleEChwXCUBZDHpibmeuj6YNy+IyXliW+hfu7gB9W2BxHkZXLuxsFzeGLdCRTWeea0T0J/mupb5WAXXPfM+OlW5LqrQG87TqnpR5eI6SmCix7MgjMblAn95mYPQdFZMpjjm6Y6qqqg8KIZkD1ZeRlLLqpI3wLulQ89kqYJpUfcglDn9hXwenO9AsmGcz3ccZhq8rhdlpgKiHP3DiIdaXJin6OM6EaywJybnO7cEcIkF12Yucathd9R1qcY1ZOLW67/EdC1m46zQd2HWE1VdZAcDtfu8J6R9/MHhC826AsoNslLXNyB0khWSEKN2nrP7wiNmaugjkq7M4xZyDAG2e7iJNveKyWsmwIKkGuYtrTEaeGRZEApYCpJC/92VfedPT5ZUpmwJcxqT5pMGuBVkWu8ss0wEMyy76I9lF/3xBU9VFTjGjCDk5vKaPuRIB1OxK67CMCTYIDT9rISCqxN9Ur4NqtLlExs68UlmsA8igGObktlVJLDiFh/DaH+6DDtrr/61kNkXSGVXqUxXQhoenCS9FzJHgYSs68q+P9/h5WiBbBAXT0DaPv9xsfBFg/GPLR+A3Joy+VxtINCqDFcxhjAQtkx8CwPG+Bav/lB1UKNeaOosRqpjVhFeDnrSW9eDJ/RWpB2yhj+1ghUWcBjFHObFcQedKcwfrZyd9ZHr9MhB7WIa61yW8OT1/69pzopcPdAoY0onldckkuzbd0K4xIS1t4tPQAFxjwOX0MenS1wHOwIYJEqVB6eI9DwBswiegn2r6K0DCDsuhJoigsnvrjuMx7IEsidGX893+UV8N12MXvB8X8t2SWxnOLprCuxE0kyig1G4ydZgeGtnNg5v7MTIPvLadjWbKOgZawfJzv21cnAYghe8FPF8UXPaFUzynngMcwrNePz48oqeHCYmdHwFDyAmOc88oQCVaECMgyhwCvWSOULGarqYIyiHCxFlJscqclEne33SJ9jFs1azZMjxrKQBNOOF6YnojJxHHqwiZF7TkBSCDIJ1SEMVQhlVSfi3z2kZmDE8R0fA2Z40wmWNyA0SG8hioijARxJKkSWFzC0t392m4k9isHwhrorqkqJ6ycIraPkQjzb20jIZyLccpiA+zQOtKBQEVl/K5Rc+i2Fd/ZB7QyTN6Sv8oUbsDQqIQRzrPmpGIng3ORRN9R4gKkrBkCDNNZHEcfhbORZjTEBDQKSJuilb4cCeggNkF7ujRQJLSs0cgyMsTymBWpRQJ1FfpWNFAMGhFB66mFgNBs6Oo5pFiFomiJfIfRMEjFjgXSRO6UVaoRnEZI9oCvILZjLisEo6YlX3KZoEhAjZ5SRmMKpy1uYeUUrqzYMRo9Q2HLH0Hgi7iOKJ8NOGViRjEsWKtCHY8+UOi4hpfjywkFZM0Qg5zWXbJPd64nIPs3XtI6PcW2nsUHI/KxhPYsQes2HPa2Y2IGMLEnKxW9jhx5s+QaxHdVGw3CD5T+phtDdZIsqHlKkXNutqhJiFCWrHgMQip3z+1rLGjuMcEomdW+IN/3hKZJZIgycQrSkFsCbJvLXrxIBIO+5U0CqfQjtV4hasTXCwkkkxB7cnEuIXNx3PuYaz0FKmRcwVhk6V4xdqgQYAs7Z4kLLRmKVZQlo3psoIEuaUdIF6/uXk/TtBNRjSHF0nwJB1tbypZ2jZPRZuHvGjLXHESVr9JH70CjblDb5DK//NbfVPayk/lM1L/7Gw13WWjpom84GWaFGnWdtcNI9UEGu+2DpY812n01y2w/SNXikvuC8V/MJIji+bFkhxTsUCU2vQ4xRk5qoxd2i4TOtdLlaJQfQ8GtiYTg+pTjSnAmGPTQqyC+t8nPErZQouT89zN0c0STNItPsanO8A7aJSlAHaSgsi0ejxX/PLnGo4usioolJwBQATpiWtWvTYRop/Pvp4BM2xvIg+4DOIYyqDxXEYwXDRcER1payA2XkOrkTUOYK0qqoO+lAGhA8EJgTATbUvrgWJ7+Cr/yWH4uhpf48rYW/+fvi/J/HLw3fvjn4GBrfnO7EtXsYxEQPC3Hqar+paje0T4nzzNC7mJXA931me52/ev/8Qv3v/89EJwAWQleh1AV2lh79DcJV5PNQ3+bh58QYx5CvAMN9OlPflZwSZ7bdfVE7+uaiOrsvQL7GcmlrMfVWWWCLTmwckY4s57A3clmo0llNFB8ivRlBkM47HMzW+1IGuxrTEtez6249G3ykahrMiPhaXhjlzPQVjXsAsUgRoanysNrp+1KwOB0O2lEiOigqsLCodF2Wc0NnFpmi2rhQCO+YSpAT5NRDtxqEnN8jPrzkOpaKcWo8PzyzaQ0gLV40t9kMyk5qH2xS5yI2duVURfebFckjP6I3kZ2kRtMVwcgngVwTDDD1ZqZATj84uLKIj2bSzn6ydy+FyaLzQDMxrH0EfYlMrte0k4ggZobtpAlWhSKwygYIBDAimf0AZDQXd6Yi2AEDOJcduKotIsO0EXEX3eD6X9XyR+S0Tro7AVJNtBvOvZHYZ0GKsuG4vTKfuRp5NarCaVTShLHWMGN3hugOzJRXEy4HkThSUz6m0YL5GmA2MpMgxmCPyhDmiZI6AzxlZsxDhcLNv9NYikkli5wjXO7dr4/WLdJmltNw059nu4GLDwHbxokWZUNS9YayfJGZ3s/E1rN4qNVhQrBXz0RbeQX6rUBt4ySMsI/H8DVxc5SCgrXDBY12HA5skYGXoNuZ91VCm86FMP3S6s0r7ikAwHqRoXwT1nWeR7wKYJtup2txp04yts1qdkxU/kmWp8iRwFR8z6HzHmDxj3v+7y69Ns7AXfJnm760XO1xMDzX5oNDUVlYGRIXt91bWbxEyrsm0sUEPRA6IeQLpgaQg0YLaav5dFGC0xIg8tbJCUL/ot4LSUzawgTdPGG62Ln8ijwLGOFC3j3rikQHiNIoU8BHmQ947fUSl2OXmeuIm/KlksQF97XBvgX3X7CCOZK6Jt+OZ9RAcCrYOgsvPSiKJ/Qd/hi9MNppdyRu9azZgkckgLkwbQCSUFlqRZzc9Nycyl9bVBEYKTBHJhlIUdNyITE2Rml9R2CS4ZxRFoefMEJSJpKBFMYBnKeeXZhUCz2n+kZ2j54UNutb5Qgz/LQLPl/5RdOJCT4Qww4NFBxhoYflD8AMdbpAdQA6/ZYXzws3RbMm2qwzMu8THRoniojTxZLfaRt2Nkq2o3pcQTBkl5OJ/hINNKaunNyIwkZOd4ax/4Qv3BvIfwoJ21ruIZ+O2RjK3fh3BJkxiYe8QimZSYxLqSs1UzklsEzvyqN+XfkvZ8o5MfVKlAJPdxBzJBgCBRDembKob0b/P1S7jmSg9rlLOKkQJUYV+T4XdieYOqdu+poQN6qqon+bN+m6FaURlFZe3Rsib8VzH9ShQURw3L5Dy2WCTgCPSnLBTn9A8AIGVmlD6qVzMR5lfQ4KNSq+oZEIFK7KyXJu0Qoe/FI55UXuniZaCBYZ683Zai75lJu2Y8A6J8vZHKPpf3vKsS+rvmD1HJB40uRgs7WqWS5msrKafzpoQ0ZGytefehZcSW05RAH4gmi77LlyH3KOdCg5RVshEBwEj8YTEhnSCSxE7BA8+iyoZgTN2SJ8IokXG7+C4wdXEA7GuFyZhPLencc756I1uv5rCjBFZm0mTJFLDuwJL4MM/a6FdEDpbUtFG0SmCJRXUZwarC/YW9I215MIPZ11K5+JZzXXHtWl9K8FVi4DZSYdDdGC6+17eSIkreU6AdFpEJzWVT47fO+aZIk9kPgL77eT4l8M3H9/2bJZuu8KcxVzcdbvAni3rnmkiXu71HS7d7JgPwbTFqoiOc1UKymcWIwCmq2GWw1HVKU1ROTSPT4/fHn2MPx4dvuk12K2EObpUJD3dUlhErTHVZkxlh6qTvGXiUeQdLjOi0RbmsEyeqVqZbzFan87C4glp6oD+rA6Exph++oxec8JwYQR/nSTWIVVFVAOKzTCyPOFXMK7frU3YskunJGEF0SrHhbFmhkaywGzV2OLWRXHJewri1q3HMOpTBdiZLjvLS6lVUyEyNZC2JPbyzVG/P2gReBAZ2zC2G0OMJu3E0Amxwi6UDYLu8EeNWq+B3sQoZ1doXsgzSRTvxAdIwJ/2+/3hxVcrMp+D7NZxuMk/32H6uL1it6MJWnrmVXj/4ra8SgplgiZz2A7ehUAs/bz9gSRMH6aT5zvm0B9pYeMymnOADd32HOH9ujf1lG+6TZcMNNvLHvezhS2X5ZkuZBnc0UEXxFuzR76GgoYAicrEGHZ6Itv+SVbazRGltZrrgE/P5F2vwFbV7fvTBooEGsEk9AtesBK0CckhCc24EjyzOybprxawW0YoKFrhUwS5I23F4JJ5TvOF8tdvbByssYdDBnL2bHgB+837wsDWVqkFm8fzHQvadfZmo7dUDaa0Z1OaYu2t2fShpn7UXz5kT2z94MNXeJ1v9Tx2c27dKn+VSbOLcNZyjjSVAu/1sxtrFrwdbGIPo+NsrLMin7p9mVWbLYI0N6fMOJrvUQIxkZVwO4OhbwTs5Ly5sVLT/kaCG5wpvN5wVuVBvuT3wsU/Z8D7FckX+pAWViPgBK+CFCfBto3antjf5N5sOHvhErKqMWwPdkXNABOrHlDqGLSx611RKRuVUVFkwRpi5h2BqjuSS+DqDrhmhEtQtroUPtY8ES7hMdYtjim/oY2yodseoHTH7Ow2zqLd07WXGbZeWQj24UdGiH7zMVaWX9IJMz6Vko6xzuinpgWeZQYhyBbzHK85PmyO3cEKumsPvPf7ARmlnNL7wFyBEPb+Q098oPQzHauQTt4z5nw6nEpf7tA+RSsUXMNumXKXdXifU+T7nU1jY+zQZd6D084UgeGz74msZU8sch4C/hZCQl6VnmE9MvZlrjjH25Y1Z67O9Whhb14AN34tqSIIBMxYUPgzwHPlYWr2BsE4hsCb7CoJI3HK5/TpSoE7ogoMKjUljjPPBNzImAIM2s8Mzdl9cwKeJFxWLILMIboIsbYfvmXDu2g3rvWs8waetyR8Nm9sg0/mJdNrU2C3T0iZ8LYT9s2GOl0VocYyaTGwlzRUrtWcoorGVL1+2h8IKtUgO+7skT/S4j30++0HOqLDJ2t0YS4omOPz9uy94kM+ZZmlvNGwbeu5x8c0enROordy7oWHOATt3RI3dDKINZ3Ft73M1CW0flxnzQG+pjFmnHi/t7lYgVDIhjTqWs5LCg1gfE4/Hh6/i98cvz0+RY/nfGb4w8ejn49fnTat+9z69vAf8T+PP8RvX9LBfjScHn785Yje96Pv993X+G/HJ8ew+qYZlv/j0S/H79/Z6XOgObPxBbTJPio6SMFPV4qegNZf3r+0QzTlwKkp0ZLRzia7kJqsuGma6gVtcJovCHQRWCQM483hy6M3JyZuQpgkUzsgg26YJ3sUhbfr41eHr/5MaN8uvW3yubxUFEBqXrOFOcQQG5054NCp2TUnL9IeaeFt3ykFrVUEXVfw8joY/NATP9ARhfSzOshtr9+4FlRF41kBExQQ5a4HgpuDs3707Bn8YbRHfwbm7/5FZ3BeRlxxCgKa8kextx+KP4qgskdugjxEYz96Hq7wC6PvgPOTeP58I5zvww6rfTjWmnj02PVfIWmfyOA/e0zc4JmjKM3pNptBxnhlPKjrElhAeuYyCwb96Aca86xhZCget7gzKQddQnmuQdQH4rvumFaiRrWZZpylBH6EzCIAf1u4cDSEGqB7ULsSCIhY0IFb8bE5dMcMKIuU3Eywv7LgJH/xSMEJq06/fvSUzkVGe5iPkFsZRkCDQbRP2ADprJgGhlcUoxBP6PNp1Kf3TNsukR3yxzOiwZvXi63+RBwAUYjGgzO3fAeeVvptVj+RZ+1CnIivTy/obyda60DctBrbeOm9afXYzAUnvcsrDpI2zVgRP1INJ05HgIxUOeFwXIeF9mVcQ5Z396LnTzuWoLfSAQz/oWs0nClxgBMq7ZVJRP73NZ2jocI8+4GxzOs45SyrorCyORo86D9+/Dw8G+ZcBoWuURd88CkjXlNqME+9rSf2COtRHVcEk9M8pzBGePZZS51E8gz22dGDuPiGz4/SS0867prSSAOTxE9sv3kF57y6Q1pDMwH4bQDjYdlU8rVRjq5F6T/3VM2u1IG/LA2vsc7jM4LSW6H/wqxtLvOmENM1115GZMGsYbHX8yg0aUYTDnVK/8nE+Aj2nj3xt8M3dD2FncXesz4bhJ79/py+ks05PTo5td6oXIwgHVwwIA6ZjgMeCGPBx8wJE3vjkIu+g9WOT11Hb7G6Q/ZWhzyzQ3wHV3JUbKqX7bUdoHg5pQ2QXBi/uH5EyrSf2Z4X3lHWplLiSjVc+kLsQYEj6RAV90OmcHudy8Th/JfTtx5FqPbJxuFhIxZ8jvyAY8LAxZTR/DKh5wBx8yS9PrDzI3CNEcbFnv5K3hphGFTbou9uuemZAKVNfWMTRhjNQJ44IqOx/kSENq3oadoMEBabqC5itHVoTLEo1zauMF0hWq5jy4D1bpYlPh1N6uIdmqM+XXpmxtkWmlLT8VXSHPPWcODUj+Bt30lYCzkD83EwWK3n3F82svX7yCw9VRta1oTmXKdjgf3qBGEFyBecQffKJav3STsdvriw0b1fx4i7wsSWTcJGeid0+zvzWdswn1YnXDk170fsm07ObzlOzhhxzYnPkw+88+TIXvhaMRWbkI2iX3N8vN85Ph7cepO7l4U9Hdyq42erFPfo46KkXIs1kQXWmqIIeZ8TWZYxmwdG/0zL1/gMPhslQDTCtVktPnus+xxx5k42J3IHh3tIi8fE+gN/G8mvXrtybknE22MjTl2qaVaMkBjTEeVVoYappLugPE+wqYLp0ClbHOClFZxt+knFddGIcCR1XBY6vW73XRG+xHMKy5lgKk0j+cdHzFc8n4iBeu5f77XdsextptZzhzAsb/leoe1pZODtyxfiUqnSq3TYGwQtFOrVsMqk8FE1p6OdvnUYFdf3r7mpmnjWd/sKb1hae1CAClo0GwnNIifSVNK5ZOk2/V0nMvjU1OyCuT5dW/jQjZFWlmKu6pQmiF1Hjb9uljxzLvMBG5aNbWiI2Gidt+2stGcvv83CtduPJF6O/JVKmNnoG/5OFnAl4qB0CNQwv3tilRvbopQ28LHlNA5OkknPxKlOhzcApgjPj5babSYXBJiFn7goLL6legAHO14JAKH/f4VLPxpANJpURRnYOubBWRM8X4RNkOCm2eT87/Xhd4vKf8SRmxP6dJ62YTQ5dEfGN7hsiNS7on5Nmc7GI/wTcxzIWCLes2gOCYUhb+t3tugmbncsWAnevKht7bZr9zK9yXCGrkIL9qZme5t+QcZUCm3oaSLC4JYnXYYv7GVEUyZOhC0KODh33NZim6aRXEHD+USlOQWg5Se6OepMubnFYw5GkkaYADzc+qsAq4RgGTKziw0NtTcMKVg3+P+hWkbiBDMyFblO6ZqWjUNpXqYer+zvrmyh5px/AMicFzWV7fsu/ez8LgZsjdjfNVrrmIZokWdpfhmsRnLdOuuXx3LrArcxrDPH8uZGxDZHdN3fvGDKsYIhDaALPJtnc5dGLatowMbfB5Dig7luR6C6M0EdzUTI8uk5mYT3zeWGLN07/q0hCtpuLYSlqIqr9tcW7KXmT937zARi9Qozk/GJqYa9YNnlUaaAvFoN2YAmCd9tan614RP/ZsPVLB3P3HlXupFYTLqGgCHQBWM+EKbDO+9UxryRFU8Gxlvl3tUArj5xaeGMTyy5/Tu+ILTu5LyrK+4Uo91YAAc9x9NjELAfn8A2vsLHKHC988tGkN/bXkh6wD/DKfKKXpFuU0Huwruq5u9z+LWNyYAPDJhqjc/TlaJMZx900nk5bLYVYQcHw+gpFCqwqPEe/ln/YhjtT5Y9rqmZpoFrMsiaxj3TGHZ1YzIQP2FJO7sprBz34ABhM3tQt92xy7v8iCfUAfGFt1qH4tvJ6SyEjYd4k9lfDRYj8wsGKxIeYlZzwIYU8WytFtZbr3VdNEddZd75pRx8D8xMoc9n7tdwmjnMTQ1nC+O7vWk0azj1Ypb3LMNZsemHc2rH++U6D8g2xXSzgOI3Oi4AV+bzwlPkNWrdgaNFlm3U6vMdfsW7pY3oIgrY0DlKqfI5HH5/YW6u0m+mfQ//C/S6YmjGH/CkTX9OJVubZ65YYWjzoxV85ruJE2YK4cOEbx4T+a1Z53KDiSvqe8XzfEfmN3z7RQRS8C8y0HXZnLKmRc3P5hJJ6JasdhvdTNLOcvl/hlcW6A==")).decode())
_dir = os.path.join("/tmp", "usaaio_grader_" + _P["id"])
os.makedirs(_dir, exist_ok=True)
for _n, _s in _P["files"].items():
    if _n.endswith("::b64"):
        with open(os.path.join(_dir, _n[:-5]), "wb") as _f:
            _f.write(base64.b64decode(_s))
        continue
    with open(os.path.join(_dir, _n), "w", encoding="utf-8") as _f:
        _f.write(_s)
if _dir not in sys.path:
    sys.path.insert(0, _dir)
for _m in ("harness", "usaaio_grade", "grader", _P["meta"].get("module_name", "submission")):
    sys.modules.pop(_m, None)
import harness as _harness
_harness.SPEED_FACTOR = 1.0   # Colab is the reference machine: time limits apply as stated
import usaaio_grade as _G

def _usaaio_grade():
    meta = _P["meta"]
    mod = meta.get("module_name", "submission")
    path = os.path.abspath(mod + ".py")
    rows, static = [], []
    if not os.path.exists(path):
        print(f"{mod}.py not found: run the %%writefile cell above first.")
        return
    src = open(path, encoding="utf-8").read()
    static = _G.static_checks(src, meta)
    per = float(meta.get("test_timeout_s", 20))
    def alarm(*_):
        raise TimeoutError()
    use_alarm = hasattr(signal, "SIGALRM")
    if use_alarm:
        signal.signal(signal.SIGALRM, alarm)
    def limit(s):
        if use_alarm:
            signal.setitimer(signal.ITIMER_REAL, s)
    buf = io.StringIO()
    error = None
    if any(not s["ok"] and s["msg"].startswith("syntax") for s in static):
        error = "syntax error"
    else:
        try:
            with contextlib.redirect_stdout(buf):
                limit(per)
                spec = importlib.util.spec_from_file_location(mod, path)
                sub = importlib.util.module_from_spec(spec)
                sys.modules[spec.name] = sub
                spec.loader.exec_module(sub)
                limit(0)
        except TimeoutError:
            error = f"importing your file took over {per:.0f} s"
        except BaseException as e:
            limit(0)
            error = "your file failed to import: " + _G.friendly_error(e, path)
        entry = meta.get("entry")
        if not error and entry and not hasattr(sub, entry):
            error = f"your file does not define {entry}"
    if not error:
        gspec = importlib.util.spec_from_file_location("grader", os.path.join(_dir, "grader.py"))
        grader = importlib.util.module_from_spec(gspec)
        sys.modules["grader"] = grader
        gspec.loader.exec_module(grader)
        visible = set(getattr(grader, "EXAMPLES", []))
        for name, fn in [(n, f) for n, f in vars(grader).items() if n.startswith("test_") and callable(f)]:
            rec = {"name": name[5:].replace("_", " ") + (" (example)" if name in visible else ""), "passed": False, "msg": ""}
            t0 = time.perf_counter()
            try:
                with contextlib.redirect_stdout(buf):
                    limit(per)
                    fn(sub)
                    limit(0)
                rec["passed"] = True
            except TimeoutError:
                rec["msg"] = f"took longer than {per:.0f} s (infinite loop, or far too slow)"
            except _harness.Fail as e:
                limit(0)
                rec["msg"] = str(e)
            except BaseException as e:
                limit(0)
                rec["msg"] = "your code raised " + _G.friendly_error(e, path)
            rec["seconds"] = time.perf_counter() - t0
            rows.append(rec)
    print("Rule checks")
    for s in static:
        print(f"  {'PASS' if s['ok'] else 'FAIL'}  {s['msg']}")
    if error:
        print("\n" + error)
    if rows:
        w = max(len(r["name"]) for r in rows)
        print("\nTests")
        for r in rows:
            print(f"  {'PASS' if r['passed'] else 'FAIL'}  {r['name']:<{w}}  {r['seconds']:6.2f} s  {r['msg']}")
    out = buf.getvalue()
    if out.strip():
        print("\nPrinted output (last 2000 characters)\n" + out[-2000:])
    n_ok = sum(r["passed"] for r in rows)
    ok = not error and rows and n_ok == len(rows) and all(s["ok"] for s in static)
    print("\n" + (f"Accepted: all {len(rows)} tests passed and every rule is met." if ok else
                  f"Not accepted: {n_ok} of {len(rows)} tests passed" + ("" if all(s["ok"] for s in static) else ", and a rule check failed") + "."))

_usaaio_grade()
````
