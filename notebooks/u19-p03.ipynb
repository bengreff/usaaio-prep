
- `z1`, `z2`: float tensors `(N, d)`, $N \ge 2$: the projection-head outputs for the first and second view of the same $N$ images (row $i$ of `z1` and row $i$ of `z2` come from image $i$). They are **not** normalized.
- `tau`: a positive float.
- Stack the views as $z = [z_1; z_2]$, shape `(2N, d)`, and normalize every row to unit length. With $s_{ij} = z_i^\top z_j$ and $p(i)$ the index of $i$'s other view ($i + N$ for $i < N$, $i - N$ otherwise),

$$\ell_i = -\log \frac{\exp(s_{i,p(i)}/\tau)}{\sum_{k \ne i} \exp(s_{ik}/\tau)}, \qquad \text{loss} = \frac{1}{2N}\sum_{i=0}^{2N-1} \ell_i .$$

- Return the 0-dimensional tensor `loss`. Gradients must flow to `z1` and `z2`. Do not modify the inputs.

**Example.**

```python
z1 = torch.tensor([[1., 0], [0, 2]])          # rows normalize to (1, 0) and (0, 1)
z2 = torch.tensor([[4., 3], [3, 4]])          # rows normalize to (0.8, 0.6) and (0.6, 0.8)
nt_xent(z1, z2, 0.5)                          # tensor(0.6271): every anchor has the same loss here
```

**Restrictions.** Only `import torch` and `import math`. No loops or comprehensions. `F.cosine_similarity` and `nn.CosineSimilarity` are not allowed; `F.normalize`, `F.cross_entropy` and everything else in torch are.

**Constraints.** $N \le 1024$, $d \le 256$. Values of $\tau$ down to 0.05. Time limit 1 second per call on CPU.
