# Lasso by proximal gradient descent (ISTA)

Train lasso with proximal gradient descent. Plain gradient descent with $\lambda\operatorname{sign}(w)$ makes useless weights jitter around 0; the proximal step sets them exactly to 0.

Implement

```python
def lasso_ista(X, y, lam, lr, steps):
```

**Objective** (no intercept):
$$F(w) = \frac{1}{2n}\lVert Xw - y \rVert^2 + \lambda \lVert w \rVert_1.$$

**Algorithm (ISTA).** Start from $w = 0$ and repeat exactly `steps` times:

1. $v = w - \eta \cdot \frac{1}{n}X^\top(Xw - y)$ (a gradient step on the squared-error part only);
2. $w = S_{\eta\lambda}(v)$, where the soft-threshold is $S_t(v)_j = \operatorname{sign}(v_j)\max(|v_j| - t,\ 0)$.

Here $\eta$ is `lr` and $\lambda$ is `lam`.

**Inputs.** `X`: float array `(n, d)`; `y`: float array `(n,)`; `lam` $\ge 0$, `lr` $> 0$: floats; `steps` $\ge 0$: int.

**Outputs.** Return exactly `return w, history`: `w` the final weights, shape `(d,)`; `history` a float array of shape `(steps + 1,)` with $F$ at the start ($w = 0$) and after each step. Do not modify the inputs.

**Example.**

```python
X = np.array([[1.0, 1], [1, -1], [1, 1], [1, -1]])
y = np.array([3.0, 1, 2, -2])
w, history = lasso_ista(X, y, lam=0.8, lr=0.5, steps=2)
```

gives `w = [0.15, 0.525]` and `history = [2.25, 2.05125, 2.0015625]`.

**Restrictions.** Only `import numpy as np`. Nothing from `np.linalg`. Loops are allowed over the steps (not needed inside a step).

**Constraints.** $n \le 2000$, $d \le 200$, `steps` $\le 500$, and `lr` is at most $1/\lambda_{\max}(\frac{1}{n}X^\top X)$ (so $F$ never increases). Time limit 1 second on CPU.
