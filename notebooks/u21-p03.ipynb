# Parameters of a damped oscillation

An open-ended parameter-regression problem in the spirit of 2026 Round 2 Problem 5 (one curve family instead of a mixture; lesson u21-l02).

Each group holds 200 noisy measurements of a damped oscillation,

$$y = A\, e^{-\gamma x} \cos(\omega x + \varphi) + \varepsilon, \qquad \varepsilon \sim \mathcal{N}(0, 0.05^2),$$

at random positions $x \in [0, 10]$, sorted. Every group has its own parameters, drawn from $A \in [0.5, 2]$, $\gamma \in [0.05, 0.4]$, $\omega \in [1, 6]$ (radians per unit of $x$) and $\varphi \in [0.5, 5.5]$. Recover them.

- `X`, `Y`: float32 arrays of shape `(N, 200)`; row `n` is one group's positions and measurements.
- Return an array of shape `(N, 4)` with columns $(A, \gamma, \omega, \varphi)$ in the labels' form: $A > 0$ and $\varphi$ in $[0, 2\pi)$. Do not modify the inputs.

**Data.** `train.npz` (300 groups) and `val.npz` (100 groups) come with the problem (in its Colab notebook, the Data cell saves them next to the notebook; load them with `np.load`). They hold `X`, `Y` and `params` (shape `(n, 4)`). The grader calls `predict` on a public set of 200 and a hidden set of 300 groups from the same distribution.

**Metric.** The RMSE over all predicted parameters of all groups, $\sqrt{\frac{1}{4N}\sum_{n, j} (\hat p_{nj} - p_{nj})^2}$, as in the official problem. Predicting the training means gives about 1.02. You pass with RMSE at most **0.03** on the hidden set (0.035 on the public set). The public test prints the normalized score between that baseline and the reference solution.

**Restrictions.** You may import `numpy`, `scipy`, `math`, `torch` and `sklearn`.

**Constraints.** `predict` on 300 groups must take at most 20 seconds on CPU.
