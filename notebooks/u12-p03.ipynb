# Max pooling forward and backward in NumPy

Implement max pooling with non-overlapping $k \times k$ windows (PyTorch's `nn.MaxPool2d(k)`: stride $k$, no padding) and its backward pass, in NumPy without loops.

```python
def maxpool_forward(x, k):
    ...
    return out, cache

def maxpool_backward(dout, cache):
    ...
    return dx
```

- `x` has shape `(N, C, H, W)`; `k` is a positive int. `out` has shape `(N, C, H // k, W // k)`: `out[n, c, i, j]` is the maximum of `x[n, c, i*k:(i+1)*k, j*k:(j+1)*k]`. Rows and columns that do not fill a whole window (when $k$ does not divide $H$ or $W$) are ignored, as in PyTorch.
- `cache` is anything you choose; the grader passes it unchanged to `maxpool_backward`.
- `dout` has the shape of `out` and holds $\partial L / \partial \text{out}$ for some loss $L$. Return `dx` $= \partial L / \partial x$, shape `(N, C, H, W)`: each `dout[n, c, i, j]` goes entirely to the position of the maximum of its window, every other entry of that window gets 0, and ignored rows and columns get 0. If several entries tie for the maximum, the gradient goes to the **first** one in row-major order (top row first, then left to right), which is what PyTorch does on the CPU.
- Do not modify `x` or `dout`.

**Example.**

```python
x = np.array([[1., 3, 2, 0], [4, 2, 1, 5], [0, 1, 7, 2], [3, 8, 2, 6]]).reshape(1, 1, 4, 4)
out, cache = maxpool_forward(x, 2)      # out[0, 0] = [[4, 5], [8, 7]]
dx = maxpool_backward(np.array([[[[10., 20], [30, 40]]]]), cache)
dx[0, 0]
# [[ 0,  0,  0,  0],
#  [10,  0,  0, 20],
#  [ 0,  0, 40,  0],
#  [ 0, 30,  0,  0]]
```

**Restrictions.** Only `numpy`. No Python loops or comprehensions.

**Constraints.** $1 \le k \le \min(H, W) \le 64$, $N \le 16$, $C \le 64$. Time limit 1 s for each call on `x` of shape `(16, 32, 64, 64)` with $k = 2$, CPU.

**Hint.** If $H$ and $W$ are multiples of $k$, `x.reshape(N, C, H//k, k, W//k, k)` puts window row and window column on axes 3 and 5; move them next to each other and flatten them to one axis of length $k^2$. `argmax` returns the first maximum.
