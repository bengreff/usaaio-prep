# Logistic regression by gradient descent or Newton (2025 style)

This is the coding part of 2025 Round 1 Problem 3, with the rules made exact. The design matrix `X` already contains a column of ones (as in 2025, the intercept is one of the coefficients). With $z = \sigma(X\beta)$ applied entrywise, the loss is the **summed** log-loss

$$L(\beta) = -\sum_{i=0}^{n-1}\Big[y_i \ln z_i + (1 - y_i)\ln(1 - z_i)\Big].$$

- `"gd"`: $\beta \leftarrow \beta - \eta\, \nabla L(\beta)$
- `"newton"`: $\beta \leftarrow \beta - \eta\, \big(\nabla^2 L(\beta)\big)^{-1}\nabla L(\beta)$ (Newton's method with a learning rate)

**Initialization.** If `solver` is not exactly `"gd"` or `"newton"`, raise `ValueError`. Otherwise store `solver`, `lr` ($\eta$) and `num_iter` under those names and set `self.coef_ = None`.

**Training.** `fit` gets `X` of shape `(n, d)` and labels `y` of shape `(n,)` (0s and 1s). Start every call from $\beta = 0$, make exactly `num_iter` updates of the chosen kind, each computed at the current $\beta$, and store the result as `self.coef_`, a float array of shape `(d,)`. Return `self`.

**Prediction.** `predict_proba(X)` returns the probabilities $\sigma(X\beta)$, shape `(m,)`. `predict(X)` returns integers, 1 where the probability is at least 0.5 and 0 elsewhere. `score(X, y)` returns the accuracy of `predict(X)` against `y` as a float. No method modifies its inputs.

**Example.**
```python
X = np.array([[1.0, 0.5], [1.0, -1.0], [1.0, 2.0], [1.0, 0.0]]); y = np.array([1, 0, 1, 0])
MyLogReg("gd", 0.1, 3).fit(X, y).coef_        # [-0.01763091, 0.4606836]
MyLogReg("newton", 0.5, 3).fit(X, y).coef_    # [-0.58730214, 1.85768779]
MyLogReg("newton", 0.5, 3).fit(X, y).score(X, y)   # 1.0
MyLogReg("adam", 0.1, 3)                      # raises ValueError
```

````python
#@title Grader: run this cell to grade your solution { display-mode: "form" }
# The grading code is hidden on purpose (double-click to see it). It writes the site's test harness and this
# problem's grader to a temporary folder, imports your file the way the contest grader does, checks the import
# and ban rules, runs every test (including hidden ones) and prints a table.
import ast, base64, contextlib, importlib.util, io, json, os, signal, sys, time, traceback, zlib
_P = json.loads(zlib.decompress(base64.b64decode("eNq9Wwt32zaW/itY95wxmZEZWUm6HafOjts6bebkdZLMjvfYPiwkQjRqilQJKrbi8X/f714AJKiHGyfdTU8tEQTuC/cN6GZHZzsHYmcx/M+9+fDxzkDszFQjMXSz/qbRTaFo7GWVa9PoiahVXitjdFWK8VLktcy0KhuRKTOhz6oWr9VVg7fRaDh6IkyzLFRMoAqsqkoPvxg+Ycy6XDTKYHQ0xGOmp1M9WRTNEiOPMACQNX3febUEBe9UzouqbFGotJQzJm1eV+NCzR6nZjGeaSaNZsmiqK5UlurZvKobQnG6Uy5m8+XOOd6OZVmqjAd1+ZHmz91nYRrzO/OuSlPVpio+qu7RTTITDUBE4PVcNx5iWlTVnDBNZWEUrzFN2uiZqhZNSi8eEZeNzC05RQVZ5knJAmPcduCCBCzDkTxjEpfNRVUmk0IaA6ERVlMt6gmLQWZy3qhsj8S+925/7+0jXiLrZoTXTb1Qt3ie6oLFfbNzIesSeJI5y/eM//tFFXNVG9FcyEY4ufIe06AVpIh+ndbVTLjlfvTBr3FyVp6VR2660EZMqhJCax4uSt2Yh9/Tx7OHDiqe5zp79tDOBhGJeEFaNNWAKkhuYrooJw020wja6YwH0weWtkZeKnxTZ2W36cLqhfgVQ78m4vjk6NXbl8fvRQHFJZaUBUvAjDAX1VUpsOjdAqr6URsNquKn4j2Ba0S9AFr1UdVLXgTWnktdLKD5opbaKEGP4ko3F0IKwDMyV+Kq1g0YFlMYwQ+qPBBGLsUVkXslDd5WZT4QJUEVGRRRXOgsw3RdzheNIenZXSidSCfVfNk+kBKRfN0jK7IA1HJOo9+IH6tCjsWPb/9phKyVMKT7NcmqJFJ2jXg1Em/raiBMJYpqIguCqMvciNkCQhkrEipEVJEBarwuNMkBJt5cYCunctJUdUKYRsmQNleKfEH7H1nMBrAKtddc1EpmwswVNgyz6mqRXxRLaEsxFaAFVEyqmgRdKxJ6I/SUCMQjW6njw5kMCeX92+Pjn9LnRz9+ePNOHBJ2YvisZCPgbYiOrydqTpoSH5yVAv/meGVnQaEgcZVF0MVsIGYm93OAt6xIyGXmRuhft7kRze2ApOU8uvZr4ZOCNX6HqnpyEYxOwb8uTSPLiYquB/Z98oG9SBwsZ7SqWdSluE4yeOLJRRQnk/kCf3mbo9hOVsymeMHojuu6qgMoluUAVjlPpJF1LZegu+PDXMi5ivKqGUApS/oL/Tw824Fmw+2e7XjKDGTdLOaFiohzzI8TXupoYZlijp9EsOKBmJ7t3BDAW2y4sbjEjYHfgZtUE3gncRPMvwW6jrJJUZm7KBuIuqmKw3219+1ASPf1O08vLOsKJLfESgQeMDotKkmE0Tjj7L8ImMktf8TSlf26hR3LgJseb+LNv2L2WgTYEG3g3nSD1aCjKKKYwiVpYfjuyr0L0ZMnlZo9YUlrdDltgTtFpv0uCidECMOJi/44cdGfUPFUXUNiLAgibiav6UOOTZSLPXEVxzHBBqP6kxIKAU0Myfg2mEpfTuzoxEdZwD+IqJB1Tm5XkcKKG3wcJI/y27i39+r3hSw+Qyv7RmWnEtFlxpo+iFmiIEI2Te3en+3wdnRANqhLoCDdnP9ztQhVg+lPnRxA3JoxhVJtIdCuHKxSDGUgapn5Dgac8Q1e/UfdI41mYai3GdqkbCK8HfTNbN0PRhjsSLdkjX4ahSgc4DhJOX9L0x45Odwf7ZzDuusn7XqofUpTU8o5Innz/+uai6pU93TKQOm18ppUkmP7ToyQmLH19umJMj2BIClxiUN6+sz1qCOAUabU/PADMr1AwRyBHyC+VfLWAcS9EEJDCcHkd9c9wWNbIjkQ4y+Xu/wsudsp1i4Y35eKXZLYGY7puwKHSFokJhrHm3wNlnd+ZuPy1k+M3Vfe275lEwcD6+2g2WW4Vx4OQwiSlyqdLWAUyG6iaTkQD+BOYRkPHlxe0TdPiU0df0QEENOSMU8pQSUekOMgC8xhXrJEyljnixkVaroU80JOVOKzTo76ZE/wi6edZcmY81lJCwjjuZ2J7IyCRxmtEmRf0xINRQbDJqalCqmMqiXi2yc9j+waxtFTcPYnrXI5J7JEiQNdzBQl+HqqKbOklLnj5ZsbLf4q9m+fiquqvqSsXrLyCto+5KOtv3RCBvGdhCmJ12VkFKWCoOpzpfw0FDG8a5hyb8ikuRxFPDTIvcEBCYhz3d12JZJ3W0MRqjcAUVMJhgJpZoglzsNfyYmYAAEtAZM266ZqhRN7Sg5QXeyNFxk8KQ1zDo60XFMBtZjDnERzpSeKAEJCGhG6mjoLBs1eooZViEamyJcofBMErFjgXSI+0AtdYxjMFLuEguKCRUYSVllPrZohZZOAkKC6nKYMRtXe2/yBKmXN5sXIURqXjjh+D4XbRPFQhGVDp5IpqWJN1hCNQr3DJgLN94cO0oorGqOmueyG5GggLkfA1vePTPJgZbDHyR+LgukkQYxYDKNgmMWAii3KKMRuEUeYb4YMsR01VcV6g+I/aw6S0fQWWT60TD11VVerxKxMMDsGJBYl1fM3TjRuHdeQKOz8Fm/4xyhRWaIMnkK1ckpgbZF54/aJAZF13GmgdZnDOlXmN6wrcLCTWTWDtKcS6pe2E8+4dbIwUuoq5Q5Dr8vxM43AAkBZ1zzQ7DQudJGR1U2oM4KCWZMt0Mx/vH/zGrSWbDmmyUAh26ptyYi9F8LjEd+7Fkea6fqZ+L7DgZd4hlX+m8eaZ2slP4wtKP+xsddNocftkP3ASLJodNENV+3X30zVNQnmsrkIlxudl7JbZpZmpb3gH2rEhbGcXLYj0OKSmgW218DNL+hG6+4wcKmbvUJhmthPvk32XU5nDqhPNKPW34BdCqoLF3y885tLDSnnZ6XHkUx1AY32j9HZDsiuakUVoOu0IBNNHvyzvCyph8MNulpU3AEAQj2nXUseuEzxl+N3x7AcJ4vkLT6jNKU2WJrGCRwXLUdWN5c1KDsrIZWEJifQVlU30RDGgPSBwMQAuKn3xb0g8Q1i9e/yQBw/Ho64E/byX0f/8z794ej16+OfQMHN2U7qmpFpSsyAMb+f9lFdq4n7hjzffptUszloPdu5PStfvnnzNn395qfj9wAXQVeS5xVslb78C4qr7Ncjsywn7YuXyCF/BAz79F4FDz8hyeyeflYlxeeqPr6ex2GL5YPtxfxRlyWVqPRmEenYYgZ/g7ClWovlUtEDCrsRlNlM0smFmlyayNQT2uJG9uPtO2vvlA0jWJEcq0srnJnJIZincIuUAdoeH5uNaXbb3eFkyLUSKVA1tbZtxV6IskHo9HxTNtvUCokdSwlagvoahPbz0PdL1OfXnIdSU06t54enjuwDaAv3hh31B+QmDS93JXJVWj9zoxL6LKvbA/qO2Sh+bh2BrrlNIQHySuCYYScrHW+S0em5I3Qs23GOk40POdwOTReGgQXjY9hDanulbpxUHCkjbFdnMBXKxGqbKFjAgGDnR1TRUNKtx9TcBzuXnLupIiHFdgi4Vx7IfCab2aIIR6bcHYGrJt8M4V/J4jKizVgJ3UGaTtOtPtvSYLWqaFNZmpgwuQfrAcy1VJAvR5InUVI+o9aCfUyADYKkzDGaIfOEO6JijoDPmFi7EfHB5tgY7EUis8zhiNcnd3sTzEvMvNC03YTzdG//fMPCbvOSxTyjrHvD2rBILO4W43N4vVVusKHYK5aja7yD/c6gNsiSVzhB4vtXSHFVgoC2IoVAdD0JbNKAlaXbhPdFS5nP+wr9yNvOKu8rCsF0kKF9FtTXgUe+C6DOtnO1edImjF2wWsXJhp/I+VyVWeQ7PnbR2Y51eda9/1dfXpuwcBT8QZdvXBQ7WuRHhmJQbHsrKwuSys17JZtXSBnXdNr6oHsSB8IChQxAUpLoQG11/z4LsFZiVZ5G2SBoXvJbReUpO9gowBPHm73LXymiQDAe1M3uQOxaIN6iyAB3gQ91b75Lrdjbzf3ETfRTy2ID+cbT3gH7pj1BHMvSkGwnFy5CcCrYBQhuPyuJIvaEP+OnthotruTS7NmjWlQyyAt1C4iU0kGrymI58DhRuXShJrJaYJtILpWipGMpCpWjNL+itEnwzCRJ4iCYISkTWUWbYgFfaK4v7S5EQdD8CwfHIApbcl3whRr+W0RBLP2L6OWFgQoBw71VBxQY4eRD8CMTb9AdQI6/ZofLyuNoj2S7XQblfeZTa0RpNbf5ZL/bRtOtka2Y3ucwTBUl9OLvwsOmkjWwGxHZzMlhOB2eh8q9gf37iKDDehfz7NzWWObRL2PYpkms7D1GMUxmTEpdqwtVchHb5o686s/l33F2e0elPq01wBTLlDPZCCBQ6KZUTfUz+jel2mM66dZIrbmqEHOoKuw7F+4kmidof3xNBRvMVdE8w4f1/Q7TmNoqvm5NUDfje5M240gladq+QMnnkk0CjkxzykF9SngAAjs1pfJT+ZyPKr+WBZeVXlHLhBpW5GW5N+mUDn8pHQuy9t4QbQUrDM3m47SOfCdMOjHhExIVnI9Q9n97w1hvab4X9gyZeNTWYvC0q1UuVbKyzj+etimiZ2XrzNF5UBI7SVECfijaKY98ug69xzg1HJKikpmJIibiIakN2QS3InYIHmIWdTIi7+xQPhFER0w4wUuDu4mHYt0ubMFIQ6fn9EylnukebWPGqqyrpEkTaeB1hS0I4Z920M6JnC2laGvolMGSCZpTS9U5Rwt6Yis5D9NZX9L5fNZw33ENbegluGsRsTjpcoiJ7PQwylst8S3PKYjWVfK+ofbJizdeeLbJk9iPyD29f/Hz0ct3rwauSndT4c5Sbu76U+DAl/VvLpEsR0NPS7865kswXbMK25npWsH47GZEoHQ1zfI0qkYTitqT+eLDi1fH79J3x0cvBy11K2mOmSvSnn4rLKHRlHoztrND3Uk+Mgk4Cm6NWdXoGnPYpsBVreBbjNfROViMkFBH9Gd1ISzGzjOn9JoLhnOr+OsssQ2pOqEeUGqXkeeJv0Bww35vwrVdei0Jp4jOOM6tN7M8kgdmr8Yet6mqSz5TEDd+Pw6SIXWAvetyWH6QRrUdItsD6VpiP7w8Hg73OwLuxcY2it3BEJNJJzF0Q6xyG+WSoDviUWvWa6A3Ccr7FcILfSaN4pP4CAX44+FweHD+xYbMNxz7fRweCu932Dn+rNifaIKXgX0V//HmdrLKKmWTJnvZDtGFQNyGdfs9WcjvZ5NnO/bSH1lhGzLae4At3+4e4R/bXh4YX77Nliw0N8td93ONLV/l2SnkGfzVQZ/EO7dHsYaShgiFytQ6dvpGvv2jrI3HkehGzUzEt2fKflRgr+rP/ekARYKMaBqHDS94CTqE5JSEMK4kzxyOSfvrBfyWVQrKVvgWQelZW3G45J51uVDh/k1sgLX+8ICBnD45OIf/5nNhUOu61ILd49mOA+0nB9joLXWDqezZVKY4f2sPfWhomAxv73Mmtn7x4QuiztdGHnc4t+6Vv8iluU047SRHlkqJ9/rdjTUP3i22uYe1cXbWRVXm/lxm1WeLSJf2lhln8wMqIKayFv5kMA6dgEPOhxsrPe2vZLilmdLrDXdV7hVL/ixawnsGfF6RfWYM6WC1Ck7wamhxFm07qB2IR5vCm0tnz31BVreO7d6hqF1gc9VDKh2jLne9KytlpzKuqiJaI8y+I1BNT3MJXNMD167wBcrWkMLXmqfCFzzWu6Up1Td0UHbgjweo3LEnu22woDPdTTeit53SDfhEe2Dvmg6Eu8/nLrwNwss2A3cvZEDHz0Rhe50cAcA5cnUtZ/PCZZM8Qu05ex4JzoNjLp1Hzcrx9TB5Ih6IaB8qVs6TRpYXUYNIOIrjTkW/4UIYq2eVBtX+frmk1h1s/Eoug1NyNUWFSI3Ok4FYDtyx6EAU+B+ySWH07R2rsa3dgPeTqisTndgbS12huHJXYnU9/ftE9gu+TsTfGV6gzjkdT54Qd5+g7Mv49GDA5ReKS7OYRfJam8Nhv7HrDnF56+3PIdqN9/9+sRQrXRIQpKsD/dtAX+49+41PAz9Zae6JTwhWEMHJirnndjkqcVnkiS1xfxmIPJjmxMIfexAcIOa9PaM3wYVlKhVtFrZd5HiSY9Xeu21kbn31DQ/f+kVi98Z+ud31PnhmS4XE//Il2rij4Q1ml8fMONg6LbSSj7mZYTGRAfemFv1pxYYpHl1vYstiIGjuYMj89gDmTIEG5sy/LzBNRb9KaDmweUsLdSsbk0pNXRrCtRpsEdmTNkzFIEBnVDFNeLZHKPk+BU9sEdhT3vBO3SyZ6sZuYEgETQSWWYgCEy1spw2EMri4QvX4ZxmhS+vdLduZpTq8XhgidQLwxT3f3m+XBIY7iLevspfC21UgcsPcjdfr/+a995rU5pRwId2iBF521mZRIbPwrgHIepT11m3Buu+dA11K2oK4Q9neoHa/dsCiOMlItsmlpn5+SYcUC8K1RkZvPxGgFFIns3ZtP7zBCegkd8uZeHaItML/pADr4028ilx/VP4CN5t9h0JOJmscGvqFzLpK0sV826bBIsoi7EO0nRgy0mXMJyoRnUR8L0i4o57ZEC7fReczl8lkUcvJcvVG1qzze5lsZEQKPRAISnyf8XA/GXoX95zsICmresb9p0/qkOqlDATvg4YHdoGdemJ98qQqFrMSmbuEPz3FAEzWRCVk+dwHJKqQo9pd+MIr8NJqWA9XBkZDGfSYIJF2fIQxnLP6gx5VVptOT8HagKL1OWpB/r6Hv+3DKPiOquY8JLgDso+XA0F//fu14IESmA+NhgmmPbLXgVx6sX2JD5VM4IZlPVa75CTklkI93pCZ0BWSTM6su/35J/v5ukVBddzdt9F74QpAPTfrKf5/U8q66T76eqXqVtyrIOj/NqUl6QY08Q8gHFk+0V9vuj8VrbNnUB29vV9vfC6e7ZB6O2TVO82zcIOoBUI3Gx8P94Ps7LLLzh6Hm0KagQXeRinprhPv2SL6ke1oOCQ3tfpmIL6NwwbPVgW1bqdOFkjaYXgRtH64z8o/2gAXWrlvEUJA7geP3HO+ueQr21uFYFV7syBGXy2IR18liM7s1oXBovjbZsD7o/sIwmJJayVBgklnyKdmyHw3SsSb2ArXj4bg83HcJZR3MsNOjOID5YSYvE9PyKTUvAtYlL6eJB/gdyPvgF1iEXOuvxI56cdkMf+0zIWf77hXxLfI/O/ELa6YfpK7KDKButedsVE8an9gTj85HVNaNtygMgq5Gf4a1ZjN8vEyOAFPy/1AQk+CUvwErC9HwcvHwcu1dHzN+7ZyppQyspj6Qwx/a1LmM0eetcG7P467LhonoxAepjC1Lo0TUZCk2hSYS+FhvC40/mFeCmNL6Y5/KLZASVbjYBj6enFwmIwQ+wZrMW94Hm9gZRTWDMTUjH5TMZHwj2M1pZTE/jq5y0LanXyyVdPJpPd9QgI0n7NxTPsjn2+6n4KMaHC9MNi+Y5stCWDZlEjOzFtX7NmZu3cn399tZ8CnCmD6Cenwn8TAOthtDOTZZxC/c3v7vxtLGlc=")).decode())
_dir = os.path.join("/tmp", "usaaio_grader_" + _P["id"])
os.makedirs(_dir, exist_ok=True)
for _n, _s in _P["files"].items():
    if _n.endswith("::b64"):
        with open(os.path.join(_dir, _n[:-5]), "wb") as _f:
            _f.write(base64.b64decode(_s))
        continue
    with open(os.path.join(_dir, _n), "w", encoding="utf-8") as _f:
        _f.write(_s)
if _dir not in sys.path:
    sys.path.insert(0, _dir)
for _m in ("harness", "usaaio_grade", "grader", _P["meta"].get("module_name", "submission")):
    sys.modules.pop(_m, None)
import harness as _harness
_harness.SPEED_FACTOR = 1.0   # Colab is the reference machine: time limits apply as stated
import usaaio_grade as _G

def _usaaio_grade():
    meta = _P["meta"]
    mod = meta.get("module_name", "submission")
    path = os.path.abspath(mod + ".py")
    rows, static = [], []
    if not os.path.exists(path):
        print(f"{mod}.py not found: run the %%writefile cell above first.")
        return
    src = open(path, encoding="utf-8").read()
    static = _G.static_checks(src, meta)
    per = float(meta.get("test_timeout_s", 20))
    def alarm(*_):
        raise TimeoutError()
    use_alarm = hasattr(signal, "SIGALRM")
    if use_alarm:
        signal.signal(signal.SIGALRM, alarm)
    def limit(s):
        if use_alarm:
            signal.setitimer(signal.ITIMER_REAL, s)
    buf = io.StringIO()
    error = None
    if any(not s["ok"] and s["msg"].startswith("syntax") for s in static):
        error = "syntax error"
    else:
        try:
            with contextlib.redirect_stdout(buf):
                limit(per)
                spec = importlib.util.spec_from_file_location(mod, path)
                sub = importlib.util.module_from_spec(spec)
                sys.modules[spec.name] = sub
                spec.loader.exec_module(sub)
                limit(0)
        except TimeoutError:
            error = f"importing your file took over {per:.0f} s"
        except BaseException as e:
            limit(0)
            error = "your file failed to import: " + _G.friendly_error(e, path)
        entry = meta.get("entry")
        if not error and entry and not hasattr(sub, entry):
            error = f"your file does not define {entry}"
    if not error:
        gspec = importlib.util.spec_from_file_location("grader", os.path.join(_dir, "grader.py"))
        grader = importlib.util.module_from_spec(gspec)
        sys.modules["grader"] = grader
        gspec.loader.exec_module(grader)
        visible = set(getattr(grader, "EXAMPLES", []))
        for name, fn in [(n, f) for n, f in vars(grader).items() if n.startswith("test_") and callable(f)]:
            rec = {"name": name[5:].replace("_", " ") + (" (example)" if name in visible else ""), "passed": False, "msg": ""}
            t0 = time.perf_counter()
            try:
                with contextlib.redirect_stdout(buf):
                    limit(per)
                    fn(sub)
                    limit(0)
                rec["passed"] = True
            except TimeoutError:
                rec["msg"] = f"took longer than {per:.0f} s (infinite loop, or far too slow)"
            except _harness.Fail as e:
                limit(0)
                rec["msg"] = str(e)
            except BaseException as e:
                limit(0)
                rec["msg"] = "your code raised " + _G.friendly_error(e, path)
            rec["seconds"] = time.perf_counter() - t0
            rows.append(rec)
    print("Rule checks")
    for s in static:
        print(f"  {'PASS' if s['ok'] else 'FAIL'}  {s['msg']}")
    if error:
        print("\n" + error)
    if rows:
        w = max(len(r["name"]) for r in rows)
        print("\nTests")
        for r in rows:
            print(f"  {'PASS' if r['passed'] else 'FAIL'}  {r['name']:<{w}}  {r['seconds']:6.2f} s  {r['msg']}")
    out = buf.getvalue()
    if out.strip():
        print("\nPrinted output (last 2000 characters)\n" + out[-2000:])
    n_ok = sum(r["passed"] for r in rows)
    ok = not error and rows and n_ok == len(rows) and all(s["ok"] for s in static)
    print("\n" + (f"Accepted: all {len(rows)} tests passed and every rule is met." if ok else
                  f"Not accepted: {n_ok} of {len(rows)} tests passed" + ("" if all(s["ok"] for s in static) else ", and a rule check failed") + "."))

_usaaio_grade()
````
