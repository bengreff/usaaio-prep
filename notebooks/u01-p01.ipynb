Students with equal scores are ordered by name in increasing Python string order (A to Z for lowercase names). Names are distinct.

Do not modify `records` or any of the dicts in it.

**Example.**

```python
records = [
    {"name": "dee", "score": 70},
    {"name": "ben", "score": 85},
    {"name": "cy", "score": 90},
    {"name": "ana", "score": 90},
]
rank_students(records)   # ["ana", "cy", "ben", "dee"]
```

`ana` and `cy` tie at 90, so they are in name order. `records` is unchanged afterwards.

**Restrictions.** No imports.

**Constraints.** $0 \le$ `len(records)` $\le 10{,}000$. Time limit 1 second on CPU.
