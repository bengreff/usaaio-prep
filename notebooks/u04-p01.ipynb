A **gradient check** compares a hand-derived gradient with a numerical estimate. Write the numerical estimate.

Implement

```python
def numerical_gradient(f, x, h=1e-5):
```

- `f` is a Python function that takes a NumPy array with the same shape as `x` and returns a float.
- `x` is a NumPy array of any shape: `(d,)`, `(m, n)`, or more dimensions. It may have an integer dtype.
- Return a float NumPy array `g` with the same shape as `x`, where each entry is the **central difference**

$$g[i] = \frac{f(x + h\,e_i) - f(x - h\,e_i)}{2h},$$

and $x \pm h\,e_i$ means `x` with only entry `i` changed by $\pm h$ (entry `i` in any number of dimensions).

Do not modify `x`. Call `f` only on arrays of the shape of `x`.

**Example.** For `f = lambda p: p[0] ** 2 * p[1] + 3 * p[1]` and `x = np.array([2.0, 1.0])`, the result is approximately `[4.0, 7.0]`.

**Restrictions.** Only `import numpy as np`. `np.gradient` is banned (it estimates something different: slopes between neighboring samples of an array). Loops are allowed.

**Constraints.** `x` has at most 400 entries, all of absolute value at most 10. `f` is smooth near `x`. Results are compared with a tolerance of about $10^{-6}$, so the one-sided difference $\frac{f(x + h e_i) - f(x)}{h}$ is not accurate enough.
