- `box_iou`: `a` is a float tensor `(n, 4)` and `b` is `(m, 4)`, boxes in `xyxy` format $(x_1, y_1, x_2, y_2)$ with $x_1 < x_2$ and $y_1 < y_2$. Return the `(n, m)` tensor of IoU values: entry $(i, j)$ is the area of the intersection of `a[i]` and `b[j]` divided by the area of their union, and 0 when they do not overlap.
- `nms`: `boxes` is `(n, 4)` in `xyxy`, `scores` is `(n,)` with distinct values, `iou_threshold` a float. Repeatedly keep the highest-scoring remaining box and remove every remaining box whose IoU with it is **greater than** `iou_threshold`. Return a `torch.long` tensor of the kept indices (into the original `boxes`) in decreasing score order. With `n = 0`, return an empty long tensor.
- Do not modify the inputs.

**Example.**

```python
boxes = torch.tensor([[0., 0, 10, 10], [1, 1, 11, 11], [5, 5, 15, 15], [6, 6, 16, 16]])
scores = torch.tensor([0.9, 0.8, 0.7, 0.6])
box_iou(boxes[:1], boxes)   # tensor([[1.0000, 0.6807, 0.1429, 0.0870]])
nms(boxes, scores, 0.5)     # tensor([0, 2])
```

**Restrictions.** Only `import torch` and `import math`. No torchvision (and no `torch.ops`). Loops are allowed in `nms`; `box_iou` must handle 2000 by 2000 boxes quickly, so vectorize it.

**Constraints.** $n, m \le 2000$. Time limit 1 second per call on CPU.
