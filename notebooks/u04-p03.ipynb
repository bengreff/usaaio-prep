# Softmax regression: loss and gradients, no loops

A linear softmax classifier maps each row $x_i$ of `X` to logits $z_i = x_i W + b$ and is trained with the mean cross-entropy loss. Compute the loss and its gradients by hand.

Implement

```python
def softmax_loss_and_grad(X, W, b, y):
```

Inputs:

- `X`: float array of shape `(n, d)`, one example per row.
- `W`: float array of shape `(d, k)`; `b`: float array of shape `(k,)`.
- `y`: integer array of shape `(n,)` with class labels in $\{0, \dots, k-1\}$.

With $Z = XW + b$ (shape `(n, k)`), $P$ the row-wise softmax of $Z$, and

$$L = -\frac{1}{n}\sum_{i=0}^{n-1} \ln P_{i, y_i},$$

return `loss, dW, db`: the float $L$, $\partial L/\partial W$ of shape `(d, k)`, and $\partial L/\partial b$ of shape `(k,)`. Do not modify any input.

**Example.** `X = np.array([[1., 0.], [0., 1.]])`, `W = np.zeros((2, 3))`, `b = np.zeros(3)`, `y = np.array([0, 2])` gives `loss` $= \ln 3 \approx 1.098612$, `dW = [[-1/3, 1/6, 1/6], [1/6, 1/6, -1/3]]` and `db = [-1/6, 1/3, -1/6]`.

**Restrictions.** Only `import numpy as np`. No loops of any kind (`for`, `while`, comprehensions). Nothing from `np.linalg`. `@` and `.T` are allowed.

**Constraints.** $1 \le n \le 20{,}000$, $1 \le d \le 100$, $2 \le k \le 20$. Logits can be as large as $\pm 1000$, so the computation must not overflow: the result must be finite and correct. Time limit 1 second on CPU.
