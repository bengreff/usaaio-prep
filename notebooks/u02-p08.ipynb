# Softmax cross-entropy loss and its gradients, stable and loop-free

A linear classifier with $K$ classes computes **logits** $Z = XW + \mathbf{1}b^\top$, that is `Z = X @ W + b`, where `X` has shape `(N, d)`, `W` has shape `(d, K)` and `b` has shape `(K,)`. Row $i$ of the softmax of $Z$ gives the predicted class probabilities $P_{ik}$ for sample $i$. With integer labels $y_i \in \{0, \dots, K-1\}$, the mean **cross-entropy loss** is

$$L = -\frac{1}{N}\sum_{i=1}^{N} \log P_{i, y_i}, \qquad \log P_{ik} = Z_{ik} - \log \sum_{j} e^{Z_{ij}}.$$

Its gradients are

$$\frac{\partial L}{\partial W} = X^\top G, \qquad \frac{\partial L}{\partial b} = \sum_{i} G_{i,:}, \qquad G = \frac{1}{N}\,(P - Y),$$

where $Y$ is the one-hot matrix of the labels ($Y_{ik} = 1$ if $k = y_i$, else 0) and $G_{i,:}$ is row $i$ of $G$.

The results must be accurate for logits as large as $10^4$ in absolute value. In particular nothing may be `nan` or `inf`, and `loss` must stay accurate when the true class's probability is too small to store as a float (for one sample with logits `[0, -2000]` and label 1, the loss is 2000).
