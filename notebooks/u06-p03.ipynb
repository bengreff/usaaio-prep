# ROC curve and AUC with tied scores, from scratch

Compute the ROC curve and the AUC of a binary classifier from its scores, handling tied scores correctly, with no loops and no scikit-learn.

Implement

```python
def roc_auc(y_true, scores):
```

**Inputs.** `y_true` is a 1D NumPy integer array of length $n$ with values in $\{0, 1\}$; both classes appear. `scores` is a 1D NumPy float array of length $n$ (higher means "more likely positive"). Scores may repeat.

**Outputs.** Let $t_1 > t_2 > \dots > t_m$ be the **distinct** values in `scores`, sorted from high to low. For the threshold $t_j$, an example is predicted positive when its score is at least $t_j$. Let $P$ and $N$ be the numbers of actual positives and negatives.

1. `fpr`, a float NumPy array of shape `(m + 1,)`: `fpr[0] = 0` (threshold above every score), and `fpr[j]` is $FP/N$ at threshold $t_j$ for $j = 1, \dots, m$.
2. `tpr`, a float NumPy array of shape `(m + 1,)`, the same with $TP/P$.
3. `auc`, a Python or NumPy float: the area under the polyline through the points $(\text{fpr}_j, \text{tpr}_j)$, $j = 0, \dots, m$, computed with the trapezoid rule:
$$\text{AUC} = \sum_{j=1}^{m} (\text{fpr}_j - \text{fpr}_{j-1}) \cdot \frac{\text{tpr}_j + \text{tpr}_{j-1}}{2}.$$
This equals the fraction of (positive, negative) pairs in which the positive has the higher score, with tied pairs counted as one half.

Return exactly `return fpr, tpr, auc`. Do not modify the inputs.

**Example.** `y_true = np.array([1, 0, 1, 1, 0, 0, 1, 0])`, `scores = np.array([0.9, 0.7, 0.7, 0.6, 0.5, 0.5, 0.5, 0.1])`. The distinct scores are 0.9, 0.7, 0.6, 0.5, 0.1, so $m = 5$:

```
fpr = [0, 0, 0.25, 0.25, 0.75, 1]
tpr = [0, 0.25, 0.5, 0.75, 1, 1]
auc = 0.75
```

At 0.7 the tied pair (one positive, one negative) moves the curve diagonally, which is where the half credit comes from.

**Restrictions.** Only `import numpy as np`. No other imports (scikit-learn is not allowed). No Python loops or comprehensions. `np.trapz` and `np.trapezoid` are banned.

**Constraints.** $2 \le n \le 10^6$. Time limit 1 second on CPU.
