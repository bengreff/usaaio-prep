# Stable softmax and log-softmax along any axis

$$\operatorname{softmax}(z)_i = \frac{e^{z_i}}{\sum_j e^{z_j}}, \qquad \operatorname{logsoftmax}(z)_i = z_i - \log \sum_j e^{z_j}.$$

Return exactly `return P, L`: two float arrays with the same shape as `Z`, where `P` holds the softmax probabilities and `L` the log-softmax values. Do not modify `Z`.

Both must be accurate for scores as large as $10^4$ in absolute value. In particular, `P` and `L` must never contain `nan` or `inf`, and `L` must stay accurate where a probability is too small to store (for the scores `[0, -2000]`, `L` is `[0, -2000]`, while `P` is `[1, 0]`).
