- `clip_loss`: `img_emb` and `txt_emb` are float tensors `(N, d)`, the projected (not normalized) embeddings of $N$ matching image-caption pairs (row $i$ of each belongs to pair $i$). `log_tau` is a 0-dimensional float tensor, usually an `nn.Parameter`. Normalize the rows of both, form $S = I T^\top / \tau$ with $\tau = e^{\texttt{log\_tau}}$, and return the 0-dimensional tensor
$$L = \tfrac12\big(\operatorname{CE}(S, [0, \dots, N-1]) + \operatorname{CE}(S^\top, [0, \dots, N-1])\big),$$
where CE is the mean cross-entropy over rows. Gradients must reach both embeddings and `log_tau`.
- `zero_shot`: `img_emb` `(M, d)` and `class_emb` `(K, d)`, neither normalized. Return a `torch.long` tensor `(M,)` with, for every image, the index of the class with the largest cosine similarity.
- Do not modify the inputs.

**Example.**

```python
clip_loss(torch.tensor([[1., 0], [0, 1]]), torch.tensor([[1., 0], [0, 1]]), torch.tensor(0.0))
# tau = 1: S = I, each row loss -log(e / (e + 1)) = 0.3133, so L = 0.3133
zero_shot(torch.tensor([[2., 1]]), torch.tensor([[10., 0], [0, 1], [1, 1]]))   # tensor([2])
```

**Restrictions.** Only `import torch` and `import math`. No loops or comprehensions. `F.cosine_similarity` and `nn.CosineSimilarity` are not allowed; `F.normalize` and `F.cross_entropy` are.

**Constraints.** $N, M \le 2048$, $K \le 1000$, $d \le 512$, $\tau$ down to 0.01. Time limit 1 second per call on CPU.
