# Project points onto a subspace

2026 Round 1 Problem 2 projected one data point onto one direction. Here you project many points onto a subspace at once, the computation behind least squares and PCA.

Implement

```python
def project_rows(X, A):
```

- `X` is a 2D NumPy float array of shape `(n, d)`: each **row** is a point in $\mathbb{R}^d$.
- `A` is a 2D NumPy float array of shape `(d, k)` with $k \le d$ and linearly independent columns. Its columns span a subspace $V$ of $\mathbb{R}^d$.

Return

- `P`: the $d \times d$ projection matrix onto $V$, $P = A (A^\top A)^{-1} A^\top$;
- `proj`: shape `(n, d)`, row $i$ is the projection of row $i$ of `X` onto $V$;
- `resid`: shape `(n, d)`, row $i$ is row $i$ of `X` minus its projection.

Return exactly `return P, proj, resid`. Do not modify the inputs.

**Example.** `X = np.array([[1.0, 2.0, 6.0], [3.0, 0.0, 0.0]])` and `A = np.array([[1.0, 0.0], [-1.0, 1.0], [0.0, -1.0]])` (the plane $x_1 + x_2 + x_3 = 0$) give `proj = [[-2, -1, 3], [2, -1, -1]]` and `resid = [[3, 3, 3], [1, 1, 1]]`.

**Restrictions.** Only `import numpy as np`. You may not use `np.linalg.pinv` or `np.linalg.lstsq`; other `np.linalg` functions (such as `solve` and `inv`) are allowed. No Python loops or comprehensions.

**Constraints.** $1 \le n \le 10^5$, $1 \le k \le d \le 50$. Time limit 1 second on CPU.
