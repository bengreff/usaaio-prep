# Revenue by category after a join

Implement

```python
def category_report(sales, catalog):
```

`sales` has columns `sku` (string, possibly with spaces at either end), `qty` (int64) and `price` (float64). `catalog` has columns `sku` and `category` (strings); it may contain the same row more than once, and a repeated `sku` always has the same category. Some sales have a `sku` that is not in the catalog.

1. Strip whitespace from `sales.sku`.
2. Attach each sale's category with a left join on `sku`, using each catalog `sku` once. A sale with no category gets `"Unclassified"`.
3. Each sale's revenue is `qty * price`.

Return a DataFrame with one row per category that occurs in the sales and exactly the columns `category`, `n_rows` (number of sales rows, int64), `revenue` (sum of revenue), `share_pct` (100 times the category's revenue divided by the total revenue of all sales). Sort by `revenue` from high to low, ties by `category` A to Z, and reset the index to 0, 1, 2, .... Do not modify the inputs.

**Example.**

```python
sales = pd.DataFrame({"sku": ["A1", " B2", "C3", "A1 ", "Z9"], "qty": [2, 1, 4, 1, 3],
                      "price": [5.0, 10.0, 2.5, 5.0, 1.0]})
catalog = pd.DataFrame({"sku": ["A1", "B2", "C3", "A1", "Q7"],
                        "category": ["Toys", "Home", "Toys", "Toys", "Garden"]})
category_report(sales, catalog)
#        category  n_rows  revenue  share_pct
# 0          Toys       3     25.0  65.789474
# 1          Home       1     10.0  26.315789
# 2  Unclassified       1      3.0   7.894737
```

`A1` is listed twice in the catalog but each A1 sale is counted once. `Z9` is not in the catalog. `Garden` has no sales, so it does not appear.

**Restrictions.** Only `import numpy as np` and `import pandas as pd`.

**Constraints.** Up to 50,000 sales rows. Time limit 2 seconds on CPU.
