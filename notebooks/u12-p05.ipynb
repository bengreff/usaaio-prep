```python
def stage_shapes(model, input_shape, names):

```

- `model` is an `nn.Module` (for example a torchvision ResNet); `input_shape` is a tuple such as `(2, 3, 224, 224)`.
- `names` is a list of dotted submodule names such as `"layer2"`, `"layer3.2"` or `"layer3.2.conv2"`. `model.get_submodule("layer3.2")` returns the module with that name (the same object as `model.layer3[2]`).
- Run **one** forward pass of `torch.zeros(input_shape)` through `model`, and return a dict mapping each name in `names` to the shape of that submodule's output, as a tuple of ints.

The model must be left exactly as it was:

- remove every hook you register;
- do not change any parameter or buffer (in training mode, batch norm updates its running mean and variance during a forward pass);
- every submodule must end in the same mode (training or eval) it started in, even if the modes were mixed;
- do not create gradients: no parameter's `.grad` may be set.

**Example.**

```python
from torchvision.models import resnet50
model = resnet50(weights=None)
```
