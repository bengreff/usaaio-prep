# Distance-weighted kNN without scikit-learn

Implement the prediction step of a **distance-weighted** k-nearest-neighbors classifier, the same rule as scikit-learn's `KNeighborsClassifier(weights="distance")`, without scikit-learn.

```python
def knn_predict(X_train, y_train, X_test, k):
```

- `X_train`: float array `(n, d)`; `y_train`: integer array `(n,)` with labels from $0$ to $C - 1$.
- `X_test`: float array `(m, d)`; `k`: a Python `int`, $1 \le k \le n$.

For each test point $q$:

1. Compute the Euclidean distance $d_i = \lVert q - x_i\rVert$ to every training point (the distance, not its square).
2. Take the $k$ training points with the smallest distances. (Hidden tests never have two training points at exactly the same distance from a test point, except in rule 3.)
3. **Weights.** If none of the $k$ neighbors is at distance 0, neighbor $i$ gets weight $1/d_i$. If one or more of them is at distance exactly 0, those neighbors get weight 1 and all other neighbors weight 0.
4. Add up the weights of each class among the $k$ neighbors and predict the class with the largest total. If two classes tie, predict the smaller label.

Return an integer NumPy array of shape `(m,)`. Do not modify any input.

**Example.**

```python
X_train = np.array([[2., 3], [1, 4], [5, 1], [3, 0], [6, 2], [6, 1]])
y_train = np.array([1, 0, 0, 1, 1, 0])
X_test  = np.array([[3., 3], [6, 2]])
knn_predict(X_train, y_train, X_test, 3)     # array([1, 1])
```

For $(3, 3)$ the three nearest are at distances $1, \sqrt5, \sqrt8$ with labels $1, 0, 0$: class 1 gets $1$, class 0 gets $0.447 + 0.354 = 0.801$, so 1 (a plain vote would say 0). The point $(6, 2)$ is itself a training point with label 1, so only it votes.

**Restrictions.** Only `import numpy as np` (no scikit-learn). Loops over $k$ or over classes are fine; a Python loop over test points will be too slow.

**Constraints.** $1 \le n \le 2000$, $1 \le m \le 2000$, $1 \le d \le 20$, $1 \le k \le 25$, $2 \le C \le 10$. Time limit 2 seconds per call on CPU.
