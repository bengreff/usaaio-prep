# A layer table: output sizes and receptive fields

Write the table you would build by hand for a shape or receptive-field question.

```python
def layer_table(h, layers):
    ...
    return rows
```

- `h` is the input height (an int; the width behaves the same way, so only the height is tracked).
- `layers` is a list of dicts, one per convolution or pooling layer, applied in order. Each has key `"k"` (kernel size) and optionally `"s"` (stride, default 1), `"p"` (padding on each side, default 0) and `"d"` (dilation, default 1). Convolutions and pooling layers behave the same for this task.
- Return a list with one tuple `(size, r, j, start)` of ints per layer, describing that layer's **output**:
  - `size`: the output height, $\lfloor (\text{size}_{\text{in}} + 2p - d(k-1) - 1)/s \rfloor + 1$;
  - `r`: the receptive field size in input pixels;
  - `j`: the jump, the distance in input pixels between neighboring output units;
  - `start`: the input row where the receptive field of output unit 0 begins. Rows of the zero padding count, so it can be negative: row $-1$ is the padding row just above the image.

Before the first layer, size $= h$, $r = 1$, $j = 1$, start $= 0$.

**Example.**

```python
layers = [{"k": 7, "s": 2, "p": 3}, {"k": 3, "s": 2, "p": 1}, {"k": 3, "p": 1}, {"k": 3, "p": 2, "d": 2}]
layer_table(224, layers)
# [(112, 7, 2, -3), (56, 11, 4, -5), (56, 19, 4, -9), (56, 35, 4, -17)]
```

The first layer's unit 0 sees input rows $-3$ to $3$ (7 rows, 3 of them padding). In the last layer, unit 0 sees rows $-17$ to $17$.

**Restrictions.** Plain Python; `import math` is allowed.

**Constraints.** At most 50 layers, $1 \le k \le 11$, $1 \le s, d \le 4$, $0 \le p \le 10$, $1 \le h \le 4096$; every output size is at least 1. Time limit 1 s, CPU.
