# Find the source of a central field

An open-ended inverse problem in the style of 2026 Round 2 Problem 2 (adapted: new grid, profiles and metric; lesson u21-l02).

A single source sits at an unknown point $\mathbf{r}_s = (x_s, y_s)$ in the square $[-1, 1]^2$. It creates a **central field**

$$\mathbf{F}(\mathbf{r}) = f(\lVert \mathbf{r} - \mathbf{r}_s \rVert)\, \frac{\mathbf{r} - \mathbf{r}_s}{\lVert \mathbf{r} - \mathbf{r}_s \rVert},$$

which points along the line through the source at every point. The **radial profile** $f(r)$ is unknown and differs between samples: it may be positive (pointing away from the source) or negative (pointing toward it), and its shape is not given (do not assume an inverse-square law). The field is measured with noise on a $32 \times 32$ grid.

- `F`: a float32 NumPy array of shape `(N, 32, 32, 2)`. With `g = np.linspace(-1, 1, 32)`, `F[n, i, j]` is the measured vector $(F_x, F_y)$ at the point $(x, y) = $ `(g[j], g[i])`: the column index `j` is $x$ and the row index `i` is $y$, as in an image.
- Return a NumPy array `out` of shape `(N, 22)`: `out[n, 0:2]` is your estimate of $(x_s, y_s)$ and `out[n, 2:]` your estimate of $f(r)$ at the 20 radii $r_k = 0.05k$, $k = 1, \dots, 20$ (so $0.05, 0.10, \dots, 1.00$). Do not modify `F`.

**Data.** `train.npz` (120 samples) and `val.npz` (40 samples) come with the problem (in its Colab notebook, the Data cell saves them next to the notebook; load them with `np.load`). They hold `F`, `source` (shape `(n, 2)`) and `profile` (shape `(n, 20)`). Use them to develop and check your method. The grader calls `predict` on a public set of 300 and a hidden set of 500 samples from the same distribution.

**Metric.** For each sample, error $= |\hat x_s - x_s| + |\hat y_s - y_s| + \sqrt{\tfrac{1}{20}\sum_k (\hat f(r_k) - f(r_k))^2}$; the score is the mean error (lower is better). Predicting $(0, 0)$ and the mean training profile gives about 1.36 on the public set (1.40 on the hidden set). You pass with a mean error of at most **0.06** on the hidden set (0.07 on the public set). The public test also prints a normalized score $(X_{\text{baseline}} - X)/(X_{\text{baseline}} - X_{\text{best}})$ as Round 2 computes it, with the reference solution as the best.

**Restrictions.** You may import `numpy`, `scipy`, `math` and `torch`. A model trained on the 120 samples is allowed, but a method that uses the geometry of a central field needs no training at all.

**Constraints.** `predict` on 500 samples must take at most 10 seconds on CPU.
