# Per-customer order summary

Implement

```python
def order_summary(orders):
```

`orders` is a DataFrame with one row per order line: `order_id` (string), `customer` (string), `item` (string), `amount` (float). An order can have several lines, and a customer several orders.

Return a DataFrame with one row per customer and exactly these columns, in this order:

| column | meaning |
|---|---|
| `customer` | the customer |
| `n_orders` | number of distinct `order_id` values of the customer (int64) |
| `total` | sum of the customer's `amount` (float) |
| `mean_order` | `total / n_orders`, the mean order value (float) |
| `top_item` | the item with the largest total `amount` for that customer; ties go to the alphabetically first item |

Sort the rows by `total` from high to low, ties by `customer` A to Z, and number them 0, 1, 2, ... (reset the index). Do not modify `orders`.

**Example.**

```python
orders = pd.DataFrame({"order_id": ["o1", "o1", "o2", "o3", "o4", "o4", "o5"],
                       "customer": ["ana", "ana", "ben", "ana", "cy", "cy", "ben"],
                       "item": ["mug", "lamp", "mug", "mug", "tin", "lamp", "lamp"],
                       "amount": [12.0, 8.0, 5.5, 4.0, 10.0, 10.0, 14.5]})
order_summary(orders)
#   customer  n_orders  total  mean_order top_item
# 0      ana         2   24.0        12.0      mug
# 1      ben         2   20.0        10.0     lamp
# 2       cy         1   20.0        20.0     lamp
```

`ana` has three lines but two orders. `ben` and `cy` tie at 20.0, so `ben` comes first. `cy` spent 10.0 on `tin` and 10.0 on `lamp`, so `lamp` wins the tie.

**Restrictions.** Only `import numpy as np` and `import pandas as pd`.

**Constraints.** Up to about 50,000 rows. Time limit 2 seconds on CPU.
