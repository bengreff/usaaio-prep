# Moving average with cumulative sums

A **moving average** smooths a time series by replacing each run of $w$ consecutive values with its mean.

Implement

```python
def moving_average(x, w):
```

- `x`: a float NumPy array of shape `(n,)` (one series) or `(n, d)` (d series side by side, one per column; time runs down the rows).
- `w`: a Python `int` with $1 \le w \le n$, the window length.

Return the array `out` with

$$\text{out}[i] = \frac{1}{w}\sum_{t=i}^{i+w-1} x[t], \qquad i = 0, 1, \dots, n - w,$$

computed along axis 0. Its shape is `(n - w + 1,)` for 1D input and `(n - w + 1, d)` for 2D input. Do not modify `x`.

**Example.** `moving_average(np.array([2., 4., 6., 8., 10., 3.]), 3)` returns `[4., 6., 8., 7.]`. For a 2D input each column is averaged separately: `moving_average(np.array([[1., 10.], [3., 20.], [5., 60.]]), 2)` returns `[[2., 15.], [4., 40.]]`.

**Restrictions.** Only `import numpy as np`. No loops of any kind (no `for`, `while` or comprehensions). Banned: `np.convolve`, `np.correlate`, anything from `np.lib` (including `sliding_window_view` and `stride_tricks`), `np.mean`, `np.average` and the `.mean()` method.

**Constraints.** $1 \le n \le 10^6$, $1 \le d \le 50$, $n \cdot d \le 10^6$, $|x| \le 1000$. Time limit 1 second on CPU. Answers are compared with relative and absolute tolerance $10^{-6}$.
