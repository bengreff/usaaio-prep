1. **Frames.** Frame $j$ is `wave[j * hop : j * hop + n_fft]`, for every $j$ such that the frame fits completely (no padding): $T = 1 + \lfloor (\text{len} - \text{n\_fft}) / \text{hop} \rfloor$ frames.
2. **Window.** Multiply each frame by the periodic Hann window $w[i] = 0.5 - 0.5\cos(2\pi i / \text{n\_fft})$, $i = 0, \dots, \text{n\_fft} - 1$ (equal to `np.hanning(n_fft + 1)[:-1]`).
3. **Power.** $P[j, k] = |\text{rfft}(\text{frame}_j)[k]|^2$ for $k = 0, \dots, \text{n\_fft}/2$; bin $k$ has frequency $f_k = k \cdot \text{sr} / \text{n\_fft}$.
4. **Mel filterbank.** With $\text{mel}(f) = 2595 \log_{10}(1 + f/700)$, take `n_mels + 2` points equally spaced in mel from $\text{mel}(0)$ to $\text{mel}(\text{sr}/2)$ and convert them back to Hz: $e_0 < e_1 < \dots < e_{\text{n\_mels}+1}$. Filter $m$ ($m = 0, \dots, \text{n\_mels} - 1$) is the triangle $H_m(f) = \max\left(0, \min\left(\frac{f - e_m}{e_{m+1} - e_m}, \frac{e_{m+2} - f}{e_{m+2} - e_{m+1}}\right)\right)$, evaluated at every $f_k$.
5. **Log.** $S[j, m] = \log_{10}\left(\max\left(\sum_k P[j, k]\, H_m(f_k),\ 10^{-10}\right)\right)$.

Return `S` as a float array of shape `(T, n_mels)`. `wave` is a 1-D float array with at least `n_fft` samples; `n_fft` is even. Do not modify `wave`.

**Example.** A 1,000 Hz tone at `sr = 8000` with `n_fft = 256`, `hop = 128`, `n_mels = 20`, 2,048 samples long, gives `S.shape == (15, 20)`, and in every frame the largest value is in mel band 9, counting from 0 (its triangle peaks at about 1,033 Hz, the center nearest 1,000 Hz).

**Restrictions.** Only `import numpy`. No STFT or spectrogram helpers and no window helpers other than `np.hanning`: the names `stft`, `spectrogram`, `hamming`, `kaiser` and `lib` (as in `np.lib.stride_tricks`) are banned. Loops over the `n_mels` filters are fine.

**Constraints.** Up to 30 seconds of audio at 16 kHz with `n_fft = 400`, `hop = 160`, `n_mels = 80`. Time limit 1 second.
