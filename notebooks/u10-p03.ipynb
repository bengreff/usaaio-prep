The grader calls `model.eval()`, then `model(X_test)` inside `torch.no_grad()` on new points `X_test` of shape `(m, d)` drawn from the same distribution, and expects **logits** of shape `(m, num_classes)`. The predicted class of a point is the index of its largest logit. Your model must reach **at least 90% accuracy** on the held-out points of every test.

The tests use 2D points labeled by which of three rings around the origin they fall in, 2D points labeled by the XOR of the signs of their coordinates, and five clusters in 10 dimensions. Neither of the 2D tasks can be solved by a linear model. Do not modify `X` or `y`. Set a seed inside your function so that your result does not depend on luck.
