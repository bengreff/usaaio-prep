# Best Gini split for one feature

A decision tree chooses each split by trying every threshold on a feature and keeping the one whose children have the smallest **weighted Gini impurity**. Write that search for a single feature.

Implement

```python
def best_split(x, y):
```

- `x`: float array of shape `(n,)`, the feature values.
- `y`: integer array of shape `(n,)`, class labels from $0$ to $K - 1$.

**Candidate thresholds.** Sort the *distinct* values of `x`: $v_0 < v_1 < \dots < v_{m-1}$. The candidates are the midpoints $t_j = (v_j + v_{j+1})/2$ for $j = 0, \dots, m - 2$. A threshold $t$ sends example $i$ left if $x_i \le t$ and right otherwise.

**Score.** For a set $S$ of examples with class fractions $p_c$, the Gini impurity is $G(S) = 1 - \sum_c p_c^2$. A threshold with $n_L$ examples left and $n_R$ right scores

$$\frac{n_L}{n}\,G(\text{left}) + \frac{n_R}{n}\,G(\text{right}).$$

Return exactly `return threshold, impurity`: the candidate threshold with the smallest score, and that score, both Python or NumPy floats. If several thresholds have the same smallest score (within $10^{-9}$), return the **smallest** of them. Do not modify `x` or `y`.

**Example.**

```python
x = np.array([3.0, 1.0, 4.0, 2.0, 5.0, 6.0, 7.0, 8.0])
y = np.array([1,   0,   0,   0,   1,   1,   1,   0  ])
best_split(x, y)    # (2.5, 0.3333333333333333)
```

Sorted by `x` the labels are $0, 0, 1, 0, 1, 1, 1, 0$. At $t = 2.5$ the left side $\{0, 0\}$ is pure and the right side has two 0s and four 1s (Gini $4/9$), so the score is $\frac68 \cdot \frac49 = \frac13$. Every other threshold scores more (the next best is $t = 4.5$ with $0.375$).

**Restrictions.** Only `import numpy as np`. Loops are allowed.

**Constraints.** $2 \le n \le 3000$; `x` has at least two distinct values; $1 \le K \le 10$; values of `x` are finite and may repeat. Time limit 1 second per call on CPU.
