# IOAI normalized scores

Turn a whole IOAI scoreboard of raw scores into task points, the way the IOAI 2026 rules (section 2.5) do it (lesson u22-l04).

```python
def ioai_points(raw, baseline, sc_solution):
    # Your code
    return points, totals
```

- `raw`: float array of shape `(n, t)`. `raw[i, j]` is contestant `i`'s raw score on task `j` (higher is better), or `nan` if the contestant has no valid submission for that task.
- `baseline`: float array `(t,)`, the score of each task's baseline solution (Min_Score).
- `sc_solution`: float array `(t,)`, the score of the Scientific Committee's solution for each task.

For each task $j$, let Max_Submission be the largest raw score on that task among all contestants (ignoring `nan`), and

$$\text{Max\_Score}_j = \max\left(0.9 \times \text{sc\_solution}_j,\ \text{Max\_Submission}_j\right).$$

Then

$$\text{points}_{ij} = 100 \cdot \frac{\text{raw}_{ij} - \text{baseline}_j}{\text{Max\_Score}_j - \text{baseline}_j},$$

clipped to the range 0 to 100 (the rules do not say what a score below the baseline gets; this problem uses 0). A missing score (`nan`) gets 0 points. Return `points`, a float array of shape `(n, t)`, and `totals`, a float array of shape `(n,)` with each contestant's sum over the tasks. Do not modify the inputs.

**Example.**

```python
raw = np.array([[0.80, 0.55],
                [0.71, np.nan],
                [0.50, 0.70]])
baseline = np.array([0.55, 0.40])
sc_solution = np.array([0.92, 0.72])
points, totals = ioai_points(raw, baseline, sc_solution)
# points = [[ 89.9281,  50.    ],
#           [ 57.554 ,   0.    ],
#           [  0.    , 100.    ]]
# totals = [139.9281, 57.554, 100.]
```

On task 0, $0.9 \times 0.92 = 0.828$ is above the best contestant's 0.80, so Max_Score is 0.828 and nobody gets 100. On task 1, the best contestant's 0.70 is above $0.9 \times 0.72 = 0.648$, so that contestant gets exactly 100. Contestant 2 scored below the baseline on task 0 and gets 0 there.

**Restrictions.** Only `import numpy`. No loops or comprehensions, and no `np.vectorize`, `np.apply_along_axis` or `np.frompyfunc`.

**Constraints.** $n \le 20000$, $t \le 8$. Every task has at least one valid score, and $\text{Max\_Score}_j > \text{baseline}_j$. Time limit 0.5 seconds.
