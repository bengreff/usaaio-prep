Inputs:

- `theta`: NumPy float array of shape `(d,)`.

Output: the gradient with respect to $\theta$ of

$$L(\theta) = \frac{1}{N}\sum_{n=0}^{N-1} \left(y_n - \theta^\top x^{(n)}\right)^2,$$
