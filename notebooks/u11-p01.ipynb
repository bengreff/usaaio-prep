# A hand-built network for a dart

Build a neural network, with every weight set by hand, that decides whether a point lies inside a **dart**: the
non-convex quadrilateral with vertices, in order around the boundary,

$$A = (0, 0), \quad B = (4, 2), \quad C = (0, 4), \quad D = (1, 2).$$

The vertex $D$ points inward, so the dart is not convex. This extends 2026 Round 1 Problem 7.4, where the shape was a
triangle.

```python

    def __init__(self):
        super().__init__()
        # Your code

    def forward(self, x):
        # Your code
        return out
```

- The constructor takes no arguments.
- `x` is a `torch.float32` tensor of shape `(B, 2)`, one point $(x, y)$ per row.
- `out` must be a floating-point tensor of shape `(B,)` with value `1.0` for points inside the dart and `0.0` for points
  outside, in input order.

**Rules (as in the 2026 problem).**

1. Build the network only from `nn.Linear` layers and a threshold module that you write, $\Theta(z) = 1$ if $z \ge 0$
   and 0 otherwise. `forward` may only apply these modules (and reshape the result); no other arithmetic or
   comparisons on the points outside your threshold module.
2. Set every weight and bias by hand. **No training.**
3. Every parameter must have `requires_grad` equal to `False`.

**Example.**

```python
model = My_Dart()
pts = torch.tensor([[2.0, 1.5], [1.5, 3.0], [0.5, 2.0], [3.0, 3.5], [-1.0, 1.0]])
model(pts)      # tensor([1., 1., 0., 0., 0.])
```

$(2, 1.5)$ is in the lower half of the dart and $(1.5, 3)$ in the upper half. $(0.5, 2)$ is in the notch to the
left of $D$, inside the triangle $ABC$ but outside the dart.

**Restrictions.** Only `import torch` and `import torch.nn as nn`. No `torch.optim`, no `backward`, no files.
The grader also bans the names `where`, `sign` and `heaviside`.

**Constraints.** $1 \le B \le 10{,}000$; points lie in $[-1, 5] \times [-1, 5]$. No test point is within $0.001$ of the
dart's boundary, so the value of $\Theta(0)$ does not matter. Time limit 1 s per construction and per forward call, CPU.

````python
#@title Grader: run this cell to grade your solution { display-mode: "form" }
# The grading code is hidden on purpose (double-click to see it). It writes the site's test harness and this
# problem's grader to a temporary folder, imports your file the way the contest grader does, checks the import
# and ban rules, runs every test (including hidden ones) and prints a table.
import ast, base64, contextlib, importlib.util, io, json, os, signal, sys, time, traceback, zlib
_P = json.loads(zlib.decompress(base64.b64decode("eNq9WwuT27a1/iuoMjcmHS0taeO0I2d9u7Y3qTt+7NjutB2tLkOJkMQuRTIE6ZWi7n+/3zkASFDS2l6n9yYzXgoEzgvnDXDXS+LeWPTq4fCkGAx7fdFbyyrC0O7wTZVUqaSxc7GKsvhkVidpJTJZ3eTltVjkpYhEHJUVzU2lUnlmAaSDEYNOsrqSCqOjAX7GyWKRzOu02mLkewzIrCrpufd6G74wgNZ5XKcyzKI1oy7KfJbK9TBU9WydKJUACWZFaZrfyDhM1kVeVoRh0qvycr7qTfF2FmWZjHkwL6pk3eOx+fVNVMZMbB7x35uVLCU9qGTJYFcy+pioJJYWTJjmeUHgF1GqJMlEqioESJnXVUgvTomzKlpqGrIsIFGxpAgg/04XJ0UaZYyJiQw0k4xE5XU5Z1ajOCoqGZ+MBqMfTt4NTy7/SAsKyGWE11VZy1v8XiQpS3TXW0VlBqkHBYvwiv//i0wLWSpRraJKGNmJZRnFNKiFJbxfFmW+Fma5HX34ix9cZVfZuZkuEiXmeVZhjx7VWVKpRz/Sn6ePDFT8LpL46SM9G0QE4mUlYrlIAFWQmMSizuYVNkwJ2s2YB8OHmrYqupZ4kldZu7FCi0X8gqFfAnHxj/PXl68u3os0URWxJDVYAqaEWuU3mcCid3UmPNo1UOU/Ee8JXCXKGmjlR1lueRFY+ylK0rrEyjJKlBT0U9wk1QpKDHgqWkpxUyYVGGbVfiazsVDRVtwQuTeRwts8W/ah/4Aq4npdiFUSx5ieZEVdKZKe3oXMiHSeF9vmB+kMydf8zLB+KwA1K2j0G/E8T6OZeH75NyWiUgpF+l2SrDIi5YESr0fissz7QuUizedRShCTbKnEuoZQZpKEChHlZGMJXqcJyWEG/lfYykU0h+oFhGkUDGhzI7Gsaf89jVkBVipPqlUpo1ioQmLDMKvM6+Uq3QrSYgFaQMU8L0nQpSShVyJZEIH4yZZo+DAWQkJ5f3lx8SL86fz5h7fvxBlhJ4avsnkaATttg3exmcuCNMUfX2UC/xV4pWdBoSBxGXvQxbgv1mpp5wBvlpOQs9iM0H/t5no0twUSZoW3sWvhdpw1dofINJ3RBfhPMlVF2Vx6m75+H3yQmcpL31nOaGVVl5nYBDG86Xzl+cG8qPEvb7Pn68mS2RQvGd1FWealA0Wz7MDKiiBSUVlGW9Dd8qFWUSG9ZV71oZQZ/Qv9PLvqQbPhWa96ljIFWVd1kUqPOMd8P+ClhhaWKebYSQTL74vFVW9HAG+x4UrjEjsFvyM3hZzDO4mdM/8W6FrK5mmuPkVZX5RVnp4N5ckPfRGZxz9ZemFZNyC5ITZS1RaMLuCriTAaZ5zdFw4zS80fsXSjH+9gRzNgpvvHeLOvmL0GATYkUXBvSYXVoCNNPV/AU5AWuu9uzDsXPXnSKGFPmNGaJFs0wI0i036nqREihGHERf8YcdE/ruLJsoTEWBBE3Dra0J9oprylOBE3vu8TbDCa/CaFRPwSAzK+I6bSlRM7OvExSuEfhJdG5ZLcriSFFTv8GQeny1u/s/fy1zpKv0Aru0alpxLRWcya3vdZoiAiqqrSvL/q8Xa0QI6oi6Mg7Zz/c7VwVYPpD40cQNyBMblSbSDQroz3KYYyELXMfAsDzniHV38oO6TRLAx1NiNRIZsIbwc9qTv3gxE6O9IuOaCfRiEKA9gPQs7RwrBDzhLuj3bOYH1gJz2wULuUhiqLCkTy6v/XNad5Ju/plIHSauWGVJJje89HSIzZerv0eHEyhyApcfFderrMdagjgF4sZXH2AZmeo2CGwA8Q3z55hwD8TgihoYBg8rtNR/DYFi/qi9nXyz36IrnrKdouGN/Xij0isTMc1XUFBlGkkShv5h/zNVje+pmjyxs/MTOPvLddyyYO+trbQbMzd68sHIbgJC95uK5hFMhuvEXWFw/hTmEZDx9SNbJUlhKdOj5HBBCLjDEvKEElHpDjIAtcwryiDCljuazXSMoRQgSqirkMbNbJUZ/sCX5x0lpW5JtSDQsI41TPRHZGwSPz9gnSr2lJAkUGw8qnpRKpjCwjxLffksLTaxhHR8HZnzTKZZzIFiUOdDGWlOAni4QyS0qZW16+2SXiOzG8fSK4tMxJjKSzgrYP+WjjL42QQXwrYUrik8xTklJBUPWlUn7iihje1U25j2TSXHIiHirk3uCABMS57oNmJZJ3XUMRqrcAUVIJhgJprYglzsNfR3MxBwJaAiZ11k3VCif2lBygukClHcOT0jDn4EjLEyqg6gLmJKqbZC4JICSUIELnC2PBoNlKVLEK0cgC+RKFb4KAFTXeBeIDvUhKDIOZ9AGhoLigkZGEZdxRq2pA2SQgBKguFyGDkaX1Np9Rpbg6vhg5SmXSEcPvmTCbKB4Jt2xoVTIkVSzJGryRq3fYRKD58cxA2nNFM9Q01+1QNOqL6xGwdf0jk9zfG+xw8nlRMJ0kiBGLYeQMsxhQsXkxhdg7xOHmmy5DbEdVnrPeoPiPq3EwWtwiy6cWxhNTdTVKzMoEs2NAos6ont8Z0Zh1XEOisLNbfOQ/RonKEmXwAqq1pARWF5k7s08MiKzjkwZaZktYp4zthrUFDnYyzteQ9iKC+oXNxCvuxtQqipI85A5Dp8vxM43AAkBZ2zxI2GmskjQmq5tTZwQFc0K2QDP/+v7tG9CaseWoKgaFbKvFtlph5OSlsHjEj6bFEcZJ+VT82OLAS/yGVf6bx6qnByU/jM0p/7GxmypNZs2Q/oORoK6StB3Om8d/qbxtEhRRtXKXU5sqapeprdprL9gfJeICNbyaEWhxRs0C3WtQ8wRsZrJxdxi4TqqTVGKaGAY/BEOT06kx9YnW1N3rs0tBdWGCj3V+RZRAysurzOIIFkkKjbY/vaseyM5LSRWg6bQgEw0e/i27zqiHo/KUOio5dwCAMClo14KHJlP8y8W7C1iOkUVwib9eGFIbLAz9AI6LliOrK6ISlF1lkEpAkwNoqywrbwBjQPpAYHwAPNb74l6Q+Aax+tdoLC6+H4y4E/bq7+f/fB8+O3/z5uIFKNhd9ULTcAxDYgaM2f3UP+VGzs0T8nz9NM/XBWi96t1eZa/evr0M37x9cfEe4DzoSvBTDlulh79DcaV+PFfbbN68eIUc8jlg6F/vpfPjBZLM9tfPMqP4nJcXm8J3WywfdC/mc12WMEKlt/ZIx+o1/A3ClmwslktFC8jtRlBmMw/nKzm/Vp4q57TFVdSNt++0vVM2jGBFcsyvtXDWagnBPIFbpAxQ9/jYbFT1oNkdToZMK5ECVVUmuq3YCVE6CE2mx7LZqpRI7FhK0BLU1yC0m4e+36I+33AeSk05eZgfTgzZY2gLt4IN9WNyk4qXmxI5z7Sf2cmA/mb57ZieMRvFz60h0DSwKSRAXgEcM+xkr6tNMppMDaGzqBnnOFnZkMPt0LBWDMwZn8EeQt0rNeOk4kgZYbtJDFOhTKzUiYIGDAh6vkcVDSXdyYz692DnmnM3mQak2AYBt8Ydma+jal2n7siCuyNw1eSbIfybKL32aDP2QreTptN0rc+6NNivKppUliYGTO74MICZlgryZS/iSZSUr6m1oH8GwAZBUuborZF5wh1RMUfA10ys3gh/fDw2OnsRRHFscPiHk9u9ceYFqkgT2m7COTkZTo8sbDcvqIuYsu4ja90iMf20GH+C19vnBhuKvWI5msY72G8N6ogseYURJJ5/hxT3JQhoe1JwRNeRwDEN2Ft6l/C+ainzeV+hn1vb2ed9TyGYDjK0L4L6xvHInwKYxHdzdXzSMYxtsNrHyYYfREUhs9izHR+96KqnXZ527//dldcxLBwFnyXZWxPFzuvluaIY5Oveyt6CIDfzXkfVa6SMBzqtfdA9iQNhjkI6IClJNKDudP82C9BWolWeRtkgaF7wr5zKU3awnoPH9497l+8ookAwFtTuQV880ECsRZEBPgA+1L3LB9SKvT3eTzxGP7UsjpCvLO0tsG+aE8RZlCmS7XxlIgSngm2A4PazjFDE/oP/+k90NZreRFt1oo9jBR+KJg0gUkoDLc/Sbd/iROXShhpPa4FuIplUipKOrUjlEqX5DaVNgmcGQeA7wQxJmYhz2hQNeJVwfal3wXOC5rccHJ0orMk1wRdq+G/hObH0W9HJCx0VAoZ7qw4oUMLIh+B7yj+iO4Ds/54dznKLozmSbXcZlHeZD7URhXmh88lut42mayPbM70vYZgqSujFn4WFTSWrYzfC05mTwTAZTF3lPsL+fUTQYv0U8+zcDljm0a9jWKdJrOwdRjFMZkxKXcqVzLiIbXJHXvWf5d9wdvuJSn1RJgCTbkPOZD2AQKEbUjXVzejfZvKE6YylmpcJVxWigKrCvpfCnETzhMQeX1PBBnOVNE/xYX23wzSjtoqtWwPUzXiuwmrmySAMmxco+UyyScCRaS44qC8ID0BgpxZUfkqb81Hl17BgslK+9UENK/Ky3Js0Sod/KR1zsvbOEG0FKwzN5uO0lnwjTDox4RMS6ZyPUPZ/u2OstzTfCnuNTNxrajF42v0qlyrZqFx+nDQpomXlzpmjqVMSG0lRAn4mmimnNl2H3mOcGg4BXYhRnsdEPCK1IZvgVkSP4CFmUSfDs84O5RNBNMS4E6w0uJt4Jg7tQheMNDSZ0m8q9VT7UzdmtMqaSpo0kQbe5NgCF/6khTYlcu4oRRtDpwyWTFBNNFVTjhb0i61k6qaztqSz+azivuMBWtdLcNfCY3HS5RDl6elulNdaYlueCxCd5MH7itonL99a4ekmT6D/eObX+5c/n79697pvqnQzFe4s5OauPQV2fFn3ohLJcjSwtHSrY74E0zarsJ1xUkoYn94MD5Tup1mWRlklhKK0ZL788PL1xbvw3cX5q35D3V6aowpJ2tNthQU0GlJvRnd2qDvJRyYOR87NMK0abWMO2+S4qj189ewQnYHFCAm1R//sL4TF6HlqQq+5YJhqxT9kiW1IlgH1gEK9jDyP/xWCG3R7E6bt0mlJGEU0xjHV3kzzSB6YvRp73CrPr/lMQezsfoyDAXWAresyWJ5FSjYdIt0DaVtiz15dDAbDloB7sXEXxeZgiMmkkxi6IZabjTJJ0CfiUWPWB6CPCcr6FcILfSaN4pN4DwX494PBYDz9akPmS4zdPg4Pufc79Bx7VmxPNMFLX7/yP7+5raziXOqkSV+2Q3QhELdu3X5PFpb3s8mrnr70R1bYhIzmHmDDt7lH+HnbWzrGt7zLljQ0M8tc9zONLVvl6SnkGezVQZvEG7dHsYaSBg+FykI7dnoi3/4xKpXFESSVXCuPb89k3ajAXtWe+9MBSgQyvIXvNrzgJegQklMSwriXPHM4Ju0va/gtrRSUrfAtgsyytudwyT0nWS3d/ZvrAKv94ZiBTB6Pp/DffC4Mak2XWrB7vOoZ0Hayg43eUjeYyp5jZYrxt/rQh4YGweD2PmdihxcfviLq/N7IYw7nDr3yV7k0swmTVnJkqZR4H97dOPDg7WKde2gbZ2ed5tnSnsvs+2zhJZm+ZcbZfJ8KiEVUCnsy6LtOwCDnw429nvbvZLihmdLrI3dV7hVL/lO0uPcM+Lwi/sIY0sJqFJzgldDi2LvroLYvTo+FN5POTm1BVjaO7d6hqFmgc9UzKh29Nnf9VFbKTmWW56l3QJh+R6CqjuYSuKoDrllhC5Q7Qwpfa14IW/Bo7xaGVN/QQdnYHg9QuaNPdptgQWe6x25Ed+8f3XVm1+fz7b65AdJ379r06dSZCGtukcPvG/8tN9G6ID84pQmXb1/9k/rg5qKPN5nAufXJw6EimXxPzyP9zOPf6+ehHYdESbfP++JZXzzvixdtgVeVNYLGZbd0vvgos5M8jqGjW0QQxfkazDqWpDB0tppzqSLa7wbITcLSau7c+J2yedMXlHtcTsagGHTxw9BoUJLRpwyat99kmSsPhTHo6YuYitQz0hD3uk97t4LnQS6duz3eZghsQyz3NiM8jXzCjFmThDDTg8fXeHzxX6KB4CRWc5Cg+DjK2w7FU7H1xR/oecTPewUJaIbBkk5LL04+gpEz9yA5yWA0SeyM7UeLDfQVqDZDkORtYbQgXTwk2vG8wfMjRs3jToTWMvufs4bab7FE/Mjgup0TPbXd7hk5jKjchnFiOtbN1sd6ExY17M7uAV0Zzhb3kD/dPbunxCuNd54mhed5l2A28sWfhTfjJxaBeXZGoUnQIffmioayRuBZ12svZtrTBA56GUD2awbsRSCkIvWjQn1Kkm7gRZtEnQ39rvhip+0UlTodtCJsJWcmX06OSRd6A5scDKduU+Va113W64B2c0GMTRavAvO1kX/8CuzaXmvMsuA1556cPJlF+msLQNFn682c/Qswa+f6S50R0Mu+QLYomxvcG8qZGFHF9ye9S2uXepQr+lN7g4htwtClr8R4frcNvMdoxxcC/eYOdrFy7x6nvvjLtN6OSTPpsynNd3PHUZPcvR+vb0lTN1Lfk6Y4YLV9Dygm0f0/hrmKPkpzB9t71ief4u30qltaRrehd/ug927m04tEhSwxuNOQO5HeJ3DOJHjg6YYTcrpQJQ6OEJ/gS9g7gstbYm67s6RBII/b28SI5lQD7H3r0X4xgA3JbfAA6Gnn64A94syde0sjAtW8SrdMGWjE6oN7Vs7nF/marsAQsn1V4wt4Jhgd+37CHPF6OX8PYgh0tIu6LCGcE0o5yNXz+H46nM63ZsnA9wMFz+DWl7xGq6WzaNAuGh5ZBCrm1Nz0TEATT+0SO/CdGMG1mDBH9+P+xK/dwadNPORXBm+LZKUDA3V/FX3LFaXcnU/0rU9NQioXHIhfmJ6vZiFv+OZZln5yQ/sN4SMfWNi91h8CRJm6kaXiJjNL99bGHtZeJZZkF1ojd1aYt6QoB5OGO2Kpe7PeTXT2/WHjI/XQZTf7GWlFfayznMdIdG3287jJhE5pzqmeczK0mj11TKR1enQtyuRbrm1oYU6QUgw54Ax4wxzbMKvG2gPkQZXTBSKv82XGxOLmfK39Z3ooi2WZxJ8RxNIKgmOw5usx4stgyEAf+9A+/B2etnI7jFytJPV19/bSwZIfZ/zY9CSs1bryMp8VEsVcuR8Z3GNO36MMZ3SwKtVn2KSWDN20HA6d5OMa+Wv3mjdtDdK8P0IGAxIy5DDoNLyP818GNYoBygpOAAAqQ18fnXnPILvHlDL7vj8ZP3OzwgMJLBpumSGxu+bL4QgP1PF4xs74iyfuSSqToBh2HpZykcpNCPuv5OZLJTbyv2zrm+qAVMaRyCA4JVU6tVI5HQyMTO7WBmPpRDiKi0YfOsPkMmm7jjBcUv/6M/wBJ91RP+Mret464N8Sxa5qvDPbLJ2guJlSHju50quEaDHXf3K64i/Wtn/u+ZxDsX81n4/bGKc/KucKrwGC6LUF8q7DoJxAE8oO18LjuwL8gVKWIyouwYEs4RpaHuDmQaQ0QKnCdG4u8Ld4hPm9/LWmS3F0va3X3rmjsjGzncJCM8VfM4eukChEFEEJEAnKYk7Ppm4eAA8GWKywnUmNFLjlxmj0N8sN8CcCFWKacl+J3neugwCmkwrN5CLng83d9Vh8tF85Wev+qGnngiqkz5O8ps9pT9W00t0RChr3P+WtZLUSfNjWEBAtKu73drG4+kNtB+d7IE3x5BrweS2eDL1ErH7r695ks5n2W5i4pmMzm54yDb3b2/8FKWukkQ==")).decode())
_dir = os.path.join("/tmp", "usaaio_grader_" + _P["id"])
os.makedirs(_dir, exist_ok=True)
for _n, _s in _P["files"].items():
    if _n.endswith("::b64"):
        with open(os.path.join(_dir, _n[:-5]), "wb") as _f:
            _f.write(base64.b64decode(_s))
        continue
    with open(os.path.join(_dir, _n), "w", encoding="utf-8") as _f:
        _f.write(_s)
if _dir not in sys.path:
    sys.path.insert(0, _dir)
for _m in ("harness", "usaaio_grade", "grader", _P["meta"].get("module_name", "submission")):
    sys.modules.pop(_m, None)
import harness as _harness
_harness.SPEED_FACTOR = 1.0   # Colab is the reference machine: time limits apply as stated
import usaaio_grade as _G

def _usaaio_grade():
    meta = _P["meta"]
    mod = meta.get("module_name", "submission")
    path = os.path.abspath(mod + ".py")
    rows, static = [], []
    if not os.path.exists(path):
        print(f"{mod}.py not found: run the %%writefile cell above first.")
        return
    src = open(path, encoding="utf-8").read()
    static = _G.static_checks(src, meta)
    per = float(meta.get("test_timeout_s", 20))
    def alarm(*_):
        raise TimeoutError()
    use_alarm = hasattr(signal, "SIGALRM")
    if use_alarm:
        signal.signal(signal.SIGALRM, alarm)
    def limit(s):
        if use_alarm:
            signal.setitimer(signal.ITIMER_REAL, s)
    buf = io.StringIO()
    error = None
    if any(not s["ok"] and s["msg"].startswith("syntax") for s in static):
        error = "syntax error"
    else:
        try:
            with contextlib.redirect_stdout(buf):
                limit(per)
                spec = importlib.util.spec_from_file_location(mod, path)
                sub = importlib.util.module_from_spec(spec)
                sys.modules[spec.name] = sub
                spec.loader.exec_module(sub)
                limit(0)
        except TimeoutError:
            error = f"importing your file took over {per:.0f} s"
        except BaseException as e:
            limit(0)
            error = "your file failed to import: " + _G.friendly_error(e, path)
        entry = meta.get("entry")
        if not error and entry and not hasattr(sub, entry):
            error = f"your file does not define {entry}"
    if not error:
        gspec = importlib.util.spec_from_file_location("grader", os.path.join(_dir, "grader.py"))
        grader = importlib.util.module_from_spec(gspec)
        sys.modules["grader"] = grader
        gspec.loader.exec_module(grader)
        visible = set(getattr(grader, "EXAMPLES", []))
        for name, fn in [(n, f) for n, f in vars(grader).items() if n.startswith("test_") and callable(f)]:
            rec = {"name": name[5:].replace("_", " ") + (" (example)" if name in visible else ""), "passed": False, "msg": ""}
            t0 = time.perf_counter()
            try:
                with contextlib.redirect_stdout(buf):
                    limit(per)
                    fn(sub)
                    limit(0)
                rec["passed"] = True
            except TimeoutError:
                rec["msg"] = f"took longer than {per:.0f} s (infinite loop, or far too slow)"
            except _harness.Fail as e:
                limit(0)
                rec["msg"] = str(e)
            except BaseException as e:
                limit(0)
                rec["msg"] = "your code raised " + _G.friendly_error(e, path)
            rec["seconds"] = time.perf_counter() - t0
            rows.append(rec)
    print("Rule checks")
    for s in static:
        print(f"  {'PASS' if s['ok'] else 'FAIL'}  {s['msg']}")
    if error:
        print("\n" + error)
    if rows:
        w = max(len(r["name"]) for r in rows)
        print("\nTests")
        for r in rows:
            print(f"  {'PASS' if r['passed'] else 'FAIL'}  {r['name']:<{w}}  {r['seconds']:6.2f} s  {r['msg']}")
    out = buf.getvalue()
    if out.strip():
        print("\nPrinted output (last 2000 characters)\n" + out[-2000:])
    n_ok = sum(r["passed"] for r in rows)
    ok = not error and rows and n_ok == len(rows) and all(s["ok"] for s in static)
    print("\n" + (f"Accepted: all {len(rows)} tests passed and every rule is met." if ok else
                  f"Not accepted: {n_ok} of {len(rows)} tests passed" + ("" if all(s["ok"] for s in static) else ", and a rule check failed") + "."))

_usaaio_grade()
````
