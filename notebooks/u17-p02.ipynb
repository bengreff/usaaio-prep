# Cosine top-k retrieval

Find the $k$ documents most similar to each query by [[cosine similarity]] (lesson u17-l03).

```python
def top_k(Q, E, k):
    # Your code
    return idx, scores
```

- `Q`: float array `(M, d)`, one query embedding per row. `E`: float array `(N, d)`, one document embedding per row. $1 \le k \le N$.
- The cosine similarity of a query and a document is their dot product divided by the product of their L2 norms. If either vector is all zeros, define the similarity as 0.
- Return `idx`, an integer array `(M, k)` of document indices, and `scores`, a float array `(M, k)` of the matching similarities. Each row is sorted from the highest score down; equal scores are ordered by **smaller document index first**. Parallel vectors can get cosines that differ in the last bits, so treat two scores as equal when they agree after rounding to 12 decimal places (`np.round(S, 12)`).
- Do not modify `Q` or `E`.

**Example.**

```python
Q = np.array([[1.0, 0.0], [0.0, 2.0]])
E = np.array([[2.0, 0.0], [1.0, 1.0], [0.0, 0.0], [0.0, 5.0], [3.0, 3.0]])
idx, scores = top_k(Q, E, 3)
# idx    = [[0, 1, 4], [3, 1, 4]]
# scores = [[1.0, 0.70710678, 0.70710678], [1.0, 0.70710678, 0.70710678]]
```

Documents 1 and 4 point the same way, so they tie, and the smaller index comes first.

**Restrictions.** Only `import numpy`. No `np.linalg`, no loops or comprehensions.

**Constraints.** $M \le 500$, $N \le 20000$, $d \le 256$. Time limit 2 seconds.
