# Least squares by QR (modified Gram-Schmidt), no np.linalg

Solve a least-squares problem the way numerical libraries do: factor $X = QR$ with Gram-Schmidt, then solve a triangular system. The normal equations square the condition number of $X$; QR does not.

Implement

```python
def qr_least_squares(X, y):
```

**Inputs.** `X`: float NumPy array of shape `(n, d)` with $n \ge d$ and independent columns. `y`: float array of shape `(n,)`.

**Outputs.** Return exactly `return Q, R, w`:

1. `Q`, shape `(n, d)`, with orthonormal columns ($Q^\top Q = I$).
2. `R`, shape `(d, d)`, upper triangular with a **positive** diagonal, such that $QR = X$. (With these rules $Q$ and $R$ are unique.)
3. `w`, shape `(d,)`, the least-squares solution $\arg\min_w \lVert Xw - y \rVert^2$. Since $\lVert Xw - y \rVert^2$ is minimized when $Rw = Q^\top y$, get $w$ from that triangular system by **back substitution** (last entry first).

Do not modify the inputs.

**Use modified Gram-Schmidt.** For $j = 0, 1, \dots, d - 1$: set $r_{jj} = \lVert v_j \rVert$ and $q_j = v_j / r_{jj}$, then immediately remove the $q_j$ component from every later working column, $r_{jk} = q_j \cdot v_k$ and $v_k \leftarrow v_k - r_{jk} q_j$ for $k > j$. (Classical Gram-Schmidt computes $r_{jk}$ from the *original* column $x_k$; in exact arithmetic it is the same, but in floating point it loses orthogonality on badly conditioned matrices, and the hidden tests include one.)

**Example.** For

```python
X = np.array([[1.0, 1], [1, 2], [1, 3], [1, 4]])
y = np.array([1.0, 3, 2, 4])
```

`R = [[2, 5], [0, 2.236068]]`, the first column of `Q` is `[0.5, 0.5, 0.5, 0.5]`, the second is `[-0.67082, -0.223607, 0.223607, 0.67082]`, and `w = [0.5, 0.8]` (intercept 0.5, slope 0.8).

**Restrictions.** Only `import numpy as np`. Nothing from `np.linalg` (no `qr`, `solve`, `lstsq`, `inv`, `pinv`, `norm`), no `np.polyfit`. Loops are allowed.

**Constraints.** $1 \le d \le 60$, $d \le n \le 5000$. Condition number of `X` up to about $10^6$. Requirements: $\max|Q^\top Q - I| \le 10^{-8}$, $\max|QR - X| \le 10^{-9}\max|X|$, and `w` within a relative error of $10^{-6}$ of the exact solution. Time limit 1 second on CPU.
