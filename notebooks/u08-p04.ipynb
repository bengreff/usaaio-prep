Implement principal component analysis with NumPy, matching scikit-learn's `PCA` exactly, signs included.

```python
def pca(X, k):
```

- `X`: float array of shape `(n, d)`, one example per row. `k`: a Python `int`, $1 \le k \le \min(n - 1, d)$.

Compute, in this order:

1. The column means $\bar x$ and the centered data $X_c = X - \bar x$.
2. The sample covariance $S = \frac{1}{n - 1} X_c^\top X_c$ and its eigenvalues $\lambda_1 \ge \lambda_2 \ge \dots \ge \lambda_d$ with unit eigenvectors $v_1, \dots, v_d$. (You may instead use the SVD of $X_c$: $\lambda_j = \sigma_j^2/(n-1)$.)
3. **Sign rule.** An eigenvector is only defined up to sign. Flip each $v_j$ if needed so that its entry with the largest absolute value is positive (if several entries tie in absolute value, use the first of them).

Return exactly `return components, explained_variance, explained_variance_ratio, Z`:

| name | shape | value |
|---|---|---|
| `components` | `(k, d)` | row $j$ is $v_{j+1}$, after the sign rule |
| `explained_variance` | `(k,)` | $\lambda_1, \dots, \lambda_k$ |
| `explained_variance_ratio` | `(k,)` | $\lambda_j / (\lambda_1 + \dots + \lambda_d)$, divided by the total over **all** $d$ eigenvalues |
| `Z` | `(n, k)` | the scores $X_c\, V_k$, where $V_k$ has $v_1, \dots, v_k$ as columns |

These are scikit-learn's `components_`, `explained_variance_`, `explained_variance_ratio_` and `transform(X)`. Do not modify `X`.

**Example.**

```python
X = np.array([[4., 6], [5, 5], [2, 2], [1, 3]])
components, ev, ratio, Z = pca(X, 1)
# components = [[0.70710678, 0.70710678]]
# ev = [6.0],  ratio = [0.9]
# Z = [[2.12132034], [2.12132034], [-2.12132034], [-2.12132034]]
```

The mean is $(3, 4)$ and $S = \frac13\begin{pmatrix}10 & 8\\ 8 & 10\end{pmatrix}$ with eigenvalues 6 and $\frac23$, so the ratio is $6 / \frac{20}{3} = 0.9$. The top eigenvector is $\pm(1, 1)/\sqrt2$; its two entries tie in absolute value, so the first must be positive.

**Constraints.** $2 \le n \le 5000$, $1 \le d \le 100$; the hidden tests have distinct eigenvalues. Time limit 1 second per call on CPU.
