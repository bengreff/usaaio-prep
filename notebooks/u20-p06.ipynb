**`q_update`** applies one Q-learning update to the float array `Q` of shape `(n_states, n_actions)`, in place:

$$Q(s, a) \leftarrow Q(s, a) + \alpha\big(\text{target} - Q(s, a)\big), \qquad \text{target} = \begin{cases} r & \text{if terminated} \\ r + \gamma \max_{a'} Q(s_2, a') & \text{otherwise.}\end{cases}$$

No other entry of `Q` may change. The grader tests it with tables whose terminal rows are nonzero.

**`train_q_learning`** starts from `Q = np.zeros((env.n_states, env.n_actions))` and runs `n_episodes` episodes. Each episode starts with `s = env.reset()` and repeats until `terminated` is `True`: choose `a = epsilon_greedy(Q[s], eps, rng)`, call `s2, r, terminated = env.step(a)`, apply `q_update`, and move to `s2`. Use only `rng` for randomness. Return `Q`.

The grader's environment (states are `12 * row + col`; actions 0 up, 1 right, 2 down, 3 left):

```python
class CliffWalk:
    ROWS, COLS, START, GOAL = 4, 12, 36, 47
    n_states, n_actions = 48, 4
    MOVES = [(-1, 0), (0, 1), (1, 0), (0, -1)]

    def reset(self):
        self.s = self.START
        return self.s

    def step(self, a):
        r, c = divmod(self.s, self.COLS)
        dr, dc = self.MOVES[a]
        r, c = min(max(r + dr, 0), self.ROWS - 1), min(max(c + dc, 0), self.COLS - 1)
        if r == 3 and 1 <= c <= 10:                     # the cliff: -100 and back to the start, not terminal
            self.s = self.START
            return self.s, -100.0, False
        self.s = r * self.COLS + c
        return self.s, -1.0, self.s == self.GOAL
```

After training, the grader follows `np.argmax(Q[s])` from the start. The route must reach the goal in **13 steps** (up, 11 times right, down), the optimal route along the cliff edge. It trains with 300 episodes, `alpha = 0.5`, `gamma = 1.0`, `eps = 0.1` and `rng = np.random.default_rng(0)`, then with two hidden settings. A correct agent takes about 7000 environment steps in 300 episodes; the grader stops a run after 100000.

**Example.** With `Q = np.arange(12, dtype=float).reshape(3, 4) / 4`, the call `q_update(Q, 0, 1, -1.0, 2, False, 0.5, 0.9)` sets `Q[0, 1]` from 0.25 to 0.8625 (the target is $-1 + 0.9 \cdot 2.75 = 1.475$). With `terminated = True` instead, it sets `Q[0, 1]` to $-0.375$.

**Restrictions.** Only `numpy`.

**Constraints.** Time limit 5 seconds per training run of 300 episodes. A correct solution needs well under a second.
