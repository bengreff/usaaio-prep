
- `q_net`, `target_net`: modules mapping states `(B, d)` to action values `(B, n_actions)`.
- `s`, `s2`: float tensors `(B, d)`; `a`: long tensor `(B,)`; `r`: float `(B,)`; `done`: float `(B,)` with 1.0 where `s2` is terminal; `gamma`: a float.

Return the mean squared error, a 0-dimensional tensor:

$$y_i = r_i + \gamma\,(1 - \text{done}_i)\max_{a'} Q_\text{target}(s2_i, a'), \qquad \mathcal{L} = \frac1B\sum_i \big(Q(s_i, a_i) - y_i\big)^2.$$

The target $y$ is a constant: `loss.backward()` must give gradients to `q_net` only, never to `target_net`. Do not modify the inputs.

**Example.** If `q_net(s)` is `[[1.0, 2.0], [3.0, 4.0]]`, `a = [1, 0]`, `target_net(s2)` is `[[0.5, 1.5], [2.0, 1.0]]`, `r = [1.0, -1.0]`, `done = [0.0, 1.0]` and `gamma = 0.9`, then $y = (2.35, -1.0)$ and the loss is $\frac12\big((2 - 2.35)^2 + (3 + 1)^2\big) = 8.06125$.

**Restrictions.** Only `torch`. No loops or comprehensions.

**Constraints.** $B \le 4096$, `n_actions` $\le 18$. Time limit 0.5 seconds with small networks.
