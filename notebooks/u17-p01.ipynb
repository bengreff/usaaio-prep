- `counts`: integer array of shape `(n, V)`; `counts[d, t]` is how often word `t` occurs in document `d`.
- Term frequency is the raw count. With $\text{df}(t)$ the number of documents with `counts[d, t] > 0`, use $\text{idf}(t) = \ln\frac{1 + n}{1 + \text{df}(t)} + 1$.
- `W[d, t]` $= \text{counts}[d, t] \cdot \text{idf}(t)$, then each row is divided by its L2 norm. A row of all zeros stays all zeros.
- Return a float array of shape `(n, V)`. Do not modify `counts`.

**Example.**

```python
counts = np.array([[2, 1, 0],
                   [1, 0, 1]])
tfidf(counts)
# idf = [1.0, 1.405465, 1.405465]
# array([[0.8181802 , 0.57496187, 0.        ],
#        [0.57973867, 0.        , 0.81480247]])
```

**Restrictions.** Only `import numpy`. No `np.linalg` (compute the norm yourself), no loops or comprehensions.

**Constraints.** $n \le 2000$, $V \le 5000$. Time limit 1 second.
