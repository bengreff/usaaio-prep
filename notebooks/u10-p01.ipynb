**Parameters.**

- `self.weight` must be an `nn.Parameter` of shape `(out_features, in_features)` and dtype `torch.float32`. Each entry is drawn independently from a normal distribution with mean 0 and standard deviation $\sigma_w = 1/\sqrt{\texttt{in\_features}}$ (the choice that keeps the output variance equal to the input variance).
- If `bias` is `True` (the default), `self.bias` must be an `nn.Parameter` of shape `(out_features,)` whose initial values are all 0.
- If `bias` is `False`, set `self.bias = None`; the module must then have no bias parameter at all.
- The module must have no other parameters.

**forward.** `x` is a float32 tensor of shape `(..., in_features)` with any number of leading axes, including none (shape `(in_features,)`). Return $x W^\top + b$ (or $x W^\top$ without bias), shape `(..., out_features)`. Gradients must flow to `weight` and `bias`.

**Example.**

