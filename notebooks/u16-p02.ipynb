- `images`: float tensor of shape `(B, C, H, W)`; `P` divides both `H` and `W`; $N = (H/P)(W/P)$.
- `patchify` returns the non-overlapping $P \times P$ patches in raster order: patch number $r \cdot (W/P) + s$ is the one whose top-left pixel is at row $rP$, column $sP$. Each patch is flattened in the order (channel, row within the patch, column within the patch), so entry `c * P * P + i * P + j` of patch $(r, s)$ is `images[b, c, r*P + i, s*P + j]`.
- `unpatchify` is the exact inverse: `unpatchify(patchify(x, P), P, C, H, W)` equals `x`.
- `patch_embed` applies the patch embedding with `weight` of shape `(d, C, P, P)` and `bias` of shape `(d,)`. The result must equal `nn.Conv2d(C, d, kernel_size=P, stride=P)` with that weight and bias, followed by `flatten(2).transpose(1, 2)`. Gradients must flow to `images`, `weight` and `bias`.

Do not modify the inputs.

**Example.**

```python
x = torch.arange(16.).reshape(1, 1, 4, 4)       # one 4 x 4 single-channel image
patchify(x, 2)
# tensor([[[ 0.,  1.,  4.,  5.],
#          [ 2.,  3.,  6.,  7.],
#          [ 8.,  9., 12., 13.],
#          [10., 11., 14., 15.]]])
```

**Restrictions.** Only `import torch`. No loops or comprehensions. You may not use convolution or unfolding functions (`conv2d`, `Conv2d`, `unfold`, `Unfold`, `fold`, `Fold`, `as_strided`, `pixel_shuffle`, `pixel_unshuffle`). `reshape`, `view`, `permute`, `transpose`, `flatten` and `@` are allowed.

**Constraints.** $B \le 64$, $C \le 4$, $H, W \le 256$, $d \le 1024$. Time limit 1 second per call on CPU.
