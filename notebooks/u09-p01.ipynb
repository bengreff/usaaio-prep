`df` has the string columns `user_id`, `email`, `city`, `age`. Return a **new** DataFrame with the same columns in the same order, the same rows in the same order, and the same index, where:

- `user_id`: surrounding whitespace removed.
- `email`: surrounding whitespace removed, then lowercased.
- `city`: whitespace removed and lowercased; then the exact value `"nyc"` becomes `"new york"`, `"la"` becomes `"los angeles"` and `""` becomes `"unknown"`; then title case (`str.title()`). No other value is changed.
- `age`: converted with `pd.to_numeric(..., errors="coerce")`, so it has dtype float64 and a value that is not a number becomes NaN.

Do not modify `df`.

**Example.**
