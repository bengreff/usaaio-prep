- `model.features` is some `nn.Module` (a convolutional stack, an MLP, anything). Freeze it: every one of its parameters must end with `requires_grad=False`. Its values must not change.
- `model.classifier` is an `nn.Linear`. Replace it with a **new** `nn.Linear` with the same `in_features`, `out_features = num_classes`, and a bias exactly when the old head had one. The new head must be trainable.
- Change `model` in place and return it (the same object) together with `params`, a Python list of exactly the parameters that should be trained, ready for `torch.optim.SGD(params, lr=...)`.
