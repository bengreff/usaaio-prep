```python
def perceptron(X, y, max_epochs):
    # Your code
    return w, b, mistakes
```

**Procedure.** Start from `w = np.zeros(d)` and `b = 0.0`. One epoch visits the examples in index order $i = 0, 1, \dots, n - 1$. For each example, if $y_i(w \cdot x_i + b) \le 0$ (a point exactly on the boundary counts as a mistake), update immediately

$$w \leftarrow w + y_i x_i, \qquad b \leftarrow b + y_i,$$

before looking at the next example; otherwise change nothing. After each epoch, append the number of updates made in that epoch to the list `mistakes`. Stop after the first epoch with 0 updates, or after `max_epochs` epochs, whichever comes first.

**Input.** `X` of shape `(n, d)` (ints or floats), `y` of shape `(n,)` with entries $-1$ and $+1$, `max_epochs` a positive int.

**Output.** `w` a float array of shape `(d,)`, `b` a float, `mistakes` a Python list of ints with one entry per epoch run. Do not modify `X` or `y`.

**Example.**
```python
X = np.array([[1.0, 2.0], [2.0, 0.0], [0.0, 0.0], [3.0, 1.0]]); y = np.array([1, -1, -1, 1])
perceptron(X, y, 20)                 # (array([0., 3.]), -1.0, [4, 1, 0])
Xor = np.array([[0, 0], [1, 1], [0, 1], [1, 0]]); yx = np.array([-1, -1, 1, 1])
perceptron(Xor, yx, 5)[2]            # [3, 4, 4, 4, 4]: XOR is not linearly separable
```

**Restrictions.** Only `import numpy as np`. No `np.linalg`, no scikit-learn.

**Constraints.** $1 \le n \le 2000$, $1 \le d \le 20$, $1 \le$ `max_epochs` $\le 50$, $|X_{i,j}| \le 100$. Time limit 2 s per call, CPU.
