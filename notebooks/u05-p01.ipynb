# Monte Carlo: how often is a majority vote right?

An ensemble of $m$ classifiers (with $m$ odd) votes on each example. Classifier $j$ is right with probability $p_j$, independently of the others, and the ensemble is right when more than half of the classifiers are right. Estimate the probability that the ensemble is right by simulation.

Implement

```python
def majority_vote_mc(p, n, seed):
```

- `p`: NumPy float array of shape `(m,)`, each entry in $[0, 1]$, $m$ odd.
- `n`: number of simulated examples (a positive integer). `seed`: an integer.

Simulate **exactly** like this, so that your numbers match the grader's:

1. Create the generator with `rng = np.random.default_rng(seed)`.
2. Make one call `U = rng.random((n, m))`. Classifier $j$ is right on simulated example $i$ exactly when `U[i, j] < p[j]`.
3. The ensemble is right on example $i$ when the number of classifiers right on it is greater than $m/2$.

Return `estimate, standard_error` as two Python floats: the fraction $\hat{q}$ of the $n$ simulated examples on which the ensemble is right, and $\sqrt{\hat q(1 - \hat q)/n}$.

**Example.** `majority_vote_mc(np.array([0.7, 0.7, 0.7]), 100000, 1)` returns `(0.78316, 0.0013031...)`. The exact probability is $0.7^3 + 3(0.7^2)(0.3) = 0.784$, about 0.65 standard errors away.

**Restrictions.** Only `import numpy as np`. No loops of any kind (`for`, `while`, comprehensions). Do not use the legacy global generator (`np.random.rand` and `np.random.random_sample` are banned by name; `np.random.seed` does not affect `default_rng` anyway), and do not use `binomial`. Draw random numbers only through the one `rng.random` call above.

**Constraints.** $1 \le m \le 9$, $1 \le n \le 1{,}000{,}000$. Time limit 2 seconds on CPU. Results are compared with a relative tolerance of $10^{-9}$.
