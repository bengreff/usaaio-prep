# Maximum likelihood for a mixture weight by Newton's method

Fit the mixing weight of a two-component Gaussian mixture by maximum likelihood, using Newton's method on the log-likelihood.

Implement

```python
def mixture_weight_newton(x, mu0, mu1, sigma, pi0, steps):
```

**Model.** Each value comes from component 1, $N(\mu_1, \sigma^2)$, with probability $\pi$, and from component 0, $N(\mu_0, \sigma^2)$, otherwise. The means and the standard deviation $\sigma$ are known; only $\pi$ is unknown. With $f_k(x) = \frac{1}{\sigma\sqrt{2\pi}}\exp\!\big(-\frac{(x - \mu_k)^2}{2\sigma^2}\big)$, the log-likelihood of the data $x_1, \dots, x_n$ is

$$\ell(\pi) = \sum_{i=1}^{n} \ln\big(\pi f_1(x_i) + (1 - \pi) f_0(x_i)\big).$$

**Task.** Starting from $\pi = $ `pi0`, do exactly `steps` Newton steps
$$\pi \leftarrow \pi - \frac{\ell'(\pi)}{\ell''(\pi)},$$
and after **each** step clip $\pi$ into $[10^{-6},\ 1 - 10^{-6}]$. Derive $\ell'$ and $\ell''$ yourself. Do not stop early.

**Inputs.** `x`: 1D float NumPy array of length $n$. `mu0`, `mu1`, `sigma` (> 0), `pi0` (in $(0, 1)$): Python floats. `steps`: Python int $\ge 0$.

**Outputs.** Return exactly `return pi, loglik`: the final $\pi$ (a float) and $\ell(\pi)$ at that final $\pi$ (a float, natural log, including the constant $\frac{1}{\sigma\sqrt{2\pi}}$). Do not modify `x`.

**Example.** `x = np.array([-0.5, 0.2, 2.8, 3.1, 0.0, 1.4])`, `mu0 = 0.0`, `mu1 = 3.0`, `sigma = 1.0`, `pi0 = 0.5`:

| steps | pi | loglik |
|---|---|---|
| 0 | 0.5 | -10.206360 |
| 1 | 0.380954 | -10.070104 |
| 10 | 0.381574 | -10.070100 |

**Restrictions.** Only `import numpy as np` (no SciPy). Loops are allowed over the Newton steps; a Python loop over the data will be too slow.

**Constraints.** $1 \le n \le 200{,}000$, $0 \le$ `steps` $\le 30$, $|x_i| \le 10$, $|\mu_k| \le 5$, $0.5 \le \sigma \le 3$. Results must match to a relative error of $10^{-7}$. Time limit 1 second on CPU.
