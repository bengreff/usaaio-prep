
In `__init__`, build the table `pe` of shape `(max_len, d_model)` (float32, `d_model` even) with, for position $p = 0, \dots, \text{max\_len} - 1$ and pair index $i = 0, \dots, d_{\text{model}}/2 - 1$,

$$\mathrm{pe}[p, 2i] = \sin\left(p \cdot \text{base}^{-2i/d_{\text{model}}}\right), \qquad \mathrm{pe}[p, 2i+1] = \cos\left(p \cdot \text{base}^{-2i/d_{\text{model}}}\right),$$

and store it with `self.register_buffer("pe", pe)`: it must be saved in `state_dict()` under the name `pe`, and the module must have no parameters.

`forward(x)` takes `x` of shape `(B, T, d_model)` with `T <= max_len` and returns `x + pe[:T]` (same shape). Do not modify `x`.

**Example.**

```python
m = SinusoidalPositionalEncoding(max_len=4, d_model=4)
m.pe[1]                              # tensor([0.8415, 0.5403, 0.0100, 0.9999])
m(torch.zeros(2, 3, 4)).shape        # torch.Size([2, 3, 4])
list(m.parameters())                 # []
```

**Restrictions.** Only `import torch`, `import torch.nn as nn` and `import math`.

**Constraints.** `max_len <= 10000`, `d_model <= 1024`. Building the module must take under 1 second.
