# Fibonacci-type sequences by spectral decomposition, no loops

2025 Round 1 Problem 1 asked for a class that computes Fibonacci-type numbers from the spectral decomposition of a $2 \times 2$ matrix, with no loops. This is that computation as one function.

The sequence is $F_n = F_{n-1} + F_{n-2}$ for $n \ge 2$, with given starting values $F_0$ and $F_1$. With $A = \begin{pmatrix} 1 & 1 \\ 1 & 0 \end{pmatrix}$,

$$\begin{pmatrix} F_{n+1} \\ F_n \end{pmatrix} = A^n \begin{pmatrix} F_1 \\ F_0 \end{pmatrix}, \qquad A = Q \Lambda Q^\top,$$

where $Q$ is orthogonal and $\Lambda$ is diagonal ($A$ is symmetric).

Implement

```python
def fib_values(f0, f1, indices):
```

`f0` and `f1` are Python ints. `indices` is a 1D NumPy integer array of indices $n \ge 0$ (any order, repeats allowed). Return a NumPy array of dtype `int64` and the same shape as `indices`, whose entry $i$ is $F_{\text{indices}[i]}$. Use the spectral decomposition; the results must be exact integers, so round the floating-point values to the nearest integer before converting. Do not modify `indices`.

**Example.** `fib_values(3, 1, np.array([0, 1, 2, 3, 4, 5, 10]))` returns `array([3, 1, 4, 5, 9, 14, 157])`.

**Restrictions.** Only `import numpy as np`. `np.linalg.eigh` and `np.linalg.eig` are allowed; `np.linalg.matrix_power` is banned. No Python loops or comprehensions.

**Constraints.** $|F_0|, |F_1| \le 100$, $0 \le n \le 50$, up to $10^5$ indices. Time limit 1 second on CPU.
