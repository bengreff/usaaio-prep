# k-means++ initialization

k-means++ picks starting centers for k-means that are spread out: each new center is a data point drawn with probability proportional to its squared distance from the nearest center chosen so far. Implement it with a random generator you are given, using the generator in exactly the way described, so that the grader can reproduce your choices.

```python
def kmeanspp_init(X, k, rng):
```

- `X`: float array `(n, d)`; `k`: a Python `int`, $1 \le k \le n$; `rng`: a `numpy.random.Generator` (from `np.random.default_rng(seed)`).

Procedure:

1. Choose the first center index with `rng.integers(n)`.
2. Repeat until there are $k$ centers:
    1. For every point $x_i$, let $D_i$ be the squared Euclidean distance from $x_i$ to the **nearest** center chosen so far (points already chosen have $D_i = 0$).
    2. Draw the next index with `rng.choice(n, p=D / D.sum())`, where `D` is the array of all $D_i$.

Use the generator for nothing else: exactly one call to `rng.integers` and then $k - 1$ calls to `rng.choice`, in this order. Return a float array of shape `(k, d)` whose row $j$ is the $j$-th chosen point (row $0$ the first). Do not modify `X`.

(scikit-learn's `KMeans` uses a greedy variant that draws several candidates per step, so its centers differ from these.)

**Example.**

```python
X = np.array([[0., 0], [1, 0], [10, 0], [11, 0], [5, 8]])
kmeanspp_init(X, 3, np.random.default_rng(0))
# array([[ 5., 8.], [ 1., 0.], [10., 0.]])
```

With seed 0, `rng.integers(5)` gives 4, so the first center is $(5, 8)$. Squared distances to it are $89, 80, 89, 100, 0$ (total 358), and the draw picks index 1, the point $(1, 0)$. Distances to the nearest of the two centers are then $1, 0, 81, 100, 0$, and the draw picks index 2, $(10, 0)$.

**Restrictions.** Only `import numpy as np`. Loops over the $k$ centers are fine.

**Constraints.** $1 \le n \le 20{,}000$, $1 \le d \le 10$, $1 \le k \le 20$; `X` has at least $k$ distinct rows (so `D.sum()` is never 0). Time limit 1 second per call on CPU.
