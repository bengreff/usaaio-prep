- `P`: float array `(S, A, S)` of transition probabilities; `R`: float array `(S, A)` of expected rewards; terminal states are absorbing with reward 0.
- `gamma`, `theta`, `max_sweeps`: as in u20-p02.

Start from $V_0 = 0$ and apply synchronous Bellman optimality sweeps:

$$V_k(s) = \max_a \Big(R(s, a) + \gamma \sum_{s'} P(s, a, s')\, V_{k-1}(s')\Big).$$

Stop after the first sweep with $\max_s |V_k(s) - V_{k-1}(s)| < \theta$, or after sweep `max_sweeps`. Return

- `V`: the float array $V_k$ of shape `(S,)`;
- `policy`: an integer array of shape `(S,)`, a greedy action for every state with respect to the returned `V`: `policy[s]` maximizes $R(s, a) + \gamma \sum_{s'} P(s, a, s') V(s')$ over $a$. If several actions tie, any of them is accepted.

Do not modify the inputs.

**Example.** Three states and two actions. In state 0, action 0 ("cash") pays 1 and stays; action 1 ("invest") pays 0 and moves to state 1. In state 1, cash pays 2 and returns to state 0; invest pays 0 and moves to state 2. State 2 pays 3 for either action and then returns to state 0 or stays in 2, each with probability 0.5.

```python
P = np.zeros((3, 2, 3))
P[0, 0, 0] = 1.0
P[0, 1, 1] = 1.0
P[1, 0, 0] = 1.0
P[1, 1, 2] = 1.0
P[2, :, 2], P[2, :, 0] = 0.5, 0.5
R = np.array([[1.0, 0.0], [2.0, 0.0], [3.0, 3.0]])
```

With `gamma = 0.9`, `theta = 1e-6`, `max_sweeps = 1000`: `V` is close to `[13.0997, 14.5552, 16.1725]` and the policy is `[1, 1, 0]` or `[1, 1, 1]` (invest, invest, and a tie in state 2). With `gamma = 0.5` the far future counts less: `V` is close to `[2.0, 3.0, 4.6667]` and the policy cashes in, `[0, 0, 0]` or `[0, 0, 1]`.

**Restrictions.** Only `numpy`. No `np.linalg`.

**Constraints.** $S \le 300$, $A \le 5$. Time limit 1 second for $S = 300$, $A = 5$, $\gamma = 0.95$, $\theta = 10^{-8}$.
