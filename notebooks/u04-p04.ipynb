# Backpropagation by hand: a two-layer ReLU network

Compute the loss and all four parameter gradients of a two-layer ReLU network by hand, with NumPy only (2025 Round 1 Problem 2 asked for the gradients of an affine layer; this goes one layer further).

Implement

```python
def mlp_loss_and_grads(X, y, W1, b1, w2, b2):
```

**Inputs** (all float NumPy arrays except `b2`):

- `X`, shape `(n, d)`: one example per row. `y`, shape `(n,)`: targets.
- `W1`, shape `(h, d)` and `b1`, shape `(h,)`: the hidden layer (PyTorch's `nn.Linear` layout: one row per hidden unit).
- `w2`, shape `(h,)` and `b2`, a Python float: the output layer.

**Model.** For each example (row $x_i$ of `X`):
$$z_i = W_1 x_i + b_1, \qquad a_i = \operatorname{ReLU}(z_i), \qquad \hat{y}_i = w_2^\top a_i + b_2,$$
and the loss is the mean squared error
$$L = \frac{1}{n}\sum_{i=1}^{n} (\hat{y}_i - y_i)^2.$$
For the ReLU derivative use $\operatorname{ReLU}'(z) = 1$ if $z > 0$ and $0$ if $z \le 0$ (the convention PyTorch uses at exactly 0).

**Outputs.** Return exactly `return loss, dW1, db1, dw2, db2`, where `loss` is a float and each gradient has the shape of its parameter: `dW1` shape `(h, d)`, `db1` shape `(h,)`, `dw2` shape `(h,)`, `db2` a float. Do not modify the inputs.

**Example.**

```python
X = np.array([[1.0, 2.0], [0.0, -1.0]])
y = np.array([1.0, 0.0])
W1 = np.array([[1.0, -1.0], [0.5, 1.0]])
b1 = np.array([0.0, -1.0])
w2 = np.array([2.0, -1.0])
b2 = 0.5
```

gives `loss = 5.125`, `dW1 = [[0, -5], [2, 4]]`, `db1 = [5, 2]`, `dw2 = [2.5, -3]` and `db2 = 0.5`. (Row 1 has $z = (-1, 1.5)$, so the first hidden unit is off for it; row 2 has $z = (1, -2)$.)

**Restrictions.** Only `import numpy as np` (no PyTorch, no autograd). No Python loops or comprehensions. Nothing from `np.linalg`. The `@` operator and `.T` are allowed.

**Constraints.** $1 \le n \le 20000$, $1 \le d \le 100$, $1 \le h \le 200$. Time limit 1 second on CPU.
