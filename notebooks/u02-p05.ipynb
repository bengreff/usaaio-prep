Nearest-neighbor classifiers and k-means need the distance from every query point to every reference point.

`A` is a float array of shape `(n, d)` and `B` a float array of shape `(m, d)`; each row is a point. Return `D` of shape `(n, m)` with

$$D_{ij} = \sum_{k=0}^{d-1} \left(A_{ik} - B_{jk}\right)^2,$$

the squared Euclidean distance from row $i$ of `A` to row $j$ of `B`. Every entry of `D` must be $\ge 0$ (a squared distance cannot be negative, even after rounding). Do not modify `A` or `B`.

**Memory.** Do not build any array with $n \cdot m \cdot d$ entries: at the largest sizes that would need 16 GB. The grader measures your function's peak memory for $n = m = 600$, $d = 100$ and allows at most 64 MB. (An `(n, m, d)` float array at that size is 288 MB.)

**Example.** `A = np.array([[0., 0.], [1., 1.], [3., 0.]])`, `B = np.array([[0., 1.], [3., 1.]])` gives

```
[[ 1., 10.],
 [ 1.,  4.],
 [10.,  1.]]
```

**Restrictions.** Only `import numpy as np`. Nothing from `np.linalg`. No loops of any kind (no `for`, `while` or comprehensions). The `@` operator is allowed.

**Constraints.** $1 \le n, m \le 2000$, $1 \le d \le 500$, entries of absolute value at most $10^4$. Time limit 1 second on CPU. Answers are compared with a relative tolerance of $10^{-6}$ and an absolute tolerance of $10^{-6}$.
