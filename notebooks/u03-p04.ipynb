# Top-k eigenpairs by power iteration and deflation, no np.linalg

Find the $k$ dominant eigenvalues and eigenvectors of a symmetric matrix using only matrix-vector products: power iteration, then deflation.

Implement

```python
def top_eigenpairs(A, k):
```

**Inputs.** `A` is a symmetric NumPy float array of shape `(n, n)`. `k` is a Python int with $1 \le k \le n$.

Order the eigenvalues of `A` by absolute value, $|\lambda_1| > |\lambda_2| > \dots$ (they may be negative).

**Outputs.**

1. `vals`, a float NumPy array of shape `(k,)`: $\lambda_1, \dots, \lambda_k$ with their signs, in that order.
2. `vecs`, a float NumPy array of shape `(n, k)`: column $j$ is a unit-length eigenvector for `vals[j]`. Either sign of each column is accepted. The columns must be orthonormal.

Return exactly `return vals, vecs`. Do not modify `A`.

**Method (suggested).** Power iteration: start from a vector $x$, repeat $x \leftarrow Bx / \lVert Bx \rVert$, and estimate the eigenvalue with the Rayleigh quotient $x^\top B x$ (for unit $x$). After finding $(\lambda, v)$, **deflate**: $B \leftarrow B - \lambda v v^\top$ removes that pair and leaves the other eigenpairs of a symmetric matrix unchanged. Repeat $k$ times, starting from $B = A$.

**Example.** For

```python
A = np.array([[2.0, 1, 0], [1, 2, 0], [0, 0, -5]])
```

the eigenvalues are $-5, 3, 1$. With `k = 2`: `vals = [-5, 3]` and `vecs` has columns $\pm(0, 0, 1)$ and $\pm(0.707107, 0.707107, 0)$.

**Restrictions.** Only `import numpy as np`. Nothing from `np.linalg` (no `eig`, `eigh`, `svd`, `norm`, `qr`, `solve`). Loops are allowed. Compute lengths yourself, for example `np.sqrt(x @ x)`.

**Constraints.** $1 \le n \le 200$, $1 \le k \le \min(n, 10)$. The tests guarantee a gap: $|\lambda_{j+1}| \le 0.8\,|\lambda_j|$ for $j = 1, \dots, k$ (with $\lambda_{n+1} = 0$). Eigenvalues must be correct to a relative error of $10^{-6}$. Hidden tests include matrices with special structure, so your method must work for every valid input. Time limit 2 seconds on CPU per call.
