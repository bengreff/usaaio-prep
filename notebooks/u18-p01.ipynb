- `x`, `x_hat`: float tensors of the same shape `(B, ...)`: a batch of `B` examples, each with any number of further axes (vectors `(B, D)` or images `(B, C, H, W)`).
- `mu`, `logvar`: float tensors of shape `(B, k)`, the encoder's mean and log-variance $\log\sigma^2$.

For example $i$, with all of its data coordinates $j$ and latent coordinates $l$:

$$\mathcal{L}_i = \sum_{j} (x_{ij} - \hat{x}_{ij})^2 + \frac12 \sum_{l=1}^{k}\Big(\mu_{il}^2 + e^{\text{logvar}_{il}} - \text{logvar}_{il} - 1\Big).$$

Return the batch mean $\frac{1}{B}\sum_i \mathcal{L}_i$ as a 0-dimensional tensor. It must stay differentiable with respect to `x_hat`, `mu` and `logvar`. Do not modify the inputs.

**Constraints.** $1 \le B \le 512$, up to 4 axes in `x`, $1 \le k \le 64$, float32 or float64. Time limit 1 second.
