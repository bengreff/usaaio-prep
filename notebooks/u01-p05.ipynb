# LRU cache with exact eviction order

A **cache** stores recent results so they can be returned again without recomputing them. It has room for at most `capacity` keys. When it is full and a new key arrives, it must throw one stored key out (**evict** it). A **least recently used (LRU)** cache evicts the key whose last use is furthest in the past.

Implement the class

```python

```

A key is **used** when `get` finds it or when `put` stores it. Nothing else counts as a use.

- `__init__(capacity)`: `capacity` is an int. If `capacity < 1`, raise `ValueError`. Set the attributes `capacity` (the given value), `hits = 0` and `misses = 0`.
- `get(key, default=None)`: if `key` is stored, add 1 to `hits`, mark `key` as the most recently used key, and return its value. Otherwise add 1 to `misses` and return `default`; nothing else changes.
- `put(key, value)`: store `value` under `key` and mark `key` as the most recently used key.
  - If `key` was already stored, replace its value. Nothing is evicted. Return `None`.
  - If `key` is new and the cache already holds `capacity` keys, first evict the least recently used key, then store the new one, and return the evicted key.
  - Otherwise return `None`.
- `key in cache` (`__contains__`): `True` if `key` is stored, else `False`. This is not a use: it changes neither the order nor `hits`/`misses`.
- `len(cache)` (`__len__`): the number of stored keys.
- `keys()`: a new list of the stored keys, from least recently used to most recently used. Changing the returned list must not change the cache.

Keys are ints, strings or tuples; values can be anything (including `None`). No method may modify its arguments.

**Example.**

```python
c = LRUCache(2)
c.put("a", 1)        # None
c.put("b", 2)        # None
c.get("a")           # 1; order is now b, a
c.put("c", 3)        # "b": b was the least recently used key
"b" in c             # False
c.get("b", -1)       # -1 (a miss)
c.put("a", 10)       # None: an update, so nothing is evicted; order is now c, a
c.keys()             # ["c", "a"]
len(c)               # 2
(c.hits, c.misses)   # (1, 1)
```

````python
#@title Grader: run this cell to grade your solution { display-mode: "form" }
# The grading code is hidden on purpose (double-click to see it). It writes the site's test harness and this
# problem's grader to a temporary folder, imports your file the way the contest grader does, checks the import
# and ban rules, runs every test (including hidden ones) and prints a table.
import ast, base64, contextlib, importlib.util, io, json, os, signal, sys, time, traceback, zlib
_P = json.loads(zlib.decompress(base64.b64decode("eNq9XAtz2ziS/is4T9WaTGRGdpK5O2U8d5mMZzZXeVWSvd0rS8VAIiRzRZFagoqteP3f7+sGQIIU5diZ3GVqLBEEGo1Gv9HQ9UGaHIzEwWZ4fLQePj0YiIOVqiSarnffVGmVKWp79f4vYiZnF0pcptWFUFdyVgn1OZ1VaZGLokxUSf0zpXWRY0C+yTKCnOabSmk0nDzFY5LO5+lsk1VbtDxBg8qrcmvhvyDwjE6RbDIV53LFU6/LYpqp1dNYb6arVGtMSL1klhWXKonT1booK5rifILmqcxzlXhPcVYUa3o9l5lWtCalq7hKV6rYVDGjNqRWuWAYB+ttdVHk0SyTWgNzvLItCRbrP8+K1TpTVykWg8YGuUiv1eyAZtfFppzxGooyXaS5zHi4LKsTNFblRt3geZ5mTKHrgwtZ5iBgtGaSjPm/P6tsrUotqgtZCUsKsShlQo1m7SL4NC+LlbDDXeuDT2E0zsf5c9tdpFrMirwCzR9t8rTSj36ij58fWah4XqfJz49MbyARiZeVSNQ8BVRBVBPzTc4brgVtTsKN8QODWyWXCt/UOG9IIcxWik9o+hSJs789f/3u1dkHkaW6oiUpA5aAaaEvikuwUi7eb3IRfE51CqzCZ+IDgatEucG06rMqtzwIS/tNptmmxMhSploJejTcKQXgabkAs5ZphQWLeVGKX1Q+ElpuxSWheyk13hb5YiBygiqSzWotLtIkQfc0X28qTdQzu5Bbks6K9bZ+IBYi+trHHOO3AlDzNbX+IF4UmZyKF+/+ooUsldDEriXRKidUDrV4fSLelcVA6EJkxUxmBDHNF1qsNiDKVBFRQaKCZCbF6ywlOkyx/gts5RwSWJQRzXQSDWlzpVhsaP8DM7MGrEwdVRelkokAU2LD0KssNouLbAtuyeYCuACLWVESoUtFRK9EOicE8ciCZddhBYaI8uHd2dmv8W/PX3x8+16c0uy04HHOIsPbEJxdzdSaOCUcjXOBf2u8Mr3AUKC4SgLwYjIQK71wfTBvXhCR88S20L9mcwPq2wCJ83Vw5cZCjXhj3A4V5ezCa51j/WmuK5nPVHA1MO+jjyrXRRl6w3laVW3KXFxFCZTj7CIIo9l6g7+8zUFoOitepnjJ052VZVF6UMySPVj5OpJalqXcAu9mHfpCrlWwKKoBmDKnv+DP0/EBOBuacnzgMNOgdbWByglo5egfRjzU4sI0RR/XiWCFAzEfH1wTwBtsuDZziWsNvaOuoKYq8MS11/8G0zWYzbJC34bZQJRVkZ0eq6MfB0Lar//m8IVkXQLlGlmpqy0WOs8KSYhRO8/ZfuEtZmHWR0u6NF/3LMcswHYP+9bmXvHy6gmwIamGeksrjAYeWRaEsGXMhf67S/vOn540qUxZE5L9g8qY18AtI9N+Z5klIohhyUV/LLnoj894qixBMSYEIbeSV/QhpzpYiCNxGYYhwcZC0y9KKJgzMSTh6xGVNp1Y0YnPMoN+EEEmywWpXUUMK67xMYoeL27C1t6rf2xkdgeubAuV6UpI5wlz+iBkigIJWVWlfT8+4O1ogPSwi8cgTZ//c7bwWYPxjy0dgNyOMPlUrSHQroy6GIMZCFtefAMDyvgar/6lbKFGvdDU2oxUxywivB30Te/dD57Q25FmyA7+1ApSWMBhFLPLFcctdBZQf7RzdtZD1+nQQW1jGutcrmHJq/9f1ZwVubqnUsaUjiuviCXZth+EMIkJS28bn4C8vwE7LqGPT3txLewIYJAotT79CE/PYzCL4EeQr4veLoCwZUKoKSKY/O6qRXhsSyAHYvrtdJd3orvpYuSC5/tWsksiO8PRbVVgJ5JmEh1Mwz5dg+GNnukdXuuJqf3Ke9uWbFrBwGg7cHbu75WDwxA856WIVxsIBbybYJ4PxAOoU0jGgwfLS/rmMDGu4wtYADHPeeY5Oai0Bvg48AIXEC+Zw2UsF5sVnHKYELHO5ExFzutkq0/yBL143kiWDNmflTSAZpyYnvDOyHjkQRch85qGpGBkLFiHNFTBlVGlhH37kq4DM4bnaDE465OauawS2SKwAS8mihz8dJ6SZ0kuc7OWH65T8VAc3zwTl0W5JK9eMvMK2j74o7W+tEQG8g2FyYlP80ArcgWB1V2p/MwnMbSr73L3eNIcQcIeavjeWAERiH3dw3oknHcTQ9FUbwGipBAMAdJK05LYD38tZ4iMcx6CRRqvm6IVduzJOUB0cTTdJNCk1Mw+ONzylAKozRriJKrLdKYIICiUwkIXcyvBwNlRVDMLUcsc/hKZb4KAERu8i8RHepGWaMZiskOaguyCmYworJIWW1VD8iYBIUJ0OY8ZjCqdtvkKKyVV/2D4KJV1R+x6T4XdRPFI+GFDw5IxsWJJ0hCc+HyHTcQ0P51aSB1VNEVMs2ya5MlALE8wW1s/MsqDTmNrJV8nBeNJhDhhMpx4zUwGRGxBQiZ2Dzl8f9NfEMtRVRTMNwj+k2oUncxv4OWDy9QzG3XVTMzMBLFjQGKTUzx/bUljx3EMicDObXHPP54SkSXC4DlYa0EOrAkyr+0+MSCSjlsFtMwXkE6VuA1rAhzsZFKsQO25BPvFdccx52w2Wsq0iDnD0Mpy/E4tkABg1iQPUlYaF2mWkNTNKDOCgDklWaCe//Xh7RvgmrPk6CoBhiyrJj0jjl4KN4/4yaY44iQtfxY/ebma9RbPkMp/clv1807ID2Hzwn9s7FWVpdO6yXygJdpUadY0F/XXv+uiSRKsZXXhD9fpIpfNML3VnfSCeyhhF6ZytqxbwMU5JQtMrkHPUiwzV7W6Q8MyrY4yhW7iOPoxOrY+nR5RnmhFiboBqxREF9b4OOW3limovBjnbo5onmbgaPcYjA+AdlEqigBtpgWeaPTgL/kypxyOLjLKqBScAcCE6Zp2LXpgPcU/n70/g+RYWkTv8BnEMaXB4jiMoLhoOLy6tSyB2TgHVSLqHIFbVVkFQwgD3AcCEwJgX+6Lc0HiB9jqf8iROHsyPOFM2Ku/Pv+fD/Evz9+8OfsVGFyPD2KbP4xjWgwW5vbTPKorNbPf4Oebb5T3A67jg5tx/urt23fxm7e/nn0AuAC8Ev1WQFbpy1/BuMp8fa63+ax+8Qo+5AvAME8flPfwK5zM5ul3lZN9Lsqzq3Xop1g+mlzM17IssUSktwqIxzYr6BuYLVVLLIeKDpCfjSDPZhbPLtRsqQNdzmiLK9m2t++NvJM3DGNFdCyWhjgrvQBhnkEtkgdocnwsNro6rHeHnSGbSiRDVZUmj6xbJsoYofNJnzdblQqOHVMJXIL4Goi2/dAPW8TnV+yHUlJO7fqH5xbtEbiFM8MW+xGpSc3DbYhc5EbPXKuIPvPiZkTf0RvBz41F0OajySSAXhEUM+Skk6QmGp1PLKJTWbeznaycyeF0aLzRDMxrn0IeYpMrte3E4nAZIbtpAlEhT6w0joIBDAimf0ARDTnd6ZTy8VjOkn03lUXE2HYCzpR7NF/JarXJ/JY5Z0egqkk3g/iXMlsGtBkd0+256dTd8LMJDbpRRe3KUseI0R3tGjCbUoG/HEjuRE75ilIL5jHCbCAkeY7BCp4n1BEFcwR8xciajQhH/bbR24tIJomdI9zt3OyN1y/S6yyl7aY5z4+OJz0Dm82LNuuEvO6esX6QmN1Oxt+g9bqrwYZir5iONvGO5TcC1UNLHmEJie9/gIpdCgJahwoe6VoU6OOAztB9xPumobzO+xL9uZOd7to7DMF4kKDdCeobTyPfBjBN9q+qv1PfjI2x6s7Jgh/J9VrlSeAyPmbQ+MCoPKPe/6NNr75Z2Ar+kuZvrRV7vlk812SDQpNb6QyICtvvtaxew2Xc4Wmjg+6JHBDzGNIDSU6iBbVX/TsvwEiJYXlqZYGgftHfCwpPWcEG3jxh2K9dHpJFAWEcqOvDgTg0QJxEkQAeYj7EvYtDSsXe9OcT+/CnlEUP+trh3gD7oT5BnMpcE21nF9ZCsCvYGAhOPyuJIPZv/Bk+M9Fodim3+sgcsiKSgV+Y1oCIKS20Is+2AzcnIpfG1ASGC0wSybpS5HRsRaYWCM0vyW0S3DOKotAzZnDKRFLQphjAFynHl2YXAs9o/omNo2eFDbrW+IIN/ykCz5b+SbT8Qo+FMMO9WQcYaGHpQ/ADHfbwDiCHf2SH88LNUR/JNrsMzNuLj40QxcXa+JPtbBt1N0LWEb27LJgiSvDFfwoHm0JWT25EYDwnO8P5cOIzd8/y70OCZtbbFs/KbWfJ3PptCzZuEjN7a6FoJjEmpi7Vhco5iK19Rx71fddvV3ZzS6Q+L1OAybYxe7IBQCDQjSmaanv0b3N1xHgmSs/KlKMKsQarQr4Xwp5Ec4fUHV9TwAZxVdRP82F9O8M0pbSKi1sjxM34XsXVNFBRHNcvEPJZZ5OAw9Ocs1Gf0zwAgZ2aU/ipnM9HkV+9BOuVXlLKhBJWpGU5N2mZDn/JHfO89lYTbQUzDPXm47QGfUtMOjHhExLlnY+Q939zzbPeUH9H7BU88aCOxaBpu1EuRbKyXHw+r11Et5S9PU8mXkhsKUUO+Kmouzx27jr4Hu2UcIiyQiY6CBiJR8Q2JBOcijggeLBZlMkInLJD+EQQLTJ+B0cNziaeil25MAHj2FbcjLm8RjePJjFjWNZG0sSJ1PCmwBb48M8baBNCZ08oWgs6ebAkgvrcYDVha0FPLCUT3511IZ3zZzXnHXem9bUEZy0CJicVh+jAdPetvOESl/KcA+m0iD5UlD55+dYRzyR5IvMR2KcPL39//ur964GN0m1XqLOYk7vuFNjTZe26JaLlydDh0o6OuQimSVZhO5O0VBA+sxkBMO26WQ5HVaU0RenQfPnx5euz9/H7s+evBjV2HTeH6p1o5a1UGFdBxZSbMZkdyk7ykYm3Iq/Qy7BGk5jDNnmqqjPfZro7nYXFE9LUAf3pDoTEmH76nF5zwDAxjL+7JJYhVUaUA4rNMNI84TcQbtjOTdi0SyslYRnRCsfEaDOzRtLArNVY41ZFseQzBXHt9mMUDSkD7FSXneUXqVWdITI5kCYl9surs+HwuEHgXsvYh7E9GGI06SSGKsQKu1HWCbrFHtVivQO6j1BOr9C84GfiKD6JDxCAPxkOh6PJNwsyFyW28zjc5Nd3mD7urNidaGItA/Mq/PrmNrRKCmWcJlNsB+tCIG78uP2eS1jcTybHB6boj6SwNhl1HWC9bltH+HXZW3jCt9gnSwaa7WXL/Wxiy0V5pgtpBlc66Jx4q/bI1pDTECBQmRvFTt9It3+WpXZzRGmlVjrg6pm8bRVYq7pzfzpAkUAjmId+wgtagg4h2SWhGTvOM5tj4v5yA71lmIK8Fa4iyN3SOgqX1HOab5S/fzNjYI0+HDGQ86ejCfQ3nwsDW5ulFqwexwcWtOvszUZvKRtMYU9fmGL1rTn0oaZhNLy5z5nYbuHDN1idP2p57OHcrlb+JpVmN+G8oRxJKjneu7UbOxq8GWx8DyPjrKyzIl+4c5muzhZBmpsqM/bmBxRAzGUp3Mlg6CsBOzkfbnRy2n9wwTXO5F731Krcy5Z8L1z8OgM+r0juaEMaWDWDE7wSXJwE+w5qB+Jxn3mz7uzEBWRlrdjubYrqAcZXPaXQMWh819u8UlYq06LIgh3EzDsCVbU4l8BVLXD1CBeg7DUpXNY8Fy7gMdotjim+oYOykTseoHDHnOzWxoLOdDsF0vsO6AZ8mD2g0+RBpzpgnNe14tDuVkurK0kV91iBfxj2mlJLrVj2JYIx2is61izVHFFaPgO/2vJHPlU21VzYS7zamusLzQvBWSSpuXiDXuEJjFxHt7YIgk7YYhLdOA60yuYDmI+1nKVVy/jTm8i9GJjHi7TS9iu5u3ye03QBJw474w3qdojBFpbiZoD/W+hUxWZ2YZFZql08eGi0LtYB3g44AAv7upzjdaP7mgnIGaqhD4Q95T9lOG2jiPdEUA/7rrJyhBAPT8Vxzzu7GFpGbxmaB5rR7azDErYN3A61ePsrW29aK2O4rcpcuurCB3K02J2lWlPv4cRSmak88NpC8fNpmyE6RGlmySn4Tkk7NbvSTS+D7/09s4MnezpZUu326hISs/NTp0fvfliCWqBt0XAV0rV4dDjSjt3llDYY0NBC6BncJbA/FID13nGprnzK+jmcpTJ+fFeabYYkcleVgrqDX7NuPdcZVwrYDuRssYyQIm0EfXwgqWTD3qviSx+6Kkq4cZAK142dTsq8u9y8aKDunZjkqjXpkNmx1ccISLvXHpwk1ZuQd8fy6qDVqmvYrRWaeeUElALvVJG3SpXr8mNKHPMD13wbfPvrpl1B212rp33jwb6it6HNdrvKLsZ4FpFCADGIPsehoZEpCGEwI1YYh/KQ3tbLbw3lCpKTvUOnhzxlZ6itGmD3/rg1iN5gvj2TcZHKY44J6soVf7bZIbs3rbHcldgL7FAHCPUwIMjv+jHkSY6IMEc9aNLajvbRxZB0eDtNhzuDWZox6NwulgFNWsNNl/ZAUhAzDDtp9TSt7Z6Yw9jlmbUdGBUcm933hgamk+2yy2d0HOgk1Gc2csfQTjSlAqajx7dXkXcVza5L/t+k8frqx3cjSzviXg58+y5Jjco1cLkJnTu+mxr3ZLFBsHW94nbArHFMl874FpXBZvBGy6U58dsv1I+94ucl0R5sM53VHiwPYL5cwjxFG7j4ZZ0T96Wxh48Tw4W/2lDcih0VvfIaCDtW3GTfQG/CcyTkvKIaYqqU5rrzmdPLLNsDIwEJJIDOPu8kAvSRWEkw0B00Bm0Bhi3paJHSVEV8d2oO98r+U0/0iRIFX0NUVzDJlOkEnrcs3JK5owL6wXgbQdXebg/61ERrU4f+ht4LNO03eW71zejb9PtTPi3BCLMDifG6Opq+b6sQhM03WRazjf66QevYsnajtVL77Ndjb694dj4N7KMFecC8as++PQ5NUEXYGo+iTQ97du822Oyw5emp3do7z2pgGAm7lGWibzEEeLCiCKiGqrvUrv3XVMeYJ5YkHd+d3NzRWmF3xovd77G9vQZq6Nj1U5p/qs/eiR27psouuB92jx9h96EHbhvQYL+HAYBfvnypl+fjbR0Og/shejlgFR0oO9+0uyU5WDE2rv7dN+Kq9nF7RPGqq5Gk8cGtNJLo0xsrlVeH1PlufoNZWj84e22DcwxcZdLnVuzgunW4PjWQ+ajMSIPFb9vg12Zl65DEdE1uL+GOdzmCnTsKeA88EtWhyfHI3jlZb6o+fuLRGyrP2jPaiB5Rha/n5Nm2T/0TqODE8SVVpzf+sQ/NRC58tfNWK2JgTbrD9xlIao9NzKGhBDBHbMLH72EmHc2XnAOyOPbf5Fxqe/+QYyGr7oy/5C6q0esBh03GM1tqzzWr6bHU0YwuBgR7rWyjg62xZWL5GlYwBNbLpgw8rZq4DG+yVJX9CvhxWwF34dRAptu9O2J+noGVsrt/9nV1AAGkXOIxmwBXMuF0BFevN4XQTuKW7RibQXgwBqL3KpzMm3twATG2kX2TmqUTUcy5R8SXzNmtCcijJAfEKRKmkknQ+IQxF2/igvK4AzoEy5lxHU3ahd2di1e5H4ss6Xwvr4Iywl+1UKUmI2Oh+X58kRFBS3vlJ2iXhfLbn8QwevK0W9zUVJoGvDoi/XKnCrcG8O+3jmfNg/GDHpyPh8NhGIZfqbVqgUvzFjauomrt3aooNzklUNBW07Z1UYFoW3Cc1y4ww5qK9flwYpLpvO79ZV+GH9D/eMKhdJc6HiSmwG2QiMMtJHycTMI7F5/xKLbLt1SYsURa7rM3yuI66+6LJeWs6T7Ysaceq4YJnwx9LqRQeYcPIRb/6mNvmLrN+T8Od1gBsCBM4gEDfQj17MOgoqKWvqBge2BOFTqBt71Baza3uT3rcUL9WwYu4UXoGX45L9aTEJs24IZV09Cp/u77vQD7qzRcGHRd8b3aEZ9VC3vLVgTXzBEUmTv9SYrH+4WCVgxez2SUMk1G31Yujbp/VlLcqfvpA6vBPUR2f0jB2RW+ZlDbmH3gbV97IFPPY8oPwZK78HfdL5oJS7EOVeOU7ZuTk5p0w8HkM6mus3famuHp5q5VAQNRX1L2M5t3uv3aud9064k7nw1yHraWF/A57RlxH5w+UnWdImLEoXxgWeurc4YwMnAe8piWMrjDtd79V1lNbnYkmiutpkapfa/Vssy1j8MNc5DuufJ6y01We5v1zpdiv8u1V/4NJTjT2qoKzceXrF7yPZzglJ69BMsGFjzOhtO3VSdGPQFK+kWd5mFUFexkujEpcRH4uja3ecjW8Wm3pzX0vmnl34wx9a0ti7k0BTP4tiC2ol8BINwGZjZXAfMD1Shl5iIfHwoYlx+UNz/GlclaCdBVf23u+nshvynRleXK4rWLQMPUpHAnfV5co4lZOTZCyIAf9smiPaRG564y9zTzyofQOk/pPXsYYd10654YtlG1vT8Ec3/t5KbYp5B2HWHDkKBEuln5prbLqCdGQwzEE/NxHA1b0c/TATc/HdKnWd2+yfjHg26b67GZi2axnyfd2YZmupNhd76Dm5v/BVbbhEs=")).decode())
_dir = os.path.join("/tmp", "usaaio_grader_" + _P["id"])
os.makedirs(_dir, exist_ok=True)
for _n, _s in _P["files"].items():
    if _n.endswith("::b64"):
        with open(os.path.join(_dir, _n[:-5]), "wb") as _f:
            _f.write(base64.b64decode(_s))
        continue
    with open(os.path.join(_dir, _n), "w", encoding="utf-8") as _f:
        _f.write(_s)
if _dir not in sys.path:
    sys.path.insert(0, _dir)
for _m in ("harness", "usaaio_grade", "grader", _P["meta"].get("module_name", "submission")):
    sys.modules.pop(_m, None)
import harness as _harness
_harness.SPEED_FACTOR = 1.0   # Colab is the reference machine: time limits apply as stated
import usaaio_grade as _G

def _usaaio_grade():
    meta = _P["meta"]
    mod = meta.get("module_name", "submission")
    path = os.path.abspath(mod + ".py")
    rows, static = [], []
    if not os.path.exists(path):
        print(f"{mod}.py not found: run the %%writefile cell above first.")
        return
    src = open(path, encoding="utf-8").read()
    static = _G.static_checks(src, meta)
    per = float(meta.get("test_timeout_s", 20))
    def alarm(*_):
        raise TimeoutError()
    use_alarm = hasattr(signal, "SIGALRM")
    if use_alarm:
        signal.signal(signal.SIGALRM, alarm)
    def limit(s):
        if use_alarm:
            signal.setitimer(signal.ITIMER_REAL, s)
    buf = io.StringIO()
    error = None
    if any(not s["ok"] and s["msg"].startswith("syntax") for s in static):
        error = "syntax error"
    else:
        try:
            with contextlib.redirect_stdout(buf):
                limit(per)
                spec = importlib.util.spec_from_file_location(mod, path)
                sub = importlib.util.module_from_spec(spec)
                sys.modules[spec.name] = sub
                spec.loader.exec_module(sub)
                limit(0)
        except TimeoutError:
            error = f"importing your file took over {per:.0f} s"
        except BaseException as e:
            limit(0)
            error = "your file failed to import: " + _G.friendly_error(e, path)
        entry = meta.get("entry")
        if not error and entry and not hasattr(sub, entry):
            error = f"your file does not define {entry}"
    if not error:
        gspec = importlib.util.spec_from_file_location("grader", os.path.join(_dir, "grader.py"))
        grader = importlib.util.module_from_spec(gspec)
        sys.modules["grader"] = grader
        gspec.loader.exec_module(grader)
        visible = set(getattr(grader, "EXAMPLES", []))
        for name, fn in [(n, f) for n, f in vars(grader).items() if n.startswith("test_") and callable(f)]:
            rec = {"name": name[5:].replace("_", " ") + (" (example)" if name in visible else ""), "passed": False, "msg": ""}
            t0 = time.perf_counter()
            try:
                with contextlib.redirect_stdout(buf):
                    limit(per)
                    fn(sub)
                    limit(0)
                rec["passed"] = True
            except TimeoutError:
                rec["msg"] = f"took longer than {per:.0f} s (infinite loop, or far too slow)"
            except _harness.Fail as e:
                limit(0)
                rec["msg"] = str(e)
            except BaseException as e:
                limit(0)
                rec["msg"] = "your code raised " + _G.friendly_error(e, path)
            rec["seconds"] = time.perf_counter() - t0
            rows.append(rec)
    print("Rule checks")
    for s in static:
        print(f"  {'PASS' if s['ok'] else 'FAIL'}  {s['msg']}")
    if error:
        print("\n" + error)
    if rows:
        w = max(len(r["name"]) for r in rows)
        print("\nTests")
        for r in rows:
            print(f"  {'PASS' if r['passed'] else 'FAIL'}  {r['name']:<{w}}  {r['seconds']:6.2f} s  {r['msg']}")
    out = buf.getvalue()
    if out.strip():
        print("\nPrinted output (last 2000 characters)\n" + out[-2000:])
    n_ok = sum(r["passed"] for r in rows)
    ok = not error and rows and n_ok == len(rows) and all(s["ok"] for s in static)
    print("\n" + (f"Accepted: all {len(rows)} tests passed and every rule is met." if ok else
                  f"Not accepted: {n_ok} of {len(rows)} tests passed" + ("" if all(s["ok"] for s in static) else ", and a rule check failed") + "."))

_usaaio_grade()
````
