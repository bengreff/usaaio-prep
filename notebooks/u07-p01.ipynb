**Input.** `X` of shape `(n, d)`, `y` of shape `(n,)`, `lam` a float ($\lambda$).

**Output.** A tuple `(w, b)`: `w` a float NumPy array of shape `(d,)`, `b` a Python or NumPy float. Do not modify `X` or `y`.

**Example.**
