# 2D convolution with stride and padding by im2col, no loops

A **convolution layer** slides small learned filters over an image. Write its forward pass for a batch, with stride and zero padding, without any Python loop.

```python
def conv2d(X, K, b, stride=1, padding=0):
```

- `X`: float array of shape `(N, C, H, W)`: $N$ images, each with $C$ channels of size $H \times W$.
- `K`: float array of shape `(F, C, kh, kw)`: $F$ filters, each covering all $C$ channels over a $kh \times kw$ window.
- `b`: float array of shape `(F,)`: one bias per filter.
- `stride` $s \ge 1$ and `padding` $p \ge 0$ are ints.

First pad every $H \times W$ image with a border of $p$ zeros on all four sides; call the result $X^{p}$, of size $(H + 2p) \times (W + 2p)$. Then return `out` of shape `(N, F, Ho, Wo)` with

$$H_o = \left\lfloor \frac{H + 2p - kh}{s} \right\rfloor + 1, \qquad W_o = \left\lfloor \frac{W + 2p - kw}{s} \right\rfloor + 1,$$

$$\text{out}[n, f, i, j] = b_f + \sum_{c=0}^{C-1} \sum_{u=0}^{kh-1} \sum_{v=0}^{kw-1} X^{p}[n, c, s i + u, s j + v] \; K[f, c, u, v].$$

This is what deep learning libraries call convolution (the kernel is **not** flipped). Windows that would run past the padded image are dropped, which is what the floor does. Do not modify any input.

**Example.**

```python
X = np.arange(16.0).reshape(1, 1, 4, 4)
K = np.array([[[[1., 0.], [0., -1.]]]])     # (1, 1, 2, 2)
b = np.array([0.5])
conv2d(X, K, b, stride=2, padding=1)
# shape (1, 1, 3, 3):
# [[[[ 0.5, -1.5,  0.5],
#    [-7.5, -4.5,  7.5],
#    [ 0.5, 13.5, 15.5]]]]
```

For instance `out[0, 0, 1, 1]` $= X^{p}[2, 2] - X^{p}[3, 3] + 0.5 = 5 - 10 + 0.5 = -4.5$.

**Restrictions.** Only `import numpy as np`. No loops of any kind (no `for`, `while` or comprehensions). Banned: anything from `np.lib` (including `stride_tricks`, `sliding_window_view` and `as_strided`), `np.convolve`, `np.correlate`. Everything else in NumPy is allowed, including `@`, `.T`, `transpose`, `reshape` and `einsum`.

**Constraints.** $1 \le N \le 16$, $1 \le C \le 8$, $1 \le F \le 32$, $1 \le kh, kw \le 7$, $1 \le s \le 3$, $0 \le p \le 3$, $H, W \le 64$, and $H_o, W_o \ge 1$. Also $N \cdot C \cdot H_o \cdot W_o \cdot kh \cdot kw \le 2 \times 10^6$. Time limit 1 second on CPU.
