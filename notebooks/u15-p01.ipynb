# Scaled dot-product attention with a mask

Implement scaled dot-product attention with an optional mask, the building block of every later problem in this unit.

```python
def attention(q, k, v, mask=None):
    # Your code
    return out, weights
```

- `q`: float tensor of shape `(..., n_q, d_k)`, the queries.
- `k`: float tensor of shape `(..., n_k, d_k)`, the keys.
- `v`: float tensor of shape `(..., n_k, d_v)`, the values. The leading axes `...` (zero or more of them, for example batch and head) are the same for `q`, `k` and `v`.
- `mask`: `None`, or a `torch.bool` tensor that broadcasts to `(..., n_q, n_k)`. `mask[..., i, j]` is `True` when query `i` may attend to key `j`. Every query has at least one allowed key.

Compute

$$\text{weights} = \operatorname{softmax}\left(\frac{q k^\top}{\sqrt{d_k}}\right) \text{ over the keys}, \qquad \text{out} = \text{weights} \cdot v,$$

where pairs with `mask` equal to `False` get weight exactly 0 and the remaining weights of each query still sum to 1. Return `out` with shape `(..., n_q, d_v)` and `weights` with shape `(..., n_q, n_k)`. Gradients must flow from `out` back to `q`, `k` and `v`. Do not modify the inputs.

**Example.**

```python
q = torch.tensor([[1., 0, 1, 0]])                                # n_q = 1, d_k = 4
k = torch.tensor([[2., 0, 0, 0], [0, 0, 2, 2], [0, 2, 0, 0]])    # n_k = 3
v = torch.tensor([[1., 0], [0, 1], [1, 1]])                      # d_v = 2
out, w = attention(q, k, v)
# w   = [[0.4223, 0.4223, 0.1554]]
# out = [[0.5777, 0.5777]]
out, w = attention(q, k, v, mask=torch.tensor([[True, False, True]]))
# w   = [[0.7311, 0.0000, 0.2689]]
```

**Restrictions.** Only `import torch` and `import math`. You may not use `scaled_dot_product_attention`, `nn.MultiheadAttention` or `multi_head_attention_forward`. Softmax, `masked_fill` and everything else in torch are allowed.

**Constraints.** Up to 64 leading slices, $n_q, n_k \le 512$, $d_k, d_v \le 128$; scores can be large (do not exponentiate by hand without subtracting the maximum). Time limit 1 second per call on CPU.
