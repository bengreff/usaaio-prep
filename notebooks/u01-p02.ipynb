# LabelEncoder: a class with fit, transform and inverse_transform

Classifiers usually need class labels as integers, while data files store them as strings such as `"cat"`. A **label encoder** learns a fixed mapping from label strings to ids $0, 1, 2, \dots$ and back.

Implement the class

```python
class LabelEncoder:
    def __init__(self):
    def fit(self, labels):
    def transform(self, labels):
    def inverse_transform(self, ids):
```

- `__init__()` sets the attribute `classes` to `None`.
- `fit(labels)`: `labels` is a list of strings, possibly with repeats. Set `self.classes` to the **sorted list of distinct labels** (Python string order). The label `self.classes[i]` gets id `i`. Return `self`. Every call to `fit` replaces everything learned by earlier calls.
- `transform(labels)`: return a list of ints, the id of each label, in order. If any label was not seen by the latest `fit`, raise `ValueError`. If `fit` has never been called, raise `RuntimeError`.
- `inverse_transform(ids)`: return the list of labels for a list of valid ids, in order. If `fit` has never been called, raise `RuntimeError`.

No method may modify its argument.

**Example.**

```python
enc = LabelEncoder()
enc.fit(["dog", "cat", "dog", "bird"])     # returns enc
enc.classes                                # ["bird", "cat", "dog"]
enc.transform(["cat", "dog", "cat"])       # [1, 2, 1]
enc.inverse_transform([2, 0])              # ["dog", "bird"]
enc.transform(["fox"])                     # raises ValueError
LabelEncoder().fit(["b", "a"]).transform(["a"])   # [0]
```

**Restrictions.** No imports.

**Constraints.** Up to 100,000 labels per call, with up to 5,000 distinct labels. Time limit 2 seconds per method call on CPU. The grader calls methods repeatedly on the same object.
