# Affine and Chain: subclasses with super() and __call__

```python
        self.name = name

        self.calls += 1

    def forward(self, x):
        raise NotImplementedError
```

Implement two subclasses of `Model` in the same file.

**1. `class Affine(Model)`** with `__init__(self, a, b)` and `forward(self, x)`.

- The constructor sets up the `Model` part with the name `"affine"`, and stores `a` and `b` in attributes named `a` and `b`.
- `x` is a list of numbers. `forward` returns a **new** list with $a v + b$ for each value $v$ of `x`, in order.

**2. `class Chain(Model)`** with `__init__(self, models)` and `forward(self, x)`.

- The constructor sets up the `Model` part with the name `"chain"`. `models` is a list of `Model` objects. Store **a copy of the list** in the attribute `models`, so that later changes to the caller's list do not change the chain.
- `forward` passes `x` through the models in order (the output of one is the input of the next) and returns the final list. Call each model as `m(x)`, so that each model's `calls` counter goes up by one per use. An empty chain returns a new list equal to `x`.

No method may modify `x`. Every object must start with `calls == 0`.

**Example.**

```python
f = Affine(2, 1)
f([1, 2])                          # [3, 5]
f.calls, f.name                    # (1, "affine")

g = Affine(1, -1)
c = Chain([f, g])
c([1, 2])                          # [2, 4]: [1, 2] -> [3, 5] -> [2, 4]
c.calls, f.calls, g.calls          # (1, 2, 1)
isinstance(c, Model)               # True
```

**Restrictions.** No imports. Do not change the class `Model`.

**Constraints.** Lists have at most 10,000 values and chains at most 50 models. Time limit 1 second per call on CPU.
