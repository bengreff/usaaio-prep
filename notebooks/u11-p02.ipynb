# Backpropagation by hand: a two-layer softmax classifier

Compute the loss and all four parameter gradients of a two-layer classifier with a tanh hidden layer and a softmax
cross-entropy loss, with NumPy only. (Problem u04-p04 did a ReLU network with squared error; this one has a matrix of
output weights, a different activation, and a loss that overflows if computed naively.)

```python
def classifier_loss_and_grads(X, y, W1, b1, W2, b2):
    # Your code
    return loss, dW1, db1, dW2, db2
```

**Inputs** (float64 NumPy arrays, except `y`):

- `X`, shape `(n, d)`: one example per row. `y`, shape `(n,)`, integer dtype: class labels in `0, ..., k - 1`.
- `W1`, shape `(h, d)`, and `b1`, shape `(h,)`: the hidden layer (`nn.Linear` layout: one row per hidden unit).
- `W2`, shape `(k, h)`, and `b2`, shape `(k,)`: the output layer.

**Model.** With rows of a matrix as examples:
$$Z_1 = XW_1^\top + b_1, \qquad H = \tanh(Z_1), \qquad Z_2 = HW_2^\top + b_2,$$
and the loss is the mean cross-entropy of the softmax of each row of $Z_2$:
$$L = -\frac{1}{n}\sum_{i=1}^{n} \ln p_{i, y_i}, \qquad p_{ij} = \frac{e^{Z_{2,ij}}}{\sum_m e^{Z_{2,im}}}.$$
This is what `nn.CrossEntropyLoss()` computes from the logits $Z_2$. Recall $\tanh'(z) = 1 - \tanh^2(z)$.

**Outputs.** Return exactly `return loss, dW1, db1, dW2, db2`: `loss` a float and each gradient a float array with the
shape of its parameter. Do not modify the inputs.

**Example.**

```python
X = np.array([[1.0, 0.0], [0.0, 2.0]])
y = np.array([0, 2])
W1 = np.array([[1.0, -1.0], [0.5, 0.0]])
b1 = np.array([0.0, 0.5])
W2 = np.array([[1.0, 0.0], [0.0, 1.0], [-1.0, 1.0]])
b2 = np.zeros(3)
loss, dW1, db1, dW2, db2 = classifier_loss_and_grads(X, y, W1, b1, W2, b2)
round(loss, 6)          # 0.645125
np.round(db2, 6)        # [-0.26622   0.332133 -0.065913]
np.round(dW1, 6)        # [[-0.164611  0.02708 ]
                        #  [ 0.124866 -0.048915]]
```

**Restrictions.** Only `import numpy as np`. No PyTorch or other autograd. No Python loops or comprehensions. Nothing from
`np.linalg`. The `@` operator and `.T` are allowed.

**Constraints.** $1 \le n \le 20000$, $1 \le d \le 100$, $1 \le h \le 200$, $2 \le k \le 20$. Some hidden tests have
logits in the hundreds, so compute the softmax in a numerically stable way. Time limit 1 second, CPU.
