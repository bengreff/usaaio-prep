# Rank, band and flag invoices

Implement

```python
def invoice_flags(lines):
```

`lines` has one row per invoice line: `invoice_id` (string of digits) and `amount` (float). Compute each invoice's `total` (the sum of its amounts) and return a DataFrame with one row per invoice and exactly these columns, in this order:

| column | meaning |
|---|---|
| `invoice_id` | the invoice |
| `total` | sum of its amounts (float) |
| `rank` | rank of `total`, largest = 1; tied totals share the smallest rank (1, 1, 3); int64 |
| `size_band` | `"Small"` if total < 50, `"Medium"` if 50 <= total <= 200, `"Large"` if total > 200 |
| `is_outlier` | 1 if total > $Q_3 + 1.5\,(Q_3 - Q_1)$, else 0 (int64). $Q_1$, $Q_3$ are the 0.25 and 0.75 quantiles of the **invoice totals**, with linear interpolation (the pandas default). |

Sort by `invoice_id` **as strings** and reset the index to 0, 1, 2, .... Do not modify `lines`.

**Example.**

```python
lines = pd.DataFrame({"invoice_id": ["101", "101", "99", "1000", "102", "99", "103", "104", "105"],
                      "amount": [30.0, 20.0, 75.0, 200.0, 12.5, 125.0, 40.0, 980.0, 49.5]})
invoice_flags(lines)
#   invoice_id  total  rank size_band  is_outlier
# 0       1000  200.0     2    Medium           0
# 1        101   50.0     4    Medium           0
# 2        102   12.5     7     Small           0
# 3        103   40.0     6     Small           0
# 4        104  980.0     1     Large           1
# 5        105   49.5     5     Small           0
# 6         99  200.0     2    Medium           0
```

The totals have $Q_1 = 44.75$ and $Q_3 = 200$, so the fence is $200 + 1.5 \times 155.25 = 432.875$ and only 980 is flagged. Invoices 1000 and 99 tie at 200 and share rank 2. `"1000"` sorts before `"101"` and `"99"` sorts last because they are compared as strings.

**Constraints.** Up to about 40,000 rows. Time limit 2 seconds on CPU.
