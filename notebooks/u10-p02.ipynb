**MLP.** `sizes` is a Python list of $k + 1$ positive ints $[d_0, d_1, \dots, d_k]$ with $k \ge 1$.

- Create $k$ layers `nn.Linear(sizes[i], sizes[i + 1])` (with bias) for $i = 0, \dots, k - 1$, and store them, in order, in an `nn.ModuleList` named `self.layers`. The model must have no other parameters.
- `forward(x)`: `x` is a float32 tensor of shape `(N, d_0)`. Apply the layers in order with ReLU after every layer **except the last**, and return the result, shape `(N, d_k)`. The output can be negative.

**count_parameters.** For any `nn.Module` `model` (not only your `MLP`), return the number of scalar learnable parameters as a Python `int`. With `trainable_only=True`, count only parameters with `requires_grad=True`. Buffers (such as batch norm's running mean) are not parameters. A layer stored under two names counts once.

**Example.**

```python
model = MLP([4, 8, 3])
model.layers                          # ModuleList of Linear(4, 8) and Linear(8, 3)
model(torch.randn(5, 4)).shape        # torch.Size([5, 3])
count_parameters(model)               # 67 = (4*8 + 8) + (8*3 + 3)
model.layers[0].requires_grad_(False)
count_parameters(model, trainable_only=True)   # 27
```

**Restrictions.** Only `import torch` and `import torch.nn as nn`.

**Constraints.** $1 \le k \le 10$, $1 \le d_i \le 512$, $N \le 1000$. Time limit 1 s per call, CPU.
