# k-means from given centers

**k-means** groups unlabeled points into $k$ clusters. Starting from given centers, it alternates two steps: assign every point to its nearest center, then move every center to the mean of the points assigned to it.

Implement

```python
def kmeans(X, init_centers, n_iters):
```

- `X`: float array of shape `(n, d)`, one point per row.
- `init_centers`: float array of shape `(k, d)`, the starting centers.
- `n_iters`: a Python `int`, $n_\text{iters} \ge 0$.

Run exactly this procedure, with `centers` starting as a copy of `init_centers`:

1. Repeat `n_iters` times:
    1. **Assign.** For every point $i$, `labels[i]` is the index $c$ of the center with the smallest squared Euclidean distance $\sum_j (X_{ij} - \text{centers}_{cj})^2$. On a tie, take the smallest index $c$.
    2. **Update.** For every cluster $c$ that received at least one point, set `centers[c]` to the mean of the points with `labels == c`. A cluster with no points keeps its center unchanged.
2. After the loop, assign every point once more (same rule) using the final centers.

Return exactly `return centers, labels`: `centers` a float array of shape `(k, d)` and `labels` an integer array of shape `(n,)`, the labels from step 2. With `n_iters = 0`, `centers` equals `init_centers` and `labels` assigns each point to its nearest initial center. Do not modify `X` or `init_centers`.

**Example.**

```python
X = np.array([[0., 0.], [3., 0.], [0., 3.], [6., 6.], [8., 4.]])
C0 = np.array([[1., 1.], [5., 5.]])
centers, labels = kmeans(X, C0, 1)
# centers = [[1., 1.], [7., 5.]],  labels = [0, 0, 0, 1, 1]
```

**Restrictions.** Only `import numpy as np`. Nothing from `np.linalg`. Loops are allowed (you will need one over the iterations), but a loop over the points will not meet the time limit.

**Constraints.** $1 \le n \le 20{,}000$, $1 \le d \le 8$, $1 \le k \le 8$, $0 \le n_\text{iters} \le 30$, $|X| \le 100$. Time limit 1 second per call on CPU.
